### 简单解答：
搞测量差异的负优势token会比相同条件的正优势token，引入更大的不稳定性，而低策略差异的负优势token可以增加探索和防止过拟合，因此前者需要提出，后者需要保留。

在off-policy场景下：
- 高策略差异、负梯度样本会剧烈太高许多不相关的token的概率，引入不稳定性
- 低策略差异、负梯度样本则起到了正则化的作用，扩展了采样空间。

可以通过token-level的梯度如何通过logit进行传播来证明以上结论：


- 若advantage为正，则会提高采样token的logit，降低其他token的logit
- 若advantage为负，降低采样token的logit，单同时会抬高其他所有token的logit

在大语言模型强化学习中，动作空间是一个大的词汇表（Qwen3词表大小为151936），而在给定状态（state）下的期望动作数量却很少。

- 而在off-policy的高策略差异的场景下，负梯度会剧烈太高许多不相关的token的概率，引入不稳定性。

- 而在低策略差异的场景下，这个特性起到了正则化的作用，扩展了采样空间

因此DeepSeek V3.2改进版GRPO要提出高策略差异的负优势样本。
