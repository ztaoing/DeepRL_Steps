### 为什么归一化？如果不归一化
- 简单任务的梯度信号过强，主导优化方向
- 复杂任务的梯度信号过弱，几乎被忽略
- 训练不稳定，模型偏向特定类型的prompt

## GRPO的全局归一化机制

全局均值
global_mean = mean(advantages_over_all_batch)
全局标准差
global_std = std(advantages_over_all_batch) + epsilon

对每个样本进行归一化
normalized_advantage = (advantage - global_mean)/global_std

```python
import torch
import torch.nn.functional as F

def global_advantage_normalization(rewards:torch.tensor,eps:float = 1e-8):
    """
    GRPO风格的全局advantage归一化
    """
    
    global_mean = rewards.mean()
    global_std = rewards.std() + eps
    
    #全局归一化
    normalized_advantages = (rewards - global_mean)/global_std
    return normalized_advantages

rewards = torch.tensor([1.0,0.9,0.1,0.0,0.95,0.05])
normalized = global_advantage_normalization(rewards)
print(f"原始rewards:{rewards}")
print(f"归一化后：{normalized}")
print(f"均值：{normalized.mean():.4f},标准差{normalized.std():.4f}")
```
