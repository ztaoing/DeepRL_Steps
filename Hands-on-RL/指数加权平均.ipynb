```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm

my_font = fm.FontProperties(fname="/System/Library/Fonts/STHeiti Light.ttc")
plt.rcParams['axes.unicode_minus'] = False 

# 假设这是一段时间的每日顾客平均甜度喜好（满分10分）
theta = np.array([8, 9, 7, 8.5, 9.5, 10, 6, 7, 7.5, 8, 8.5, 9, 9.2, 8.8])
days = np.arange(len(theta))

# 设置beta值，我们选一个比较“稳重”的调配师
beta = 0.9

# 初始化
v = 0
v_standard = []
v_corrected = []

print("天数 (t) | 原始甜度 (θ_t) | 标准EWA (v_t) | 偏差修正后EWA")
print("-" * 65)

for t in range(len(theta)):
    # 调配师更新他的记忆
    v = beta * v + (1 - beta) * theta[t]
    v_standard.append(v)
    
    # 给他来一杯“清醒咖啡”
    v_corr = v / (1 - beta**(t + 1))
    v_corrected.append(v_corr)
    
    print(f"{t+1:^8} | {theta[t]:^15.2f} | {v:^15.4f} | {v_corr:^20.4f}")

# 可视化对比
plt.figure(figsize=(14, 7))

# 绘制原始数据
plt.plot(days, theta, 'o-', label='原始甜度数据 (θ_t)', color='gray', alpha=0.6, markerfacecolor='white')

# 绘制标准EWA曲线
plt.plot(days, v_standard, 'r-o', label=f'标准EWA (β={beta}) - 启动缓慢', linewidth=2, markersize=4)

# 绘制偏差修正后的EWA曲线
plt.plot(days, v_corrected, 'g--s', label=f'偏差修正后EWA (β={beta}) - 快速启动', linewidth=2.5, markersize=4)

# 设置图表标题和标签
plt.title('指数加权平均(EWA) - 奶茶店甜度趋势分析', fontproperties=my_font, fontsize=18, pad=20)
plt.xlabel('天数', fontproperties=my_font, fontsize=14)
plt.ylabel('甜度喜好（分）', fontproperties=my_font, fontsize=14)

# 显示图例
plt.legend(prop=my_font, fontsize=12)
plt.grid(True, linestyle='--', alpha=0.3)
plt.tight_layout()
plt.show()
```

plot() 函数可以绘制线图、散点图等，是数据可视化中最基本的工具之一。

参数说明:
plt.plot() 函数的参数非常灵活，以下是一些常用的参数：
- x, y：数据点的 x 和 y 坐标。可以是列表、数组等可迭代对象。
- fmt：格式字符串，用于指定线条的样式、颜色和标记点。例如：
   - '-b'：蓝色实线
   - '--r'：红色虚线
   - '*g'：绿色星号标记
- label：图例的标签。
- linewidth 或 lw：线条的宽度。
- linestyle 或 ls：线条的样式，如 '-'（实线）、'--'（虚线）、':'（点线）等。
- < font color='red'> marker：标记点的样式，如 'o'（圆点）、'*'（星号）、's'（正方形）等。<font>
- color 或 c：线条的颜色，可以是颜色名称（如 'red'）、十六进制颜色代码（如 '#FF5733'）或 RGB 元组（如 (1, 0, 0)）。
- alpha：透明度，取值范围为 0（完全透明）到 1（完全不透明）。

```python
import matplotlib.pyplot as plt

# x 和 y 坐标
x = [1, 2, 3, 4]
y = [1, 4, 9, 16]

# 绘制线图
plt.plot(x, y, label='y = x^2', linewidth=2, color='blue')

# 添加标题和标签
plt.title('Simple Line Plot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')

# 添加图例
plt.legend()

# 显示图形
plt.show()
```

```python
import matplotlib.pyplot as plt

# x 坐标
x = [1, 2, 3, 4]

# y 坐标
y1 = [1, 4, 9, 16]
y2 = [2, 4, 6, 8]

# 绘制两条线图
plt.plot(x, y1, label='y = x^2', linestyle='-', color='blue', linewidth=2)
plt.plot(x, y2, label='y = 2x', linestyle='--', color='red', linewidth=2)

# 添加标题和标签
plt.title('Multiple Lines Plot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')

# 添加图例
plt.legend()

# 显示图形
plt.show()
```
