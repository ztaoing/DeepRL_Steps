列表（List）、字典（Dict）和元组（Tuple）是三种最常用的数据结构。它们在数据的增、删、改操作上有着本质的区别，主要体现在<font color='red'>可变性</font>上。

简单来说：
- 列表和字典是可变的，支持增删改。
- 元组是不可变的，一旦创建就不能修改，因此不支持增删改操作。


# 📝 列表 (List)

列表是一个有序的、可变的集合，使用方括号 [] 定义。

## 添加 (Add)
- append(item): 在列表末尾添加一个元素。
- insert(index, item): 在指定索引位置插入一个元素。
- extend(iterable): 将一个可迭代对象（如另一个列表）中的所有元素添加到原列表末尾。

```python
my_list = [1,2,3]
my_list.append(4)     # [1, 2, 3, 4]
my_list.insert(1,1.5) # 在位置1添加1.5
my_list.extend([5,6]) # [1, 1.5, 2, 3, 4, 5, 6]
```

## 删除 (Delete)
- pop(index): 移除并返回指定索引的元素，默认移除最后一个。
- remove(item): 移除列表中第一个匹配的元素。
- del list[index]: 删除指定索引的元素。
- clear(): 清空整个列表。

```python
my_list = [1, 2, 3, 4, 5]
my_list.pop()     # 移除5
my_list.remove(2) # 移除2
del my_list[0]    # 移除1
my_list.clear()   # 清空列表
```

## 修改 (Modify)
- list[index] = new_value: 直接通过索引赋值来修改元素。

```python
my_list = [1, 2, 3]
my_list[1] = 20          # [1, 20, 3]
```

# 🗺️ 字典 (Dict)

字典是存储键值对（key-value pair）的<font color='red'>无序集合</font>，使用花括号 {} 定义。

## 添加 (Add)
- dict[key] = value: 如果键不存在，则会创建一个新的键值对。
- update(other_dict): 将另一个字典的键值对批量添加到当前字典中。

```python
my_dict = {'name': 'Alice'}
my_dict['age'] = 25   # {'name': 'Alice', 'age': 25}
my_dict.update({'city': 'Beijing'}) # {'name': 'Alice', 'age': 25, 'city': 'Beijing'}
```

## 删除 (Delete)
- pop(key): 移除指定键的元素，并返回其值。
- popitem(): 移除并返回最后一个插入的键值对（Python 3.7+）。
- del dict[key]: 删除指定键的元素。
- clear(): 清空整个字典。

```python
my_dict = {'name': 'Alice', 'age': 25, 'city': 'Beijing'}
my_dict.pop('age')       # 移除 'age': 25, 返回 25
print(my_dict.popitem())        # 移除 'city': 'Beijing'
del my_dict['name']      # 移除 'name': 'Alice'
my_dict.clear()          # 字典变为 {}
```

# 🔒 元组 (Tuple)

元组是一个有序的、不可变的集合，使用圆括号 () 定义。


元组不支持任何形式的添加、删除或修改元素的操作。 尝试进行这些操作会引发 TypeError 异常。

```python
my_tuple = (1, 2, 3)

# 以下操作都会报错
# my_tuple[0] = 10       # TypeError: 'tuple' object does not support item assignment
# my_tuple.append(4)     # AttributeError: 'tuple' object has no attribute 'append'
# del my_tuple[0]        # TypeError: 'tuple' object doesn't support item deletion
```

## 变通方法
- <font color='red'>虽然不能直接修改，但你可以通过组合切片来创建一个新的元组。</font>

```python
my_tuple = (1, 2, 3)
# 创建一个新元组，将第二个元素替换为 20
new_tuple = my_tuple[:1] + (20,) + my_tuple[2:] 
# new_tuple 为 (1, 20, 3)
```
