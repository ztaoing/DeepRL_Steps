## 最长无重复字符子串--滑动窗口/双指针
- left窗口左边界，初始为0
- right窗口右吧边界
- seen 记录当前窗口里有哪些字符
- ans 全局最大长度

```python
def length_of_longest_substring(s:str)->int:
    left=ans=0
    seen = set()
    for right,ch in enumerate(s):
        while ch in seen:
            seen.remove(s[left])
            left +=1 # 发现重复，窗口左边界右移动
        seen.add(ch)
        # 只保留最大的长度
        ans = max(ans,right - left + 1)
    return ans
```

##  求“要等多少天才能遇到更暖和的温度”
- 单调栈
- 维护一个“单调递减”的下标栈，遇到升温就把之前所有‘熬出头’的天数一次性结算

```python
def  daily_temperatures(t):
    '''
    当前元素 v 把栈里所有比它小的都‘唤醒’，并记录等待天数，然后自己睡进去继续单调递减。”
    '''
    n = len(t)
    res = [0] * n
    st = [] # 单调栈，存的是温度的下标，而不是温度本身
    for i,v in enumerate(t):
        while st and t[st[-1]] < v: # 只要当前温度 v 比栈顶那天的温度高，就说明栈顶那天“等到升温”了。
            prev = st.pop()
            res[prev] = i - prev
        st.append(i)
    return res
```

## 归并排序
- “在 merge 阶段，每当右半区的元素被选中时，左半区剩余元素都能与该右元素构成逆序对。”

- 先递归排好左半边 [l, m]
- 再递归排好右半边 [m+1, r]
- 最后把两段有序数组合并到辅助数组 tmp，并拷回原数组
- 归并排序合并时，右段小就一次性把左段剩余个数全算进逆序对

```python
def  reverse_pairs(nums):
    n = len(nums)
    tmp = [0] * n
    def merge_sort(l,r):
        if l >= r:
            return 0
        m = (l + r) // 2
        # 左右子问题
        ans = merge_sort(l,m) + merge_sort(m+1,r)
        # 合并 + 统计
        i,j = l,m+1
        k = l  # tmp的写入指针
        while i <=n and j <=r:
            if nums[i] <=nums[j]:
                tmp[k] = nums[i];
                i +=1
            else:
                ans += m - i +1
                tmp[k] = nums[j];
                j+=1
            k +=1
            
        # 剩余元素
        while i <=m:
            tmp[k] = nums[i];
            i+=1
        while j <=r:
            tmp[k] = nums[j]
            j +=1
        # 拷贝回原数组
        nums[l:r+1] = tmp[l:r+1]
        return ans
    return merge_sort(0,n-1)
        
```
