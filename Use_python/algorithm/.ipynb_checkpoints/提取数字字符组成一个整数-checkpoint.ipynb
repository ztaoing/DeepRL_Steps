## 提取所有数字字符并组成一个整数

```python
s = '45894sdf5'
# 使用列表推导式过滤数字字符
digits = [c for c in s if c.isdigit()]
# 合并成字符串并转换为整数
result = int(''.join(digits))
print(result)
```

## 提取连续完整的数字块


```python
import re
s = '45894sdf5'
# 使用正则表达式查找所有连续的数字
numbers = [int(x) for x in re.findall(r'\d+',s)]
print(numbers)
```

## 将每个数字单独转换为整数列表

```python
s = '45894sdf5'
result = [int(c) for c in s if c.isdigit()]
print(result)
```
