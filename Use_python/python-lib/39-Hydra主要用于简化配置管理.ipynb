Hydra 是一个非常强大的库，用于配置和管理 Python 应用程序。它由 Facebook 的研究团队开发，<b>主要用于简化配置管理</b>，特别是在复杂的机器学习和数据科学项目中。
- Hydra 允许开发者通过配置文件（通常是 YAML 格式）来管理应用程序的参数，而不是硬编码在代码中


Hydra 的主要特点
- 配置文件管理：使用 YAML 文件来定义配置，支持默认配置和覆盖配置。
- 动态配置：可以在运行时动态修改配置。
- 组合配置：可以组合多个配置文件，以实现更灵活的配置管理。
- 命令行参数支持：可以通过命令行参数覆盖配置文件中的值。
- 插件系统：支持插件扩展，可以自定义行为。

# 安装 
pip install hydra-core

```python
1. 创建配置文件
假设你有一个项目结构如下：
my_project/
    config/
        config.yaml
    main.py
```

```python
config.yaml 文件内容如下：

# config/config.yaml
db:
  driver: mysql
  user: root
  pass: password
```

2. 在代码中使用 Hydra

在 main.py 文件中，你可以这样使用 Hydra：

```python
# main.py
import hydra
from omegaconf import DictConfig

@hydra.main(config_path="config", config_name="config")
def main(cfg: DictConfig) -> None:
    print(cfg.db.driver)
    print(cfg.db.user)
    print(cfg.db.pass)

if __name__ == "__main__":
    main()
```

```python
# 运行代码 

python main.py

# 输出

mysql
root
password
```

## 3. 使用命令行参数覆盖配置
你可以通过命令行参数覆盖配置文件中的值：



```python
python main.py db.driver=postgresql db.user=admin
```

## 高级用法
- 1. 组合配置

假设你有多个配置文件，例如：

```python
my_project/
    config/
        config.yaml
        db/
            mysql.yaml
            postgresql.yaml
    main.py
```

```python
# config/db/mysql.yaml
driver: mysql
user: root
pass: password


# config/db/postgresql.yaml
driver: postgresql
user: admin
pass: secret

# config/config.yaml
defaults:
  - db: mysql
```

```python
# main.py
import hydra
from omegaconf import DictConfig

@hydra.main(config_path="config", config_name="config")
def main(cfg: DictConfig) -> None:
    print(cfg.db.driver)
    print(cfg.db.user)
    print(cfg.db.pass)

if __name__ == "__main__":
    main()
```

```python
# 可以通过以下命令切换数据库
python main.py db=postgresql
```

## 插件系统
- Hydra 支持插件系统，你可以通过安装插件来扩展其功能。例如，你可以安装 hydra-optuna-sweeper 插件来支持超参数优化：

```python
# 安装 
pip install hydra-optuna-sweeper
```

```python
# 在配置文件中使用它：

# config/sweeper.yaml
defaults:
  - _self_

hydra:
  sweeper:
    _target_: hydra_plugins.hydra_optuna_sweeper.optuna_sweeper.OptunaSweeper
    study:
      storage: sqlite:///optuna.db
      study_name: my_study
    sampler:
      _target_: optuna.samplers.TPESampler
```
