是 Python 中用于定义数据类（dataclass）的语法，它提供了更简洁、更易读的方式来创建数据结构。数据类主要用于存储数据，类似于传统的类，但更简洁且自动生成了一些方法（如 __init__、__repr__ 等）

```python
from dataclasses import dataclass, field

@dataclass
class Product:
    name: str
    price: float = field(default=0.0, metadata={"description": "The price of the product"})
    quantity: int = field(default=1, metadata={"description": "The quantity of the product"})

    def total_cost(self) -> float:
        return self.price * self.quantity

# 创建实例
product = Product(name="Apple", price=1.5, quantity=10)

# 访问属性和方法
print(product.name)  # 输出: Apple
print(product.total_cost())  # 输出: 15.0
```
