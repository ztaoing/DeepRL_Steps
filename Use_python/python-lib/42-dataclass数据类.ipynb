是 Python 中用于定义数据类（dataclass）的语法，它提供了更简洁、更易读的方式来创建数据结构。数据类主要用于存储数据，类似于传统的类，但更简洁且自动生成了一些方法（如 __init__、__repr__ 等）

```python
from dataclasses import dataclass, field

@dataclass
class Product:
    name: str
    price: float = field(default=0.0, metadata={"description": "The price of the product"})
    quantity: int = field(default=1, metadata={"description": "The quantity of the product"})

    def total_cost(self) -> float:
        return self.price * self.quantity

# 创建实例
product = Product(name="Apple", price=1.5, quantity=10)

# 访问属性和方法
print(product.name)  # 输出: Apple
print(product.total_cost())  # 输出: 15.0
```

#  field() 函数用于定义数据类中的字段，并可以指定字段的默认值、默认工厂函数、元数据等。

field(default=DEFAULT, default_factory=DEFAULT_FACTORY, init=True, repr=True, hash=None, compare=True, metadata=None)

- default：字段的默认值。
- default_factory：一个可调用对象，用于生成字段的默认值。如果指定了 default_factory，则不能同时指定 default。
- init：布尔值，表示该字段是否包含在生成的 __init__ 方法中。默认为 True。
- repr：布尔值，表示该字段是否包含在生成的 __repr__ 方法中。默认为 True。
- hash：布尔值，表示该字段是否包含在生成的 __hash__ 方法中。默认为 None，表示继承自 compare 的值。
- compare：布尔值，表示该字段是否包含在生成的比较方法（如 __eq__）中。默认为 True。
- metadata：一个映射对象，用于存储与字段相关的元数据。

## 使用默认值

```python
from dataclasses import dataclass, field

@dataclass
class Person:
    name: str
    age: int = field(default=30)  # 默认值为 30
    hobbies: list = field(default_factory=list)  # 默认值为一个空列表

# 创建实例
person1 = Person(name="Alice")
person2 = Person(name="Bob", age=25, hobbies=["reading", "coding"])

print(person1)  # Person(name='Alice', age=30, hobbies=[])
print(person2)  # Person(name='Bob', age=25, hobbies=['reading', 'coding'])
```

## 使用默认工厂函数

```python
from dataclasses import dataclass, field
from typing import List

@dataclass
class Product:
    name: str
    price: float
    tags: List[str] = field(default_factory=lambda: ["new", "sale"])

# 创建实例
product1 = Product(name="Laptop", price=999.99)
product2 = Product(name="Smartphone", price=499.99, tags=["popular"])

print(product1)  # Product(name='Laptop', price=999.99, tags=['new', 'sale'])
print(product2)  # Product(name='Smartphone', price=499.99, tags=['popular'])
```

## 控制字段的初始化和表示

```python
from dataclasses import dataclass, field

@dataclass
class User:
    username: str
    password: str = field(init=False, repr=False)  # 不包含在 __init__ 和 __repr__ 中
    email: str = field(default="no-email@example.com", repr=False)  # 默认值，不包含在 __repr__ 中

    def __post_init__(self):
        self.password = "hashed_password"  # 模拟密码哈希

# 创建实例
user = User(username="Alice")

print(user)  # User(username='Alice')
print(user.password)  # hashed_password
print(user.email)  # no-email@example.com
```

## 只读字段

```python
from dataclasses import dataclass, field

@dataclass
class Item:
    name: str
    price: float
    discount: float = 0.0
    final_price: float = field(init=False)  # 只读字段

    def __post_init__(self):
        self.final_price = self.price * (1 - self.discount)

# 创建实例
item = Item(name="Book", price=100.0, discount=0.1)

print(item)  # Item(name='Book', price=100.0, discount=0.1, final_price=90.0)
```
