Python 的 math 模块是一个标准库模块，提供了许多数学函数和常量，用于执行各种数学运算。这些功能涵盖了基本的数学运算、三角函数、对数函数、幂函数等。

1. 数学常量
- math.pi：圆周率 π，约等于 3.141592653589793。
- math.e：自然对数的底数 e，约等于 2.718281828459045。
- math.tau：τ（τ = 2π），约等于 6.283185307179586。
- math.inf：表示无穷大。
- math.nan：表示非数字（NaN）。

math.cos(x) 是余弦函数，它计算的是角度 x （以弧度为单位）的余弦值。余弦函数的值域是 [−1,1] ，即它的结果始终在 -1 和 1 之间。

```python
import math

# 不同 x 值的余弦值
x_values = [0, math.pi / 6, math.pi / 4, math.pi / 3, math.pi / 2, math.pi, 3 * math.pi / 2, 2 * math.pi]
for x in x_values:
    print(f"cos({x}) = {math.cos(x)}")
```

总结
- 当 x  接近 0 时，cos(x)  接近 1。
- 当 x  增大时，cos(x)  的值会在 -1 和 1 之间周期性变化，而不是单调趋向于 -1。
- 余弦函数是周期函数，周期为 2π ，因此它的值会在每个 2π  区间内重复。
