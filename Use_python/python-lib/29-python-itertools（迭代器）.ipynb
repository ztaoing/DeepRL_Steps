itertools 是 Python 标准库中的一个模块，提供了许多用于处理迭代器的高效工具函数。它被称为"Python 的迭代器宝石"，能够帮助你编写更简洁、高效的代码。

## 无限迭代器

| 函数 | 说明 | 示例 |
|------|------|------|
| `count(start, step)` | 从start开始无限计数 | `count(0, 2)` → 0, 2, 4, 6... |
| `cycle(iterable)` | 无限循环迭代元素 | `cycle([1,2,3])` → 1,2,3,1,2,3... |
| `repeat(object, times)` | 重复对象指定次数 | `repeat('A', 3)` → A,A,A |



```python
from itertools import count,cycle,repeat

# count
for i in count(10,5):
    if i > 25:
        break
    print(i,end=' ')

# cycle
from itertools import islice
print(list(islice(cycle('abc'),7)))

# repeat
print(list(repeat('hello',3)))
```

## 有限迭代器

| 函数 | 说明 |
|------|------|
| `accumulate(iterable, func)` | 累积计算（默认求和） |
| `chain(*iterables)` | 连接多个迭代器 |
| `compress(data, selectors)` | 根据选择器过滤元素 |
| `dropwhile(predicate, iterable)` | 跳过满足条件的元素 |
| `filterfalse(predicate, iterable)` | 过滤不满足条件的元素 |
| `groupby(iterable, key)` | 按关键函数分组 |
| `islice(iterable, start, stop, step)` | 切片迭代器 |
| `starmap(func, iterable)` | 解包参数后映射函数 |
| `takewhile(predicate, iterable)` | 取满足条件的元素 |
| `tee(iterable, n)` | 创建n个独立迭代器 |
| `zip_longest(*iterables, fillvalue)` | 最长长度zip |



```python
from itertools import accumulate,chain,compress,groupby,islice,zip_longest

# accumulate
print(list(accumulate([1,2,3,4])))

# chain
print(list(chain([1,2],[3,4],[5])))

# compress
data = ['a','b','c','d']
selectors = [1,0,1,0]
print(list(compress(data,selectors)))

# groupby
data = [('a',1),('a',2),('b',3)]
for key,group in groupby(data,key=lambda x:x[0]):
    print(key,list(group))
    
# islice
print(list(islice(range(10),2,8,2)))

# zip_longest
print(list(zip_longest([1,2],[3,4,5],fillvalue=0)))
```

```python
import itertools

# filterfalse - 反向过滤
print(list(itertools.filterfalse(lambda x: x%2==0, range(10))))
# [1,3,5,7,9] (奇数)

# dropwhile - 跳过开头
print(list(itertools.dropwhile(lambda x: x<5, [1,3,5,2,4])))
# [5,2,4] (遇到5后停止跳过)

# takewhile - 取开头
print(list(itertools.takewhile(lambda x: x<5, [1,3,5,2,4])))
# [1,3] (遇到5后停止)

# compress - 掩码过滤
data = ['A', 'B', 'C', 'D', 'E']
selectors = [1, 0, 1, 0, 1]
print(list(itertools.compress(data, selectors)))  # ['A', 'C', 'E']

# groupby - 分组（重要：输入必须先排序！）
data = [('A', 1), ('A', 2), ('B', 3), ('B', 4), ('A', 5)]
# 错误！A不连续，会被分成两组
for key, group in itertools.groupby(data, key=lambda x: x[0]):
    print(key, list(group))
# A [(A,1), (A,2)]
# B [(B,3), (B,4)]
# A [(A,5)]  <- A 再次出现！

# 正确用法：先排序
data.sort(key=lambda x: x[0])
for key, group in itertools.groupby(data, key=lambda x: x[0]):
    print(key, list(group))
# A [(A,1), (A,2), (A,5)]
# B [(B,3), (B,4)]
```

## 组合迭代器

| 函数 | 说明 |
|------|------|
| `product(*iterables, repeat)` | 笛卡尔积 |
| `permutations(iterable, r)` | 排列 |
| `combinations(iterable, r)` | 组合（无重复） |
| `combinations_with_replacement(iterable, r)` | 组合（可重复） |



```python
from itertools import product, permutations, combinations, combinations_with_replacement

# product 笛卡尔积
print(list(product('AB', [1, 2])))  # [('A',1), ('A',2), ('B',1), ('B',2)]

# permutations 排列
print(list(permutations('ABC', 2)))  
# [('A','B'), ('A','C'), ('B','A'), ('B','C'), ('C','A'), ('C','B')]

# combinations 组合
print(list(combinations('ABC', 2)))  # [('A','B'), ('A','C'), ('B','C')]

# combinations_with_replacement
print(list(combinations_with_replacement('AB', 2)))  # [('A','A'), ('A','B'), ('B','B')]
```

```python
# 读取大文件，分块处理

from itertools import islice

def read_chunks(file, chunk_size=1000):
    while True:
        chunk = list(islice(file, chunk_size))
        if not chunk:
            break
        yield chunk
```

```python
# 扁平化嵌套列表

from itertools import chain

nested = [[1, 2], [3, 4], [5]]
flat = list(chain.from_iterable(nested))  # [1, 2, 3, 4, 5]
```

```python
# 分组统计

from itertools import groupby

data = sorted(['apple', 'apricot', 'banana', 'blueberry'])
for key, group in groupby(data, key=lambda x: x[0]):
    print(f"{key}: {list(group)}")
```

```python
# 高效处理百万级数据

from itertools import islice,filterfalse

data = range(1000000)

# 只处理前10个偶数，不会遍历全部数据
result = list(islice(filterfalse(lambda x:x %2,data),10))
print(result)
```

| 分组处理 | `groupby` |
| 排列组合 | `product`, `permutations`, `combinations` |
