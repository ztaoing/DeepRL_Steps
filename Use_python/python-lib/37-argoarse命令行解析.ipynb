### argparse 模块用于解析命令行参数和选项。
-它支持自动生成帮助和使用消息，并且能够在用户给出不正确的参数时发出错误消息。

#### 创建 ArgumentParser 对象

首先，需要创建一个 ArgumentParser 对象，这是解析命令行参数的基础。

```python
import argparse

parser = argparse.ArgumentParser(description="这是一个示例程序")
```

#### 添加参数

使用 add_argument 方法向解析器添加参数。你可以指定参数的名称、类型、默认值、帮助信息等。

- metavar：在帮助信息中显示的参数名称。
- type：参数的类型。
- nargs：参数的数量。
- help：参数的帮助信息。
- dest：将参数值存储到命名空间的哪个属性。
- action：指定参数的处理方式。
- const：当 action='store_const' 时，存储的常量值。
- default：参数的默认值。

```python
parser.add_argument('integers',metavar = 'N',type=int,nargs = '+',help='一个整数列表')
parser.add_argument('--sum',dest='accumulate', action='store_const',const=sum,default=max,help='将整数相加')
```

#### 解析参数
- 使用 parse_args 方法解析命令行参数。这个方法会返回一个命名空间对象，其中包含所有解析的参数值。

```python
args = parser.parse_args()
print(args.integers)
print(args.accumulate)
```

```python
import argparse

# 创建 ArgumentParser 对象
parser = argparse.ArgumentParser(description="这是一个示例程序")

# 添加参数
parser.add_argument('integers', metavar='N', type=int, nargs='+',
                    help='一个整数列表')
parser.add_argument('--sum', dest='accumulate', action='store_const',
                    const=sum, default=max,
                    help='将整数相加 (默认: 找出最大值)')

# 解析参数
args = parser.parse_args()

# 使用解析的参数
if args.accumulate == sum:
    result = sum(args.integers)
else:
    result = max(args.integers)

print(f"Result: {result}")
```

假设将上述代码保存为 example.py，可以在命令行中运行以下命令：

```python
python example.py 1 2 3 4 --sum
```
