### 几个常用的非贪婪匹配Pattern

*? 重复任意次，但尽可能少重复

+? 重复1次或更多次，但尽可能少重复

?? 重复0次或1次，但尽可能少重复

{n,m}? 重复n到m次，但尽可能少重复

{n,}? 重复n次以上，但尽可能少重复

### .*?的用法

. 是任意字符


* 是取 0 至 无限长度

? 是非贪婪模式。

合在一起就是 ：取尽量少的任意字符，一般不会这么单独写，他大多用在：

.*?x ： 就是取前面任意长度的字符，直到一个x出现

re.match(pattern, string, flags=0)

pattern：正则表达式，用于定义匹配的模式。
string：要匹配的字符串。
flags：可选参数，用于修改正则表达式的匹配行为。常见的标志包括：
- re.IGNORECASE 或 re.I：忽略大小写。
- re.MULTILINE 或 re.M：多行模式，^ 和 $ 匹配每一行的开头和结尾。
- re.DOTALL 或 re.S：点号 . 匹配任意字符，包括换行符。

```python
import re

pattern = r"^hello"
string = "hello world"

match = re.match(pattern,string)
if match:
    print("匹配成功")
    print("匹配的内容：",match.group())
else:
    print("匹配失败！")
```

```python
import re

pattern = r"^hello"
string = "Hello World"

# 忽略大小写
match = re.match(pattern, string, flags=re.IGNORECASE)
if match:
    print("匹配成功！")
    print("匹配的内容:", match.group())
else:
    print("匹配失败！")
```

- match.group()：返回匹配的整个字符串。
- match.start()：返回匹配的起始位置。
- match.end()：返回匹配的结束位置。
- match.span()：返回一个元组 (start, end)，表示匹配的起始和结束位置。

```python
import re

pattern = r"^hello"
string = "hello world"

match = re.match(pattern, string)
if match:
    print("匹配成功！")
    print("匹配的内容:", match.group())
    print("起始位置:", match.start())
    print("结束位置:", match.end())
    print("匹配范围:", match.span())
else:
    print("匹配失败！")
```

re.match 只从字符串的开头进行匹配。如果需要在整个字符串中查找匹配项，可以使用 re.search。

## 常量、属性
1、re.A（re.ASCII）

让\w，\W，\b，\B，\d，\D，\s和\S 执行ASCII-只匹配完整的Unicode匹配代替。这仅对Unicode模式有意义，而对于字节模式则忽略。

2、re.I（re.IGNORECASE）

执行<b>不区分大小写</b>的匹配；类似的表达式也[A-Z]将匹配小写字母。

3、re.L（re.LOCALE）

　　让\w，\W，\b，\B和区分大小写的匹配取决于当前的语言环境。该标志只能与字节模式一起使用。不建议使用此标志，因为语言环境机制非常不可靠，它一次只能处理一种“区域性”，并且仅适用于8位语言环境。默认情况下，Python 3中已为Unicode（str）模式启用了Unicode匹配，并且能够处理不同的语言环境/语言。

4、re.M（re.MULTILINE）

　　指定时，模式字符'^'在字符串的开头和每行的开头（紧随每个换行符之后）匹配；模式字符''在字符串的末尾和每行的末尾（紧接在每个换行符之前）匹配。默认情况下，'^' 仅在字符串的开头，字符串'
'的末尾和字符串末尾的换行符（如果有）之前立即匹配。

5、re.S（re.DOTALL）

使'.'特殊字符与任何字符都匹配，包括换行符；没有此标志，'.'将匹配除换行符以外的任何内容。

### re.compile(pattern,flags = 0)

将正则表达式模式 编译为 正则表达式对象，可使用match()，search()以及下面所述的其他方法将其用于匹配

- re.search(pattern,string,flags=0)
  - 扫描字符串以查找正则表达式模式产生匹配项的第一个位置 ，然后返回相应的match对象。
  - 如果字符串中没有位置与模式匹配，则返回None；否则返回false。
  - 请注意，这与在字符串中的某个点找到零长度匹配不同。

```python
import re
# 将正则表达式模式 编译为 正则表达式对象
prog = re.compile('\d{2}')
# 通过调用group()方法得到匹配的字符串,如果字符串没有匹配，则返回None。
string1 = '12abc'
out = prog.search(string1).group()
print(out)

string2 = '123abc'
out = prog.search(string2).group()
print(out)
```

## re.match(pattern,string,flags=0)
- 如果字符串开头的零个或多个字符与正则表达式模式匹配，则返回相应的匹配对象。
- 如果字符串与模式不匹配，则返回None；否则返回false。
- 请注意，这与零长度匹配不同


```python
# 同search,不过在字符串开始处进行匹配，只会匹配一个对象
re.match('a','abcdef').group()
```
