```python
# 初始化一个长度为 N 的一维数组
array = [0] * 8
print(array)
```

```python
# 初始化一个形状为 MxN 的二维数组(矩阵)
matrix = [[0]* 8 for _ in range(9)]
print(matrix)
```

```python
# 排序
# Python 中排序主要使用 sorted() 和 .sort() 函数，在官网[2]有详细介绍，大家可以自行阅读。

returns = sorted([3,6,1,9,2,6])
print(returns)

a = [3,6,1,9,2,6]
a.sort()
print(a)

sorted({1: 'D', 2: 'B', 3: 'B', 4: 'E', 5: 'A'})
```
