## 动态获取对象的属性值 getattr(对象，属性名)

```python
class Person:
    def __init__(self,a,b):
        self.name = a 
        self.age = b
p1 = Person('小明',20)
field = input('请输入你想要的属性：')

# 获取属性
print(getattr(p1,field))
```

## setattr(对象，属性名，新值) 设置对象的属性

```python
field = input('请输入想要修改的属性：')
new_value = input('请输入更新后的值：')
setattr(p1,field,new_value)
print(vars(p1))
```

## vars()是 Python 的内置函数，用来查看当前局部作用域或某个对象的“属性字典”（即 __dict__）

- dir() 列出所有可访问属性名（含类属性、方法、继承来的）。
- vars() 只给实例自己拥有的可写属性字典，速度更快，内容更“干净”。

```python
def demo():
    a = 1
    b = 2
    print(vars())      # {'a': 1, 'b': 2}

demo()
```

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

p = Point(3, 4)
print(vars(p))         # {'x': 3, 'y': 4}
```

## isinstance(对象，类名)
- 判断对象是否是 某个类 创建出来的

```python
class Animal: pass
class Dog(Animal): pass

d = Dog()
type(d) is Animal       # False
isinstance(d, Animal)   # True   ✅ 把子类也算进去
```

```python
>>> isinstance(123, int)
True
>>> isinstance('abc', (str, bytes))   # 多个候选写元组
True
>>> isinstance([], (list, tuple))
True
```
