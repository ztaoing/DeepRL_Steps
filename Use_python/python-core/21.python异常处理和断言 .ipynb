## 常见的异常类型如下：
- Exception：<b>所有异常的基类，因为所有python异常类都是基类Exception的其中一员，异常都是从基类Exception继承的，并且都在exceptions python 模块中定义。</b>

- AttributeError：   属性错误，特性引用和赋值失败时会引发属性错误
- NameError：        试图访问的变量名不存在
- SyntaxError：      语法错误，代码形式错误
- IOError：          python ioerror，一般常见于打开不存在文件时会引发IOError错误，也可以解理为输出输入错误
- KeyError：         使用了映射中不存在的关键字（键）时引发的关键字错误
- IndexError：       索引错误，使用的索引不存在，常索引超出序列范围
- TypeError：        类型错误，内建操作或是函数应于在了错误类型的对象时会引发类型错误
- ZeroDivisonError： 除数为0，在用除法操作时，第二个参数为0时引发了该错误
- ValueError：       值错误，传给对象的参数类型不正确，像是给int()函数传入了字符串数据类型的参数。

核心原则是：捕获你预期并能处理的错误，让无法处理的错误暴露出来以便修复。

一个完整的异常处理结构通常包含四个部分：
- try: 放置可能出错的代码。
- except: 放置出错后执行的代码。
- else: （可选）只有当 try 没有出错时才执行的代码。
- finally: （可选）无论是否出错都会执行的代码（常用于清理资源）。

```python
try:
    # 1-尝试执行的代码
    result = 10/0
except (ZeroDivisionError, FileNotFoundError, PermissionError)as e:
    # 2-如果发生ZeroDivisionError，执行这里
    print(f'出错了：不能除以0！详情：{e}')
except Exception as e:
    # 3-如果发生其他任何异常，执行这里
    print(f'发生未知错误：{e}')
else:
    # 4-如果没有发生任何异常，执行这里
    print('计算成功，结果是：',result)
finally:
    # 5-无论是否出错，最后都会执行这里
    print('执行完毕，清理资源')
```

### 最佳实践
#### A 记录日志而不是打印

在生产环境中，不要只用 print，应该使用 logging 模块记录堆栈信息，方便排查。

```python
import logging

try:
    risky_operation()
except Exception as e:
    # 记录完整堆栈跟踪
    logging.error("操作失败", exc_info=True) 
    # 或者 logging.exception("操作失败")
```

#### 经典应用场景：文件操作

```python
file = None
try:
    file = open('data.txt', 'r')
    content = file.read()
    # 假设这里可能发生转换错误
    number = int(content)
except FileNotFoundError:
    print("错误：文件不存在。")
except ValueError:
    print("错误：文件内容不是有效的数字。")
except Exception as e:
    print(f"发生了未预料的错误：{e}")
else:
    print(f"读取成功，数字是：{number}")
finally:
    # 无论成功与否，都要关闭文件
    if file:
        file.close()
        print("文件已关闭。")
```

注：在现代 Python 中，更推荐使用 上下文管理器 (with 语句) 来自动处理 finally 中的关闭操作，代码更简洁：

```python
try:
    with open('data.txt', 'r') as file:
        content = file.read()
        number = int(content)
except FileNotFoundError:
    ...
```
