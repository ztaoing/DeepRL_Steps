```python
import warnings
```

```python
# 忽略UserWarning类型的警告
warnings.filterwarnings("ignore",category=UserWarning)
```

```python
warnings.warn("this is a waring message", UserWarning)
```

```python
# 只显示一次警告：
warnings.filterwarnings("once",category=UserWarning)
```

```python
# 将警告记录到日志文件中
import logging

logging.basicConfig(filename="warnings.log",level=logging.WARNING)
warnings.simplefilter("always",UserWarning)
warnings.showwarning = lambda msg,cat,fname,lineno,file=None,\
line =None:logging.warning(f"{cat.__name__}:{msg}(in {fname}:{lineno})")
```
