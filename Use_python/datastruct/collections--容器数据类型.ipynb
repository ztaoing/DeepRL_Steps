## defaultdict  
- 字典的子类，通过调用用户指定的工厂函数，为键提供默认值。

```python
from collections import defaultdict
```

```python
s = [('yellow',1),('blue', 2), ('yellow', 3), ('blue', 4), ('red', 1)]
# 使用list作为工厂函数，很轻松地将（键-值对组成的）序列转换为（键-列表组成的）字典：
d = defaultdict(list)
for k,v in s:
    d[k].append(v)
sorted(d.items())
```

```python
s =  'mississippi'
d = defaultdict(int)
for k in s:
    d[k]+=1
sorted(d.items())
```

```python
s = [('red', 1), ('blue', 2), ('red', 3), ('blue', 4), ('red', 1), ('blue', 4)]
d = defaultdict(set)

for k,v in s:
    d[k].add(v)

sorted(d.items())
```
