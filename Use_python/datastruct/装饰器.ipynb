在 Python 中，装饰器（Decorator）是一种非常强大的功能，
- 用于在不修改原始函数或类定义的情况下，动态地添加或修改函数或类的行为。
- 装饰器本质上是一个函数，它接受一个函数或类作为参数，并返回一个新的函数或类。通过使用装饰器，你可以实现诸如日志记录、性能测试、事务处理、权限校验等功能，而无需修改原始代码。

## 1. 装饰器的基本概念
装饰器是一个函数，它接受一个函数或类作为参数，并返回一个新的函数或类。装饰器通常用于在函数或类的调用前后添加额外的逻辑。
## 2. 函数装饰器
函数装饰器用于修改函数的行为。

### 日志记录
- 假设你有一个函数，你希望在每次调用时记录日志：

```python
def log_decorator(func):
    def wrapper(*args, **kwargs):
        print(f"Calling function {func.__name__} with args {args} and kwargs {kwargs}")
        result = func(*args, **kwargs)
        print(f"Function {func.__name__} returned {result}")
        return result
    return wrapper

@log_decorator
def add(a, b):
    return a + b

result = add(3, 4)
```

## 3. 类装饰器
- 类装饰器用于修改类的行为。以下是一个简单的示例：

示例：添加类属性
假设你有一个类，你希望在类定义时自动添加一个属性：

```python
# add_class_attribute 是一个类装饰器，它在类定义时添加了一个新的属性 new_attribute。
def add_class_attribute(cls):
    cls.new_attribute = "This is a new attribute"
    return cls

@add_class_attribute
class MyClass:
    pass

obj = MyClass()
print(obj.new_attribute)  # 输出: This is a new attribute
```

## 4. 带参数的装饰器
- 装饰器也可以接受参数，这使得装饰器更加灵活。以下是一个带参数的装饰器示例：

示例：重复执行
假设你有一个函数，你希望它能够重复执行指定的次数：

```python
def repeat(times):
    
    def decorator(func):
        def wrapper(*args, **kwargs):
            results = []
            for _ in range(times):
                results.append(func(*args, **kwargs))
            return results
        return wrapper
    
    return decorator

@repeat(times=3)
def say_hello(name):
    return f"Hello, {name}"

results = say_hello("Alice")
print(results)  # 输出: ['Hello, Alice', 'Hello, Alice', 'Hello, Alice']
```

## 5. 使用 functools.wraps
当你定义装饰器时，使用 functools.wraps 可以保留原始函数的元信息（如函数名、文档字符串等）。

```python
import functools

def log_decorator(func):
    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        print(f"Calling function {func.__name__} with args {args} and kwargs {kwargs}")
        result = func(*args, **kwargs)
        print(f"Function {func.__name__} returned {result}")
        return result
    return wrapper

@log_decorator
def add(a, b):
    """This function adds two numbers."""
    return a + b

print(add.__name__)  # 输出: add
print(add.__doc__)   # 输出: This function adds two numbers.
```

## 6. 内置装饰器
Python 提供了一些内置的装饰器，用于常见的功能：
- @staticmethod：将方法定义为静态方法。
- @classmethod：将方法定义为类方法。
- @property：将方法定义为属性。
- @functools.lru_cache：为函数添加缓存功能。
- @typing.final：标记类或方法为不可覆盖。

```python
class Person:
    def __init__(self, first_name, last_name):
        self.first_name = first_name
        self.last_name = last_name

    @property
    def full_name(self):
        return f"{self.first_name} {self.last_name}"

person = Person("John", "Doe")
print(person.full_name)  # 输出: John Doe
```
