## Any
- Any 是一个特殊的类型，表示可以是任何类型。
- 当一个变量被标注为 Any 时，静态类型检查器不会对其进行任何类型检查。这在某些情况下非常有用，例如当你不确定变量的具体类型，或者需要处理动态类型时。

```python
from typing import Any

def process_value(value:Any) -> None:
    print(value)

process_value(10)
process_value("hello")
process_value([1,2,3])
```

# 2. Callable
- Callable 用于表示一个可调用对象，例如函数、方法或类。你可以指定 Callable 的参数类型和返回类型，使其更加具体。


```python
from typing import Callable

def add(a:int,b:int) -> int:
    return a+b
def process_function(func:Callable[[int,int],int]) -> None:
    print(func(2,3))
    
process_function(add)
```

## 3. Optional
- Optional 用于表示一个变量可以是某种类型，也可以是 None。这是处理可选值时非常有用的类型提示。

```python
from typing import Optional

# Optional[int] 表示函数 get_value 的返回值可以是 int 类型，也可以是 None。
def get_value(flag:bool)-> Optional[int]:
    if flag:
        return 66
    else:
        return None
value = get_value(True)
if value is not None:
    print(value)
```

## 4. Sized
- Sized 是一个协议（protocol），表示一个对象具有长度或大小，即支持 len() 函数。任何实现了 __len__ 方法的类都可以被视为 Sized。

```python
from typing import Sized

class Myclass:
    def __init__(self,data:list):
        self.data = data
    def __len__(self)-> int :
        return len(self.data)
def process_sized(size_obj:Sized) -> None:
    print(len(size_obj))

my_obj = Myclass([1,2,3])
process_sized(my_obj)
```

## List、Tuple、Dict、Set
- 用于表示集合类型。

```python
from typing import List, Tuple, Dict, Set

def process_list(values: List[int]) -> None:
    print(values)

def process_tuple(values: Tuple[int, str]) -> None:
    print(values)

def process_dict(values: Dict[str, int]) -> None:
    print(values)

def process_set(values: Set[int]) -> None:
    print(values)
```

## TypeVar
- 用于定义泛型类型变量。

```python
from typing import TypeVar, List

T = TypeVar('T')

def first_element(lst: List[T]) -> T:
    return lst[0]
```

## NewType
- 用于创建新的类型，这些类型在运行时与原始类型相同，但在类型检查时被视为不同的类型。

```python
from typing import NewType

UserId = NewType('UserId', int)

def get_user(user_id: UserId) -> None:
    print(user_id)

user_id = UserId(123)
get_user(user_id)

print(int == UserId)
```

## TypedDict
- 用于定义字典的键和值的类型。

```python
from typing import TypedDict

class Point(TypedDict):
    x: int
    y: int

def process_point(point: Point) -> None:
    print(point['x'], point['y'])
```

## Literal 
- 是一个非常有用的类型注解工具，用于指定一个变量或函数参数只能取特定的字面量值。这在某些情况下可以提高代码的可读性和类型安全性。
- Literal 的主要作用是为静态类型检查器提供类型信息。


### 限制函数参数的值
- 假设你有一个函数，它接受一个字符串参数，但这个参数只能是 "start" 或 "stop"。你可以使用 Literal 来指定这一点：

```python
from typing import Literal
```

```python
def control_system(command:Literal['start','stop'])->None:
    if command == 'start':
        print("systemstarting...")
    elif command == 'stop':
        print("systemstoping...")
    else:
        raise ValueError("Invalid command")
control_system('start')
control_system('stop')
```

```python
control_system('pause')
```

# Protocol
- 是一个非常强大的工具，用于定义鸭子类型（duck typing）的接口。它允许你指定一个类必须实现哪些方法或属性，而不需要这些类显式地继承某个基类。这在编写灵活且可扩展的代码时非常有用。

MyProtocol 定义了一个接口，要求实现该接口的类必须提供 method1 和 method2 方法。

```python
from typing import Protocol

class MyProtocol(Protocol):
    def method1(self) -> None:
        ...
    def method2(self) -> int:
        ...
```

```python
from typing import Protocol

class Starter(Protocol):
    def start(self) -> None:
        ...
    def stop(self) -> None:
        ...
```

```python
class Engine:
    def start(self) -> None:
        print("Engine starting...")

    def stop(self) -> None:
        print("Engine stopping...")

class Car:
    def start(self) -> None:
        print("Car starting...")

    def stop(self) -> None:
        print("Car stopping...")
```

```python
def control_vehicle(vehicle: Starter) -> None:
    vehicle.start()
    vehicle.stop()

# 使用示例
engine = Engine()
car = Car()

control_vehicle(engine)  # 输出: Engine starting... Engine stopping...
control_vehicle(car)     # 输出: Car starting... Car stopping...
```

```python
def get_positive_number(prompt):
    while True:
        try:
            value = float(input(prompt))
            if value <= 0:
                raise ValueError("输入的值必须大于零")
            return value
        except ValueError as e:
            print(f"无效输入: {e}")

# 使用函数
number = get_positive_number("请输入一个正数: ")
print(f"你输入的正数是: {number}")
```
