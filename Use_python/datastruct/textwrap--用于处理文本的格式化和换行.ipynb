```python
import textwrap
```

```python
text = "这是一个非常长的文本，需要被分割成多个较短的行。我们将使用 textwrap 模块来实现这一点。"
```

 # textwrap.wrap()
 - wrap 函数将长文本分割成多个较短的行，并返回一个包含这些行的列表。每个行的长度不超过指定的宽度。
 - textwrap **模块不会改变原始文本的内容，只是在显示时进行格式化。**

```python
wapped_lines = textwrap.wrap(text,width=20)
for line in wapped_lines:
    print(line)
```

# textwrap.fill()
- fill 函数与 wrap 类似，但它返回的是一个字符串，而不是一个列表。它会将长文本格式化为一个段落，每行的长度不超过指定的宽度。

```python
wrapped_text = textwrap.fill(text, width=20)

print(wrapped_text)
```

# 缩进和对齐
- textwrap 模块还提供了 dedent 和 indent 函数，用于处理文本的缩进。
## textwrap.dedent()
- dedent 函数用于去除字符串中的缩进。这对于多行字符串特别有用，可以去除不必要的缩进。

```python
import textwrap

text = """
    这是一个多行字符串，
    每一行都有缩进。
    我们将使用 dedent 函数
    去除这些缩进。
"""
dedented_text = textwrap.dedent(text)

print(dedented_text)
```

## textwrap.indent()
- indent 函数用于给字符串添加缩进。你可以指定缩进的字符和前缀。

```python
import textwrap

text = "这是一个需要缩进的文本。"
indented_text = textwrap.indent(text, "    ")

print(indented_text)
print(text)
```

## 自定义格式化
- textwrap 模块还提供了一个 TextWrapper 类，允许你自定义文本的格式化行为。

```python
import textwrap

text = "这是一个非常长的文本，需要被分割成多个较短的行。我们将使用 textwrap 模块来实现这一点。"

wrapper = textwrap.TextWrapper(width=20, initial_indent=">", subsequent_indent="   ")
wrapped_text = wrapper.fill(text)

print(wrapped_text)
```
