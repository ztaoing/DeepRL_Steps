```python
class Pepole:
    def __init__(self,x):
        self.__age = x
    @property
    def age(self):
        return self.__age
    @age.setter
    def age(self,newVal):
        if newVal > 0:
            self.__age = newVal
    @age.deleter
    def age(self):
        self.__age = 0
        print('删除age属性')
child = Pepole(15)
print(child.age)
child.age = 18
print(child.age)
del(child.age) 
print(child.age)
```
