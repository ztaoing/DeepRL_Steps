# tokenizer
- 用于将文本数据转换为模型可以理解的数值化表示

Tokenizer 会自动处理一些特殊标记，这些标记对于模型的训练和推理非常重要：
- [CLS]：分类任务中用于表示分类标记。
- [SEP]：分隔符，用于分隔句子。
- [PAD]：填充标记，用于填充序列以达到固定长度。
- [MASK]：掩码标记，用于掩码语言模型任务。


```python
# 初始化 Tokenizer
from transformers import BertModel,BertTokenizer
from pathlib import Path

# 加载本地的bert
BERT_PATH_LIB = Path('../../model/bert-base-uncased')


# 列出目录中的所有文件
files = [item.name for item in BERT_PATH_LIB.iterdir() if item.is_file()]

print("Files in '", BERT_PATH_LIB, "':")
for file in files:
    print(file)

# 加载预训练的tokenizer
BERT_PATH = '../../model/bert-base-uncased'
tokenizer = BertTokenizer.from_pretrained(BERT_PATH)
```

```python
# 编码
# 将文本编码为模型可以理解的编码表示

input = tokenizer("Hello, my dog is cute",return_tensors='pt')
print(input)
```

```python
# 增强的编码函数
out = tokenizer.encode_plus(
    text=sents[0],
    text_pair=sents[1],
    truncation=True,
    padding='max_length',
    max_length=30,
    add_special_tokens=True,
    return_tensors=None,
    return_token_type_ids=True,
    return_attention_mask=True,
    return_special_tokens_mask=True,
    return_length=True,
)
#input_ids 就是编码后的词
#token_type_ids 第一个句子和特殊符号的位置是0,第二个句子的位置是1
#special_tokens_mask 特殊符号的位置是1,其他位置是0
#attention_mask pad的位置是0,其他位置是1
#length 返回句子长度
for k,v in out.items():
    print(k,':',v)

tokenizer.decode(out['input_ids'])
```

```python
# 批量编码句子
out = tokenizer.batch_encode_plus(
    batch_text_or_text_pairs=[sents[0], sents[1]],
    add_special_tokens=True,

    #当句子长度大于max_length时,截断
    truncation=True,

    #一律补零到max_length长度
    padding='max_length',
    max_length=15,

    #可取值tf,pt,np,默认为返回list
    return_tensors=None,

    #返回token_type_ids
    return_token_type_ids=True,

    #返回attention_mask
    return_attention_mask=True,

    #返回special_tokens_mask 特殊符号标识
    return_special_tokens_mask=True,

    #返回offset_mapping 标识每个词的起止位置,这个参数只能BertTokenizerFast使用
    #return_offsets_mapping=True,

    #返回length 标识长度
    return_length=True,
)

#input_ids 就是编码后的词
#token_type_ids 第一个句子和特殊符号的位置是0,第二个句子的位置是1
#special_tokens_mask 特殊符号的位置是1,其他位置是0
#attention_mask pad的位置是0,其他位置是1
#length 返回句子长度
for k, v in out.items():
    print(k, ':', v)

tokenizer.decode(out['input_ids'][0]), tokenizer.decode(out['input_ids'][1])
```

# 中文分类

```python
import torch 
from datasets import load_from_disk

class Dataset(torch.utils.data.Dataset):
    def __init__(self,split):
        all_dataset = load_from_disk('../../datasets/ChnSentiCorp')
        self.dataset = all_dataset[split]
        
    def __len__(self):
        return len(self.dataset)
    
    def __getitem__(self,i):
        text = self.dataset[i]['text']
        label = self.dataset[i]['label']
        return text,label
    
dataset = Dataset('train')

len(dataset),dataset[0]
```

```python
from transformers import BertTokenizer

# 加载字典和分词工具
token = BertTokenizer.from_pretrained('../../model/bert-base-uncased')
token
```

```python
# 批处理
def collate_fn(data):
    sents = [i[0]  for i in data]
    labels = [i[1] for i in data]
    
    # 编码
    data = token.batch_encode_plus(batch_text_or_text_pairs=sents,
                                  truncation=True,
                                  padding='max_length',
                                  max_length = 500,
                                  return_tensors='pt',
                                  return_length=True)
    input_ids = data['input_ids'] # 文字编码之后的数字
    attention_mask = data['attention_mask'] # 填充0的位置
    token_type_ids = data['token_type_ids']
    labels = torch.LongTensor(labels)
    
    print(data['length'],data['length'].max())
    return input_ids,attention_mask,token_type_ids,labels

# 数据加载

loader = torch.utils.data.DataLoader(dataset=dataset,
                                    batch_size=16,
                                    collate_fn=collate_fn,
                                    shuffle=True,
                                    drop_last=True)
for i,(input_ids,attention_mask,token_type_ids,labels) in enumerate(loader):
    break
    
print(len(loader))

input_ids.shape,attention_mask.shape,token_type_ids.shape,labels
    
```

```python
from transformers import BertModel

# 加载预训练模型
pretrained = BertModel.from_pretrained('../../model/bert-base-uncased')

# 不进行训练，所以不需要计算梯度
for param in pretrained.parameters():
    param.requires_grad_(False)
    
# 模型试计算
out = pretrained(input_ids=input_ids,attention_mask=attention_mask,token_type_ids=token_type_ids)

out.last_hidden_state.shape
```

```python
# 定义下游任务模型

class Model(torch.nn.Module):
    def __init__(self):
        super().__init__()
        self.fc = torch.nn.Linear(768,2)
    def forward(self,input_ids,attention_mask,token_type_ids):
        with torch.no_grad():
            out = pretrained(input_ids=input_ids,
                             attention_mask=attention_mask,
                             token_type_ids=token_type_ids)
        out = self.fc(out.last_hidden_state[:,0])
        out = out.softmax(dim=1)
        return out
model = Model()
model(input_ids=input_ids,attention_mask=attention_mask,token_type_ids=token_type_ids).shape
```

```python
from transformers import AdamW

# 训练
optimizer = AdamW(model.parameters(),lr=5e-4)
criterion = torch.nn.CorssEntropyLoss()

model.train()
for i,(input_ids,attention_mask,token_type_ids,labels) in enumerate(loader):
    out = model(input_ids=input_ids,attention_mask=attention_mask,token_type_ids=token_type_ids)
    
    loss = criterion(out,labels)
    loss.backward()
    optimizer.step()
    optimizer.zero_grad()
    
    # 打印
    if i%5 == 0:
        out = out.argmax(dim=1)
        accuracy = (out == labels).sum().item()/len(labels)
        print(i,loss.item(),accuracy)
        
    if i==300:
        break
```

```python
# 测试
def test():
    model.eval()
    correct = 0
    total = 0
    
    loader_test = torch.utils.data.DataLoader(dataset = Dataset('validation'),
                                             batch_size = 32,
                                             collate_fn=collate_fn,
                                             shuffle=True,
                                             drop_last=True)
    for i,(input_ids,attention_mask,token_type_ids,labels) in enumerate(loader_test):
        if i==5:
            break
        print(i)
        
        with torch.no_grad():
            out = model(input_ids=input_ids,
                       attention_mask=attention_mask,
                       token_type_ids=token_type_ids)
            out = out.argmax(dim=1)
            correct += (out == labels).sum().item()
            total += len(labels)
    print(correct/total)
    
test()
            
```
