# tokenizer
- 用于将文本数据转换为模型可以理解的数值化表示

Tokenizer 会自动处理一些特殊标记，这些标记对于模型的训练和推理非常重要：
- [CLS]：分类任务中用于表示分类标记。
- [SEP]：分隔符，用于分隔句子。
- [PAD]：填充标记，用于填充序列以达到固定长度。
- [MASK]：掩码标记，用于掩码语言模型任务。


```python
# 初始化 Tokenizer
from transformers import BertModel,BertTokenizer
from pathlib import Path

# 加载本地的bert
BERT_PATH_LIB = Path('../../model/bert-base-uncased')


# 列出目录中的所有文件
files = [item.name for item in BERT_PATH_LIB.iterdir() if item.is_file()]

print("Files in '", BERT_PATH_LIB, "':")
for file in files:
    print(file)

# 加载预训练的tokenizer
BERT_PATH = '../../model/bert-base-uncased'
tokenizer = BertTokenizer.from_pretrained(BERT_PATH)
```

```python
# 编码
# 将文本编码为模型可以理解的编码表示

input = tokenizer("Hello, my dog is cute",return_tensors='pt')
print(input)
```

```python
# 解码
decoded_text = tokenizer.decode(input['input_ids'][0],skip_special_tokens=True)
print(decoded_text)
```

```python
# 分词
tokens = tokenizer.tokenize("Hello, my dog is cute")
print(tokens)
```

```python
# 将分词结果转换为 ID
token_ids = tokenizer.convert_tokens_to_ids(tokens)
print(token_ids)
```

```python
# 将 ID 转换为分词结果
tokens = tokenizer.convert_ids_to_tokens(token_ids)
print(tokens)
```
