# 加入噪声的目的 ——> 负载均衡
- 如果不加入噪声，模型会总是选择同样的少数专家，
- 加入噪声后，所有专家都有机会被选择，负载更均衡
- 类比于强化学习中的epsilon-贪心算法

加噪声 = 负载均衡 + 防止崩溃 + 探索多样性

没有噪声注入，MoE很容易图画成”只有一个专家的稠密模型“

```python
# 伪代码示意

def select_experts_with_noise(logits,noise_type='uniform',num_samples=4):
    """
    logits:[num_experts] 原始分数
    """
    # 1st expert :确定选择(无噪声)
    first_expert = argmax(logits)
    
    # mask掉1st expert
    masked_logits = logits.clone()
    masked_logits[first_expert] = -inf
    # 加噪声：从某种分布中采样
    if noise_type == 'uniform':
        noise = torch.rand(num_experts) * noise_scale # uniform(0,1)
    elif nosie_type == 'gaussian':
        noise = torch.randn(num_experts) * noise_std # 正太分布 
    # 加噪声后选择最大的：
    noisy_logits = masked_logits + noise
    # 2st
    second_expert = argmax(noisy_logits[num_samples]) # 从采样中选择
    
    return first_expert,second_expert
```
