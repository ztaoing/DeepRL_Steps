##  trainer继承

```python
transformers.Trainer (基础)
    ↑
trl.trainer.base.BaseTrainer (TRL 基础)
    ↑
    ├── SFTTrainer (监督微调)
    ├── RewardTrainer (奖励模型)
    ├── PPOTrainer (PPO 强化学习)
    ├── DPOTrainer (直接偏好优化)
    ├── GRPOTrainer (组相对策略优化) ⭐
    ├── PRMTrainer (过程奖励模型)
    ├── NashMDTrainer (Nash-MD)
    └── ... (其他方法)
```
