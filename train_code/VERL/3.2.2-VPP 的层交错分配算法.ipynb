# VPP 的层交错分配算法

Megatron 的 VPP 层分配策略 (interleaved schedule):

总层数 = L = 16
PP = p = 4
VPP = v = 2

每个物理 Stage 的层数 = L / p = 4
每个 VPP Stage 的层数 = L / (p * v) = 2

分配方式:
VPP Stage 0:  Layer 0, 1          (GPU 0, 前半)
VPP Stage 1:  Layer 2, 3          (GPU 1, 前半)
VPP Stage 2:  Layer 4, 5          (GPU 2, 前半)
VPP Stage 3:  Layer 6, 7          (GPU 3, 前半)

VPP Stage 4:  Layer 8, 9          (GPU 0, 后半)
VPP Stage 5:  Layer 10, 11        (GPU 1, 后半)
VPP Stage 6:  Layer 12, 13        (GPU 2, 后半)
VPP Stage 7:  Layer 14, 15        (GPU 3, 后半)

## 可视化：

```python
GPU 0: part1:[0,1]        part2:[8,9]      ← 2 个 VPP Stage
GPU 1: part1:[2,3]        part2:[10,11]
GPU 2: part1:[4,5]        part2:[12,13]
GPU 3: part1:[6,7]        part2:[14,15]

普通 PP (无 VPP):
GPU 0: [0,1,2,3]                 ← 1 个大 Stage
GPU 1: [4,5,6,7]
GPU 2: [8,9,10,11]
GPU 3: [12,13,14,15]
```
