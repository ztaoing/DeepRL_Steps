# DataProto (协议)：

在强化学习（RLHF）训练中，数据非常复杂：
- 既有需要在 GPU 上计算的 Tensor（如 input_ids），
- 又有需要保留的原始文本（如 prompt），
- 还有各种元数据（如 temperature）。

##### DataProto 的出现就是为了解决这些异构数据的统一管理、切分和传输问题。

- 这是 VERL 框架定义的统一数据协议（Data Protocol）。
- 它是一个专门设计的类（Class），不仅仅是存数据，还定义了数据如何在 Actor、Critic、Reward Model 等不同模块之间传递。
- 结构：DataProto 内部通常包含三个主要部分：
| 属性名 | 数据类型 | 用途 | 示例 |
| :--- | :--- | :--- | :--- |
| `batch` | `TensorDict` | 计算核心。存放所有参与模型前向/反向传播的 Tensor 数据。支持像单个 Tensor 一样进行切片、拼接、设备转移（CPU/GPU）。 | `input_ids`, `attention_mask`, `values`, `log_probs` |
| `non_tensor_batch` | `Dict` | 辅助信息。存放无法放入 Tensor 或不需要计算的数据，通常是字符串或列表。 | 原始 `prompt` 文本, 图片路径, 数据库查询结果 |
| `meta_info` | `Dict` | 全局配置。存放与具体样本无关的全局信息或统计量。 | `temperature`, `n_samples`, `global_steps` |


###### 局限性与演进
<font color='red'>虽然 DataProto 在标准的 RLHF（如 PPO）中非常高效，但在Agent 训练场景下（特别是异步、长尾延迟场景），它的“按 Batch 组织数据”的方式可能会成为瓶颈。</font>

## repeat() 是 VERL 框架中用于数据增强或多副本生成的关键方法。

它的核心作用是：将当前的 DataProto 对象中的样本复制多份，生成一个新的、样本数量扩大的 DataProto 对象。

在强化学习（特别是 PPO）的训练流程中，这个方法通常用于“多次采样”场景。


1. 核心功能：它是如何工作的？
假设你有一个 DataProto 对象 data，其中包含 4 个样本（Batch Size = 4）：
- 原始数据：
    - input_ids: [A, B, C, D] (Tensor)
    - prompts: ["Prompt A", "Prompt B", "Prompt C", "Prompt D"] (List)
当你调用 new_data = data.repeat(n=2) 时，它会执行以下操作：
- Tensor 数据 (batch)：
    - 沿着 Batch 维度（通常是第 0 维）进行复制拼接。
    - 结果：[A, B, C, D, A, B, C, D]
- 非 Tensor 数据 (non_tensor_batch)：
    - 自动将列表内容复制并拼接。
    - 结果：["Prompt A", ..., "Prompt D", "Prompt A", ..., "Prompt D"]
- 元数据 (meta_info)：
    - 通常会保持不变，或者根据逻辑进行广播。
    - 结果：你得到了一个新的 DataProto，其 Batch Size 变成了 8 (4 * 2)。
    
2. 为什么要用 repeat()？（应用场景）

在 VERL 的 PPO 实现中，repeat() 主要用于实现 n_samples_per_prompt 功能。

###### DataProto 帮你把 Tensor 和 List 的处理统一了：

```python
# 伪代码逻辑
def repeat(self, n):
    # 1. 处理 Tensor
    # torch.repeat_interleave 或 torch.cat 用于复制张量
    new_tensor_batch = {k: v.repeat_interleave(n, dim=0) for k, v in self.batch.items()}
    
    # 2. 处理 List/Non-tensor
    # 列表乘法复制内容
    new_non_tensor = {k: v * n for k, v in self.non_tensor_batch.items()}
    
    return DataProto(batch=new_tensor_batch, non_tensor_batch=new_non_tensor)
```

# DataProto.from_single_dict(batch_dict)

from_single_dict (方法)：

- 这是一个静态工厂方法。它的作用是将那个“朴素”的 batch_dict 解析，并自动归类填充到 DataProto 的内部结构中。
- 目的：让后续的所有计算（如 actor_rollout_wg.generate_sequences）都能接收统一格式的数据，而不需要关心数据最初长什么样。
