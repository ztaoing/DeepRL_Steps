- https://github.com/verl-project/verl/blob/main/verl/utils/megatron_utils.py#L1251



````python
# 文件: verl/workers/sharding_manager/megatron_vllm.py

from verl.utils.megatron_utils import load_megatron_model_to_gpu, offload_megatron_model_to_cpu, per_tensor_generator

def per_tensor_generator(actor_module, model_config, weight_converter,
                         transformer_config, layer_name_mapping):
    """
    逐张量遍历 Megatron 模型参数，转换为 vLLM 格式
    """
    for model in actor_module:
        for name, param in model.named_parameters():
            # 1. 获取 Megatron 格式的参数 (已按 TP/PP 分片)
            megatron_param = param
            
            # 2. AllGather PP 维度 (在 micro DP 组内)
            # AllGatherPPModel 内部实现
            gathered_param = all_gather_pp_param(megatron_param)
            
            # 3. 层名映射: Megatron 层名 → vLLM 层名
            vllm_name = layer_name_mapping.get(name, name)
            
            # 4. 权重格式转换 (如 QKV 重排)
            if weight_converter is not None:
                converted_param = weight_converter(gathered_param, vllm_name)
            else:
                converted_param = gathered_param
            
            # 5. 按 vLLM 的 TP 方式重新分片
            vllm_shard = shard_for_vllm_tp(converted_param, vllm_tp_size)
            
            yield vllm_name, vllm_shard
````

````python
def per_tensor_generator(
    actor_module,
    model_config,
    weight_converter,
    transformer_config,
    layer_name_mapping,
    convert_qkv_gate_up_by_simple_split=True,
):
    from megatron.core import parallel_state as mpu

    pp_rank = mpu.get_pipeline_model_parallel_rank()
    ep_size = mpu.get_expert_model_parallel_world_size()
    etp_size = mpu.get_expert_tensor_parallel_world_size()
    ep_group = mpu.get_expert_model_parallel_group()
    etp_group = mpu.get_expert_tensor_parallel_group()
    vpp_size = len(actor_module)
    all_gather_group = mpu.get_tensor_model_parallel_group()
    all_gather_group_size = torch.distributed.get_world_size(group=all_gather_group)

    def tensor_generator():
        for scan_vpp_idx in range(vpp_size):
            existing_keys = set()
            model = unwrap_model(actor_module[scan_vpp_idx])
            for name, param in model.named_parameters():
                existing_keys.add(name)
                yield name, param
            # note
            # there is a bug in megatron GPTModel
            # decoder.layers[n].mlp.router.expert_bias" in GPTModel is not registered in named_parameter, but in
            # state_dict(). for now we patch it by adding those keys to extra_keys.
            extra_keys = [x for x in model.state_dict().keys() if "_extra_state" not in x and x not in existing_keys]
            for name in extra_keys:
                yield name, model.state_dict()[name].to(get_device_id())

    # we need first make all rank get full model information
    meta_info = []
    for scan_vpp_idx in range(vpp_size):
        existing_keys = set()
        model = unwrap_model(actor_module[scan_vpp_idx])
        for idx, (name, _) in enumerate(model.named_parameters()):
            existing_keys.add(name)
            meta_info.append((pp_rank, scan_vpp_idx, idx, name))
        extra_keys = [x for x in model.state_dict().keys() if "_extra_state" not in x and x not in existing_keys]
        for name in extra_keys:
            meta_info.append((pp_rank, scan_vpp_idx, idx, name))

    obj_spec_output = [None] * mpu.get_pipeline_model_parallel_world_size()
    torch.distributed.all_gather_object(
        object_list=obj_spec_output, obj=meta_info, group=mpu.get_pipeline_model_parallel_group()
    )
    '''
    展平: 把所有 rank 的 meta_info 合并为一个列表
    '''
    layer_list_meta = [item for sublist in obj_spec_output for item in sublist]

    gen_func = tensor_generator()

    # lazy load tensor for full model
    for cur_pp_rank, scan_vpp_idx, idx, name in layer_list_meta:
        if model_config.tie_word_embeddings and ("output_layers" in name):
            import warnings

            warnings.warn(
                "Current model sharing word and embedding weights, skip output layer conversion", stacklevel=2
            )
            continue

        if cur_pp_rank == pp_rank:
            try:
                cur_name, cur_tensor = next(gen_func)
            except StopIteration:
                cur_name, cur_tensor = None, None
            cur_name = normalize_model_name(name, cur_pp_rank, scan_vpp_idx, transformer_config)
        else:
            cur_tensor, cur_name = None, None

        # pp broadcast model tensor and name
        cur_name = broadcast_str_from_megatron_pp(cur_name)
        broad_pp_tensor = broadcast_from_megatron_pp(cur_tensor)

        # (xya): this is a hack to fix the name of the parameters
        while cur_name.startswith("module."):
            cur_name = cur_name[len("module.") :]

        # EP
        if ".mlp.experts.linear_fc" in cur_name and ep_size > 1:
            num_experts = weight_converter.mcore_config.num_moe_experts
            num_experts_per_rank = num_experts // ep_size
            infer_params = [torch.empty_like(broad_pp_tensor) for _ in range(ep_size)]
            torch.distributed.all_gather(infer_params, broad_pp_tensor, group=ep_group)

            name_prefix, local_expert_id = cur_name.split(".weight")
            local_expert_id = int(local_expert_id)
            global_expert_ids = [num_experts_per_rank * ep_rank + local_expert_id for ep_rank in range(ep_size)]
            global_expert_names = [f"{name_prefix}.weight{expert_id}" for expert_id in global_expert_ids]

            for name, param in zip(global_expert_names, infer_params, strict=True):
                if etp_size > 1:
                    # gather etp
                    etp_params = [torch.empty_like(param) for _ in range(etp_size)]
                    torch.distributed.all_gather(etp_params, param, group=etp_group)
                    params = etp_params
                else:
                    params = [param]

                merge_params = default_tp_concat_fn(
                    layer_name_mapping,
                    name,
                    broad_pp_tensor,
                    params,
                    model_config,
                    weight_converter.hf_config,
                    convert_qkv_gate_up_by_simple_split,
                )
                if not isinstance(merge_params, list):
                    merge_params = [merge_params]
                converted_names, converted_params = weight_converter.convert_param(name, merge_params)

                yield from zip(converted_names, [param.detach() for param in converted_params], strict=True)
            continue

        # tp all gather
        if tp_utils.is_tensor_parallel_param(broad_pp_tensor):
            # allocate a new tensor with proper size
            if all_gather_group_size <= 1:
                infer_params = [broad_pp_tensor]
            else:
                infer_params = [torch.empty_like(broad_pp_tensor) for _ in range(all_gather_group_size)]
                torch.distributed.all_gather(infer_params, broad_pp_tensor, group=mpu.get_tensor_model_parallel_group())
            infer_params = default_tp_concat_fn(
                layer_name_mapping,
                cur_name,
                broad_pp_tensor,
                infer_params,
                model_config,
                weight_converter.hf_config,
                convert_qkv_gate_up_by_simple_split,
            )
        else:
            infer_params = broad_pp_tensor

        if not isinstance(infer_params, list):
            infer_params = [infer_params]
        converted_names, converted_params = weight_converter.convert_param(cur_name, infer_params)

        yield from zip(converted_names, [param.detach() for param in converted_params], strict=True)
````

# 第二部分：本地张量生成器

````python
def tensor_generator():
    '''
    遍历vpp收集模型所有层的参数
    '''
    for scan_vpp_idx in range(vpp_size):
        existing_keys = set()
        model = unwrap_model(actor_module[scan_vpp_idx])  # 解包 DDP/FSDP 包装
        
        for name, param in model.named_parameters():
            existing_keys.add(name)
            yield name, param  # ← 产出 (层名, 参数张量)
        
        # 处理 Megatron 的 bug: 某些参数在 named_parameters() 中缺失
        # 但在 state_dict() 中存在 (如 router.expert_bias)
        extra_keys = [
            x for x in model.state_dict().keys() 
            if "_extra_state" not in x and x not in existing_keys
        ]
        for name in extra_keys:
            yield name, model.state_dict()[name].to(get_device_id())
````

## 对比：什么时候需要 param？

````python
# 场景 1: 只需要名称 → 用 _
for idx, (name, _) in enumerate(model.named_parameters()):
    meta_info.append((pp_rank, scan_vpp_idx, idx, name))

# 场景 2: 需要名称和张量 → 正常解包
for name, param in model.named_parameters():
    print(f"{name}: shape={param.shape}, dtype={param.dtype}")

# 场景 3: 只需要张量 → 用 _
for _, param in model.named_parameters():
    total_params += param.numel()

# 场景 4: 两者都需要 → 正常命名
for name, param in model.named_parameters():
    state_dict[name] = param.data.clone()
````

## 这一步解决了什么问题？

````python
问题: 每个 GPU 只知道自己 PP rank 的参数
      但生成完整模型需要知道所有层的参数名和顺序

解决: 通过 all_gather_object 让所有 GPU 获得完整的参数列表

示例:
  PP rank 0 的 meta_info: [(0, 0, 0, "layers.0.attn.q"), (0, 0, 1, "layers.0.attn.k"), ...]
  PP rank 1 的 meta_info: [(1, 0, 0, "layers.2.attn.q"), ...]
  
  all_gather 后:
  layer_list_meta = [
    (0, 0, 0, "layers.0.attn.q"),   # PP=0 的参数
    (0, 0, 1, "layers.0.attn.k"),
    ...
    (1, 0, 0, "layers.2.attn.q"),   # PP=1 的参数
    ...
    (2, 0, 0, "layers.4.attn.q"),   # PP=2 的参数
    ...
    (3, 0, 0, "layers.6.attn.q"),   # PP=3 的参数
  ]
````

````python
PP 组有 4 个 rank: [PP=0, PP=1, PP=2, PP=3]

执行前:
┌─────────────────────────────────────────┐
│  PP=0: obj_spec_output = [None, None, None, None]  │
│  PP=1: obj_spec_output = [None, None, None, None]  │
│  PP=2: obj_spec_output = [None, None, None, None]  │
│  PP=3: obj_spec_output = [None, None, None, None]  │
└─────────────────────────────────────────┘

all_gather_object 后:
┌─────────────────────────────────────────┐
│  PP=0: obj_spec_output = [meta_0, meta_1, meta_2, meta_3] │
│  PP=1: obj_spec_output = [meta_0, meta_1, meta_2, meta_3] │
│  PP=2: obj_spec_output = [meta_0, meta_1, meta_2, meta_3] │
│  PP=3: obj_spec_output = [meta_0, meta_1, meta_2, meta_3] │
│           ↑                              │
│           └── 每个 rank 都获得了所有 rank 的 meta_info │
└─────────────────────────────────────────┘
````

## 为什么用 all_gather_object 而不是 all_gather？

| 函数                  | 用途             | 数据类型                    |
| ------------------- | -------------- | ----------------------- |
| `all_gather`        | 收集张量           | `torch.Tensor`          |
| `all_gather_object` | 收集任意 Python 对象 | `list`, `dict`, `str` 等 |

<font color='red'>meta_info 是列表的列表，元素是元组 (int, int, int, str)，不是张量，所以必须用 all_gather_object。</font>

#### Megatron 命名的复杂性
##### 问题 1：<font color='red'>VPP 导致层号不连续</font>
- 问题：VPP 模型内部可能用局部层号（0, 1, 2...）而不是全局层号。

````python
无 VPP (PP=4, VPP=1):
  GPU 0 (PP=0): Layer 0-3
  GPU 1 (PP=1): Layer 4-7
  GPU 2 (PP=2): Layer 8-11
  GPU 3 (PP=3): Layer 12-15
  
  参数名: "decoder.layers.0.self_attention.query.weight"  ← 层号 = 物理层号

有 VPP (PP=4, VPP=2):
  GPU 0: [VPP0: Layer 0-1] + [VPP1: Layer 8-9]
  GPU 1: [VPP0: Layer 2-3] + [VPP1: Layer 10-11]
  ...
  
  参数名: "decoder.layers.0.self_attention.query.weight"  ← 还是 0?
          或者 "decoder.layers.8.self_attention.query.weight"?
````

##### 问题 2：不同 Megatron 版本的命名差异

````python
Megatron-LM 旧版:
  "language_model.encoder.layers.0.self_attention.query.weight"

Megatron-Core 新版:
  "decoder.layers.0.self_attention.query.weight"

自定义模型:
  "transformer.layers.0.attention.query.weight"
````

##### 问题 3：模块包装前缀

````python
DDP 包装: "module.decoder.layers.0..."
FSDP 包装: "_fsdp_wrapped_module.decoder.layers.0..."
原始:     "decoder.layers.0..."
````

#### normalize_model_name 的典型实现
基于 verl 的代码逻辑，该函数大致如下：

````python
def normalize_model_name(name, cur_pp_rank, scan_vpp_idx, transformer_config):
    """
    将 Megatron 的原始参数名标准化为统一格式
    """
    
    # 1. 去掉模块包装前缀 (module., _fsdp_wrapped_module. 等)
    '''
    在分布式训练中，模型可能被多层包装：
    
    model = DDP(model)           # 第 1 层 module.
    model = FSDP(model)          # 可能再加一层
    model = torch.compile(model) # 或其他包装
    
    被 DDP 再包装一次 (某些情况下)
    "module.module.decoder.layers.0.self_attention.query.weight"
    '''
    while name.startswith("module."):
        name = name[len("module."):]
    
    # 2. 计算全局层号 (处理 VPP 的层号偏移)
    '''
    num_layers	模型总层数（如 64 层 Transformer）
    pipeline_model_parallel_size	PP（流水线并行）大小
    virtual_pipeline_model_parallel_size	VPP（虚拟流水线）大小
    '''
    num_layers_per_pp_stage = transformer_config.num_layers // transformer_config.pipeline_model_parallel_size
    # 计算每个 VPP Stage 的层数
    num_layers_per_vpp_stage = num_layers_per_pp_stage // transformer_config.virtual_pipeline_model_parallel_size
    
    
    # VPP 层号偏移: 当前 PP rank 的起始层 + VPP stage 的偏移
    '''
    cur_pp_rank 	当前 PP Stage 的 rank	mpu.get_pipeline_model_parallel_rank()
    num_layers_per_pp_stage 	每个 PP Stage 的层数	num_layers // PP
    scan_vpp_idx 	当前 VPP Stage 的索引	0, 1, ..., vpp_size-1
    num_layers_per_vpp_stage	每个 VPP Stage 的层数	num_layers_per_pp // VPP
    '''
    pp_offset = cur_pp_rank * num_layers_per_pp_stage
    vpp_offset = scan_vpp_idx * num_layers_per_vpp_stage
    
    total_offset = pp_offset + vpp_offset  # 全局层号起始偏移
    
    # 3. 替换局部层号为全局层号
    # 例如: "decoder.layers.0." → "decoder.layers.8." (如果 offset=8)
    import re
    
    def replace_layer_number(match):
        local_layer_id = int(match.group(1))
        global_layer_id = local_layer_id + total_offset
        return f"layers.{global_layer_id}."
    
    # 匹配 "layers.数字." 的模式
    name = re.sub(r'layers\.(\d+)\.', replace_layer_number, name)
    
    # 4. 统一命名前缀 (把 language_model.encoder/decoder 统一为 decoder)
    name = name.replace("language_model.encoder.", "decoder.")
    name = name.replace("language_model.decoder.", "decoder.")
    
    # 5. 统一注意力命名
    name = name.replace("self_attention.", "self_attn.")
    name = name.replace("mlp.dense_h_to_4h", "mlp.gate_up_proj")
    name = name.replace("mlp.dense_4h_to_h", "mlp.down_proj")
    
    return name
````

## 5.3 PP 广播（<font color='red'>关键！</font>）
- 在 Megatron 中，模型被纵向切分成多个 pipeline stages（流水线阶段），每个 stage 运行在独立的 GPU 上。不同 stage 之间需要通信来传递激活值（activations）、梯度（gradients）或元数据。

````python
# 在 PP 组内广播参数名和张量

'''
为什么需要	因为不同 stage 的代码执行路径可能不同，需要确保所有 stage 对某个名称达成一致

典型场景：
- 在加载 checkpoint 或进行参数同步时，某个 stage 生成了一个张量名称，需要让所有 stage 知道这个名字
- 例如：PP stage 0 构造了 cur_name = "attention.query_key_value"，需要广播给 stage 1, 2, 3...
'''
cur_name = broadcast_str_from_megatron_pp(cur_name)      # 广播字符串
broad_pp_tensor = broadcast_from_megatron_pp(cur_tensor)  # 广播张量
````

### broadcast_str_from_megatron_pp

- 输入	cur_name — 一个 Python 字符串（如张量名称、键名等）
- 输出	所有 PP rank 上相同的字符串值

````python
# 伪代码示意
def broadcast_str_from_megatron_pp(string):
    # 获取当前进程在 pipeline parallel 组中的信息
    pp_group = get_pipeline_model_parallel_group()
    src_rank = get_pipeline_model_parallel_first_rank()  # 通常是 pp_rank=0
    
    # 使用 PyTorch distributed 广播字符串
    # 先将字符串编码为字节张量，广播后再解码
    if is_pipeline_first_stage():
        # 发送方：编码并发送
        byte_tensor = torch.ByteTensor(list(string.encode('utf-8')))
        broadcast(byte_tensor, src=src_rank, group=pp_group)
        return string
    else:
        # 接收方：接收并解码
        byte_tensor = torch.empty(max_length, dtype=torch.uint8)
        broadcast(byte_tensor, src=src_rank, group=pp_group)
        return byte_tensor.tobytes().decode('utf-8')
````

### broad_pp_tensor = broadcast_from_megatron_pp(cur_tensor)

- 输入	cur_tensor — 一个 torch.Tensor（可以是任何形状和 dtype）
- 输出	所有 PP rank 上完全相同的张量副本

#### 典型场景：
- 共享 Embedding：如果 embedding 层在 stage 0，但 stage N 也需要访问（如 tied embedding）
- 全局统计量：如梯度裁剪时的全局梯度范数
- Checkpoint 加载：某些元数据张量需要广播到所有 stages

````python
# 伪代码示意
def broadcast_from_megatron_pp(tensor):
    pp_group = get_pipeline_model_parallel_group()
    src_rank = get_pipeline_model_parallel_first_rank()
    
    if is_pipeline_first_stage():
        # 发送方：直接广播
        broadcast(tensor, src=src_rank, group=pp_group)
        return tensor
    else:
        # 接收方：创建相同形状的空张量，接收数据
        # 注意：需要先知道形状！可能也需要先广播 metadata
        output_tensor = torch.empty_like(tensor)  # 或从元数据构造
        broadcast(output_tensor, src=src_rank, group=pp_group)
        return output_tensor
````

### 两者的关键区别

| 对比项       | `broadcast_str_from_megatron_pp` | `broadcast_from_megatron_pp` |
| --------- | -------------------------------- | ---------------------------- |
| **传输内容**  | Python `str`                     | `torch.Tensor`               |
| **序列化方式** | 编码为字节张量 → 广播 → 解码                | 直接广播张量数据                     |
| **用途**    | 元数据、名称、配置字符串                     | 实际的数值张量、参数、激活值               |
| **性能**    | 极小（字符串很短）                        | 取决于张量大小，可能很大                 |
| **内存影响**  | 可忽略                              | 所有 stage 都会复制一份张量            |


### 为什么需要这两个函数？
- 在 Pipeline Parallelism 中，每个 stage 只拥有模型的一部分，但某些操作需要全局信息：

````python
问题场景：
- Stage 0 有 Embedding 层，生成了一个张量 "word_embeddings"
- Stage 3 需要知道这个张量的名字来构建 optimizer 分组
- Stage 3 也需要这个张量的实际值来做某些计算

解决方案：
1. 先用 broadcast_str 把名字 "word_embeddings" 告诉所有 stages
2. 再用 broadcast_tensor 把实际的 embedding 张量复制到所有 stages
````

### 这是什么操作？

````python
PP 组内广播:
  GPU 0 (PP=0): 持有 layers.0 的参数
  GPU 16 (PP=1): 持有 layers.2 的参数
  GPU 32 (PP=2): 持有 layers.4 的参数
  GPU 48 (PP=3): 持有 layers.6 的参数

处理 layers.0 时:
  - GPU 0: cur_tensor = 真实参数
  - GPU 16/32/48: cur_tensor = None
  
  broadcast_from_megatron_pp:
  - GPU 0 发送参数给 GPU 16, 32, 48
  - 最终所有 GPU 都有 layers.0 的完整参数 (但仍按 TP 分片)
````

###  torch.distributed.all_gather(infer_params, broad_pp_tensor, group=ep_group)

| 参数                | 含义                                              |
| ----------------- | ----------------------------------------------- |
| `infer_params`    | **输出列表**，接收来自所有 EP rank 的聚合结果                   |
| `broad_pp_tensor` | **输入张量**，当前 EP rank 要贡献给聚合的数据                   |
| `group=ep_group`  | 通信组，限定在 **Expert Parallel（EP）组** 内进行 all-gather |


##### 关键点：<font color='red'>每个 EP rank 只持有 部分 expert 的参数，但推理/训练时需要 所有 expert 的参数。</font>

#### all_gather 详解
操作示意图

````python
EP Rank 0:  [A0] ──┐
EP Rank 1:  [A1] ──┼──→ All-Gather →  [A0, A1, A2, A3]  (每个 rank 都拿到全部)
EP Rank 2:  [A2] ──┤         ↑
EP Rank 3:  [A3] ──┘         └ 每个 rank 输出相同
````

#### 为什么需要 broad_pp_tensor 作为输入？
- 注意这里的输入是 broad_pp_tensor —— <font color='red'>这是 经过 Pipeline Parallel 广播后的张量。</font>
这暗示了一个 多级并行的组合架构：

````python
┌─────────────────────────────────────────────┐
│           3D 并行：TP × PP × EP              │
├─────────────────────────────────────────────┤
│  Tensor Parallel (TP): 切分 attention 计算   │
│  Pipeline Parallel (PP): 切分 layer 深度      │
│  Expert Parallel (EP): 切分 MoE expert       │
└─────────────────────────────────────────────┘

数据流：
1. 先在 PP 组内广播得到 broad_pp_tensor (跨 stage)
   └── 确保当前 stage 拥有完整的 expert 参数块

2. 再在 EP 组内 all-gather (跨 expert)
   └── 收集所有 expert 的完整参数
````

##### <font color='red'>典型场景：MoE 推理时，需要把分散在不同 GPU 上的 expert 参数聚合起来。</font>



````python
import torch
import torch.distributed as dist
from megatron.core import parallel_state

# 获取 Expert Parallel 通信组
ep_group = parallel_state.get_expert_model_parallel_group()

# 假设当前 rank 只持有部分 expert 的参数
# 先通过 PP 广播确保当前 stage 有完整的本地 expert 块
local_expert_param = get_local_expert_param()  # shape: [E_local, D, D]
broad_pp_tensor = broadcast_from_megatron_pp(local_expert_param)  
# ↑ 如果 expert 在 PP 的某个 stage 上，先广播到该 stage 的所有 TP rank

# 在 EP 组内 all-gather：收集所有 expert 的参数
# infer_params 是一个列表，长度 = EP world size
ep_world_size = dist.get_world_size(group=ep_group)
infer_params = [torch.empty_like(broad_pp_tensor) for _ in range(ep_world_size)]

# 执行 all-gather：每个 rank 贡献自己的 expert，收集所有人的
dist.all_gather(infer_params, broad_pp_tensor, group=ep_group)

# 结果：infer_params[i] 是 EP rank i 的 expert 参数
# 现在当前 rank 拥有了所有 expert 的完整参数！
all_expert_params = torch.cat(infer_params, dim=0)  # [E_total, D, D]
````

| `all_gather_into_tensor` | `Tensor`       | 直接拼接成连续内存，更高效     |


````python
# 方式一：all_gather（旧 API，输出列表）
dist.all_gather(infer_params, broad_pp_tensor, group=ep_group)

# 方式二：all_gather_into_tensor（新 API，输出连续张量）
output = torch.empty(ep_size * local_size, ...)
dist.all_gather_into_tensor(output, broad_pp_tensor, group=ep_group)
````

### moe混合并行的实际应用
- <font color='red'>在实际的大规模 MoE 模型训练中，PP 和 EP 通常会结合使用。</font>
  - <font color='red'>PP 负责解决模型深度带来的显存和计算瓶颈，将模型切分为多个 Stage；</font>
  - <font color='blue'>而在每个 Stage 内部，如果 MoE 层的 Expert 依然过大，则通过 EP 将 Expert 进一步切分到多张卡上。</font>
  - <font color='red'>此外，针对 Attention 等稠密（Dense）层，可能还会同时结合张量并行（TP）来进一步降低单卡显存压力。</font>

#### 总结而言，
- PP 决定了“哪些 Layer（包含 Expert）”在哪张卡上计算，
- 而 EP 决定了“这些 Layer 里的 Expert”如何跨卡分布

#### 场景设定

````python
模型：GPT-MoE，共 16 层
并行配置：
  - PP = 4（4 个 Pipeline Stage）
  - 每层 MoE 有 8 个 Expert
  - EP = 4（Expert Parallel，4 张卡分担 8 个 expert）
  - TP = 2（Tensor Parallel，每个 expert 的权重再切一半）

总 GPU 数 = PP × EP × TP = 4 × 4 × 2 = 32 张卡
````

##### PP 先切 Layer
- 只有 PP Stage 2 需要处理 MoE，我们聚焦这里。

````python
PP Stage 0 (GPU 0-7):   Layers 0-3    ← 纯 Dense，无 MoE
PP Stage 1 (GPU 8-15):  Layers 4-7    ← 纯 Dense，无 MoE  
PP Stage 2 (GPU 16-23): Layers 8-11   ← Layer 10 是 MoE！
PP Stage 3 (GPU 24-31): Layers 12-15  ← 纯 Dense，无 MoE
````

##### Stage 2 内部：8 个 Expert 用 EP 切分

````python
PP Stage 2 占用的 GPU：16, 17, 18, 19, 20, 21, 22, 23（共 8 张）

但 EP = 4，意味着 EP group 只有 4 个 rank
TP = 2，每个 expert 拆成 2 份

所以 Stage 2 的 8 张卡组织为：
┌─────────────────────────────────────────────────────────┐
│              Stage 2 的 8 张卡 = 4 EP × 2 TP              │
├─────────────────────────────────────────────────────────┤
│                                                         │
│   EP Group（4 个逻辑 rank，每个 rank 有 2 张 TP 卡）       │
│                                                         │
│   EP-rank-0:  GPU 16 (TP-0)  +  GPU 17 (TP-1)           │
│                     └──────┬──────┘                          │
│              共同持有 Expert 0, 1（各持一半权重）         │
│                                                         │
│   EP-rank-1:  GPU 18 (TP-0)  +  GPU 19 (TP-1)           │
│                     └──────┬──────┘                          │
│              共同持有 Expert 2, 3                     │
│                                                         │
│   EP-rank-2:  GPU 20 (TP-0)  +  GPU 21 (TP-1)           │
│              共同持有 Expert 4, 5                     │
│                                                         │
│   EP-rank-3:  GPU 22 (TP-0)  +  GPU 23 (TP-1)           │
│              共同持有 Expert 6, 7                     │
│                                                         │
└─────────────────────────────────────────────────────────┘
````

#### 为什么需要 EP？Expert 有多大？
假设一个 Expert 是标准 FFN：
- EP 的作用：<font color='red'>把 8 个 expert 分散到 4 个 EP rank，每个 rank 只存 2 个 expert，显存压力降低。</font>

#### 推理时的通信流程
##### 前向传播：需要所有 Expert 参与路由

````python
输入 token 经过 router，决定走 top-2 expert
比如 token 0 需要 Expert 3 和 Expert 7

但 Expert 3 在 EP-rank-1，Expert 7 在 EP-rank-3
当前 GPU 只有 Expert 0,1（假设是 EP-rank-0）

怎么办？All-Gather！
````

````python
Expert 结构：
  Linear 1: [hidden_size=4096] → [ffn_dim=16384]
  Linear 2: [16384] → [4096]

参数量 = 4096×16384 + 16384×4096 = 134M 参数
       = 268MB (fp16/bf16)

8 个 Expert = 2.1GB 每层

如果不用 EP，Stage 2 的某张卡要存 8×268MB = 2.1GB 的 expert 参数
加上 TP 切分后，每张卡存 4×268MB/2 = 536MB（合理）
````

````python
# 在 Stage 2 的每个 GPU 上执行：

# Step 1: 本地 TP all-gather，还原完整 expert
# GPU 16 和 17 各自只有 Expert 0 的一半权重
tp_group = [GPU16, GPU17]
local_shard = expert_0_weight[:, 0::2]  # 列切分，我有一半列
full_expert_0 = all_gather(local_shard, group=tp_group)  # 还原 [4096, 16384]

# Step 2: 在 EP group 内 all-gather 所有 expert 的完整参数
ep_group = [EP0, EP1, EP2, EP3]  # 每个 EP rank 派一个代表（通常是 TP-rank-0）

# 当前 EP-rank-0 持有 expert_0, expert_1（已 TP 聚合）
my_experts = [full_expert_0, full_expert_1]  # 2 个 expert

# All-gather：每个 EP rank 贡献自己的 2 个 expert
infer_params = [None] * 4  # 4 个 EP rank
all_gather(infer_params, my_experts, group=ep_group)

# 结果：infer_params[0] = [expert_0, expert_1]  (来自 EP-rank-0)
#       infer_params[1] = [expert_2, expert_3]  (来自 EP-rank-1)
#       infer_params[2] = [expert_4, expert_5]  (来自 EP-rank-2)
#       infer_params[3] = [expert_6, expert_7]  (来自 EP-rank-3)

# 现在当前 GPU 有了所有 8 个 expert 的完整参数！
all_experts = flatten(infer_params)  # [expert_0, ..., expert_7]
````

#####  图示：通信发生在哪

````python
Stage 2 内部（8 张卡）：

GPU16 GPU17 GPU18 GPU19 GPU20 GPU21 GPU22 GPU23
  │     │     │     │     │     │     │     │
  └──TP─┘     └──TP─┘     └──TP─┘     └──TP─┘
   EP0         EP1         EP2         EP3
    │           │           │           │
    └───────────┴───────────┴───────────┘
              EP All-Gather
              (infer_params)
                    ↓
        每张卡都拿到 8 个 expert 的完整参数
````

##### <font color='red'> 关键问题：为什么推理时要 all-gather 所有 expert？</font>
- <font color='red'>训练时不需要：</font>每个 token 只路由到本地持有的 expert，用 all_to_all 交换 token 即可。
- <font color='blue'>推理时可能需要：</font>
  - 动态路由：推理 batch 小，可能某个 expert 被频繁选中，需要本地有完整参数
  - 量化/压缩：需要对完整 expert 做 INT8/INT4 量化
  - 专家聚合：某些优化需要合并 expert 参数
  - 缓存预热：预先把所有 expert 加载到显存，避免推理时反复通信

#### 一句话总结
- PP 决定 "哪几层在哪个 stage"，Stage 2 有幸（或不幸）拿到了 MoE 层。
- EP 决定 "这个 stage 里，8 个 expert 怎么分给 4 组 GPU"，每组只存 2 个。
- TP 决定 "每个 expert 的权重矩阵怎么再切开"，两张卡各算一半。
推理时通过 all_gather 把分散的 expert 拼回来，才能做路由计算。

### MoE 处理的复杂性：

````python
DeepSeek-V3 (671B) 的 MoE 结构:
- 256 个专家，EP=8
- 每个 GPU 负责 256/8 = 32 个专家
- 但 vLLM 需要看到所有 256 个专家

EP all_gather: 把 8 个 GPU 上的专家聚合成完整 256 个
ETP all_gather: 如果专家内部还有 TP，再聚合
````

## 5.6 普通 TP 参数处理（分支 B）


````python
# 判断是否是 TP 切分的参数
if tp_utils.is_tensor_parallel_param(broad_pp_tensor):
    
    if all_gather_group_size <= 1:
        infer_params = [broad_pp_tensor]  # 无 TP，无需聚合
    else:
        # 在 TP 组内 all_gather
        infer_params = [torch.empty_like(broad_pp_tensor) for _ in range(all_gather_group_size)]
        torch.distributed.all_gather(
            infer_params, 
            broad_pp_tensor, 
            group=mpu.get_tensor_model_parallel_group()
        )
    
    # TP 拼接 + 格式转换
    infer_params = default_tp_concat_fn(
        layer_name_mapping,
        cur_name,
        broad_pp_tensor,
        infer_params,
        model_config,
        weight_converter.hf_config,
        convert_qkv_gate_up_by_simple_split,
    )
else:
    infer_params = broad_pp_tensor  # 非 TP 参数，直接使用
````

### TP all_gather 示例：

````python
TP=8 组内 all_gather:
  GPU 0 持有: Q 投影的 1/8 列分片 [hidden, hidden/8]
  GPU 1 持有: Q 投影的 1/8 列分片 [hidden, hidden/8]
  ...
  GPU 7 持有: Q 投影的 1/8 列分片 [hidden, hidden/8]

  all_gather 后:
  infer_params = [
    [hidden, hidden/8],  # GPU 0
    [hidden, hidden/8],  # GPU 1
    ...
    [hidden, hidden/8],  # GPU 7
  ]
  
  default_tp_concat_fn 拼接后:
  [hidden, hidden]  # 完整 Q 投影
````

## 5.7 最终转换与产出

````python
if not isinstance(infer_params, list):
    infer_params = [infer_params]

# 调用权重转换器: Megatron 格式 → vLLM 格式
converted_names, converted_params = weight_converter.convert_param(cur_name, infer_params)

# 逐张量产出 (使用 generator 惰性求值)
yield from zip(
    converted_names, 
    [param.detach() for param in converted_params], 
    strict=True
)
````

# 完整数据流总结

````python
┌─────────────────────────────────────────────────────────────────────────┐
│                    per_tensor_generator 数据流                           │
├─────────────────────────────────────────────────────────────────────────┤
│                                                                         │
│  输入: actor_module (Megatron 模型, 按 TP×PP×EP 分片)                    │
│                                                                         │
│  Step 1: 收集本地参数元信息                                              │
│     └── meta_info = [(pp_rank, vpp_idx, idx, name), ...]               │
│                                                                         │
│  Step 2: PP 组内 all_gather_object 元信息                                │
│     └── layer_list_meta = 所有 PP rank 的完整参数列表                     │
│                                                                         │
│  Step 3: 遍历 layer_list_meta (按顺序处理每一层)                         │
│     ├── 如果 cur_pp_rank == 本地 pp_rank: 从本地生成器取参数             │
│     ├── 否则: cur_tensor = None                                         │
│     ├── PP 广播: broadcast_from_megatron_pp()                          │
│     │   └── 所有 GPU 获得该层的参数 (仍按 TP 分片)                       │
│     ├── 名称规范化: 去掉 module. 前缀                                    │
│     ├── 分支判断:                                                       │
│     │   ├── MoE 参数 → EP all_gather → ETP all_gather → 转换 → yield    │
│     │   └── 普通参数 → TP all_gather → 转换 → yield                     │
│     └── 产出: (vllm_name, vllm_shard)                                   │
│                                                                         │
│  输出: Generator of (vllm_layer_name, vllm_parameter_tensor)           │
│        逐张量产出，内存友好                                               │
└─────────────────────────────────────────────────────────────────────────┘
````

# 关键设计亮点

| 设计                          | 作用                                  |
| --------------------------- | ----------------------------------- |
| **Generator 模式**            | 惰性求值，避免一次性加载全部模型权重到内存               |
| **all\_gather\_object**     | 用 PyTorch 对象序列化传递元信息，支持任意 Python 对象 |
| **PP 广播**                   | 在流水线维度聚合层，每个 GPU 逐步获得完整模型视图         |
| **TP all\_gather**          | 在张量并行维度拼接列/行分片，恢复完整张量               |
| **EP/ETP 处理**               | 支持 MoE 模型的专家并行转换                    |
| **default\_tp\_concat\_fn** | 处理 QKV 重排、GateUp 合并等复杂格式转换          |


# 与 MegatronVLLMShardingManager.__enter__ 的配合

````python
def __enter__(self):
    # 创建生成器
    per_tensor_param = per_tensor_generator(
        self.actor_module,
        self.model_config,
        self.weight_converter,
        self.transformer_config,
        self.layer_name_mapping,
    )
    
    # 逐张量加载到 vLLM (惰性消费生成器)
    model = self.model_runner.model
    loaded_params = model.load_weights(per_tensor_param)  # ← 内部不断 next()
    
    return self
````

## Generator 的优势：
- 不需要同时持有 Megatron 和 vLLM 两份完整模型
- load_weights 消费一个张量，生成器就释放一个张量
- 峰值内存 = 1 个张量大小，而非整个模型大小
