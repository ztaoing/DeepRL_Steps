# 完整调用流程
## Phase 1: 初始化（Initialization）

```python
# ray_trainer.py
class RayPPOTrainer:
    def __init__(self, config, tokenizer, role_worker_mapping):
        # 1. 初始化各个 Worker
        self.actor_rollout_wg = ActorRolloutWorker(...)  # 生成回答
        self.ref_policy_wg = RefPolicyWorker(...)        # 参考模型
        self.reward_fn = RewardFunction(...)             # 奖励函数
        
        # 2. 初始化 Actor Worker（用于策略更新）
        self.actor_wg = DataParallelPPOActor(
            config=actor_config,
            actor_module=actor_model,
            actor_optimizer=optimizer
        )
```

## Phase 2: 训练循环（Training Loop）

```python
# ray_trainer.py - fit() 方法
def fit(self):
    for epoch in range(self.config.epochs):
        for batch in self.train_dataloader:
            # ========== Step 1: 生成回答（Rollout）==========
            batch = self._generate_responses(batch)
            
            # ========== Step 2: 计算奖励（Reward）==========
            batch = self._compute_rewards(batch)
            
            # ========== Step 3: 计算优势（Advantage）==========
            batch = self._compute_advantages(batch)
            
            # ========== Step 4: 策略更新（Policy Update）==========
            metrics = self._update_policy(batch)
            
            # ========== Step 5: 记录日志 ==========
            self._log_metrics(metrics)
```

## Phase 3: 详细调用链
### Step 1: 生成回答（Rollout）


```python
# ray_trainer.py
def _generate_responses(self, batch):
    # 调用 ActorRollout Worker 生成回答
    batch = self.actor_rollout_wg.generate_sequences(batch)
    
    # batch 现在包含：
    # - input_ids: (bsz, prompt_len + response_len)
    # - responses: (bsz, response_len)
    # - response_mask: (bsz, response_len)
    # - old_log_probs: (bsz, response_len)  ← 生成时的 log_prob
    
    return batch
```

#### ActorRollout Worker 内部：

```python
# 调用模型生成
outputs = self.actor_module.generate(
    input_ids=batch.input_ids,
    max_new_tokens=max_length,
    temperature=temperature,
    do_sample=True,
    return_dict_in_generate=True,
    output_scores=True,  # 返回每个 token 的分数
)

# 提取生成的 token 和 log_prob
responses = outputs.sequences[:, prompt_len:]
old_log_probs = self._extract_log_probs(outputs.scores, responses)
```

### Step 2: 计算奖励（Reward）

```python
# ray_trainer.py
def _compute_rewards(self, batch):
    # 方法1：使用 Reward Model
    if self.use_reward_model:
        rewards = self.reward_model_wg.compute_reward(batch)
    
    # 方法2：使用可验证奖励函数（如 RLVR）
    else:
        rewards = self.reward_fn(batch.responses)  # 例如：检查答案是否正确
    
    # rewards: (bsz,) 每个样本的奖励
    batch.batch["token_level_scores"] = rewards  # 广播到每个 token
    
    return batch
```

### Step 3: 计算优势（Advantage）


```python
# ray_trainer.py
def _compute_advantages(self, batch):
    # 使用 GAE 或其他方法计算 advantage
    batch = compute_advantage(
        batch,
        adv_estimator=self.config.adv_estimator,  # "gae", "grpo", "remax"
        gamma=self.config.gamma,
        lam=self.config.lam,
    )
    
    # batch 现在包含：
    # - advantages: (bsz, response_len) 每个 token 的 advantage
    # - returns: (bsz, response_len) 累积回报
    
    return batch
```

#### GAE 计算：

```python
def compute_gae(rewards, values, gamma=0.99, lam=0.95):
    advantages = []
    gae = 0
    
    for t in reversed(range(len(rewards))):
        if t == len(rewards) - 1:
            next_value = 0
        else:
            next_value = values[t + 1]
        
        delta = rewards[t] + gamma * next_value - values[t]
        gae = delta + gamma * lam * gae
        advantages.insert(0, gae)
    
    return advantages
```

### Step 4: 策略更新（Policy Update）⭐ 核心

```python
# ray_trainer.py
def _update_policy(self, batch):
    # 调用 Actor Worker 更新策略
    # 这就是你提供的代码中的 update_policy 方法
    
    metrics = self.actor_wg.update_policy(batch)
    
    return metrics
```

#### 进入 DataParallelPPOActor.update_policy()（你提供的代码）：

```python
# dp_actor.py
@GPUMemoryLogger(role="dp actor", logger=logger)
def update_policy(self, data: DataProto):
    # ========== 4.1 数据准备 ==========
    select_keys = ["responses", "response_mask", "input_ids", 
                   "attention_mask", "position_ids", 
                   "old_log_probs", "advantages"]
    data = data.select(batch_keys=select_keys)
    
    # ========== 4.2 分批次 ==========
    mini_batches = data.split(self.config.ppo_mini_batch_size)
    
    # ========== 4.3 多 epoch 训练 ==========
    for epoch in range(self.config.ppo_epochs):
        for mini_batch in mini_batches:
            
            # 进一步拆分为 micro_batch（适应 GPU 显存）
            micro_batches = mini_batch.split(
                self.config.ppo_micro_batch_size_per_gpu
            )
            
            for micro_batch in micro_batches:
                # ========== 4.4 前向传播 ==========
                # 关键：决定是否计算熵
                calculate_entropy = False
                entropy_top_ratio = self.config.get('entropy_top_ratio', None)
                
                if entropy_coeff != 0 or entropy_top_ratio is not None:
                    calculate_entropy = True
                
                # 调用 _forward_micro_batch
                entropy, log_prob = self._forward_micro_batch(
                    model_inputs, 
                    temperature=temperature, 
                    calculate_entropy=calculate_entropy
                )
                # entropy: (bsz, response_len) 每个 token 的熵
                # log_prob: (bsz, response_len) 当前策略的 log_prob
                
                # ========== 4.5 高熵 Token 筛选（论文核心）==========
                entropy_top_mask = None
                if entropy_top_ratio is not None:
                    entropy_top_mask = get_global_entropy_top_mask(
                        entropy=entropy,
                        response_mask=response_mask,
                        top_ratio=entropy_top_ratio  # 如 0.2
                    )
                
                # ========== 4.6 计算 PPO Loss ==========
                pg_loss, pg_clipfrac, ppo_kl, pg_clipfrac_lower = policy_loss_fn(
                    old_log_prob=old_log_prob,
                    log_prob=log_prob,
                    advantages=advantages,
                    response_mask=response_mask,
                    entropy_top_mask=entropy_top_mask,  # 传入高熵 mask
                    ...
                )
                
                # ========== 4.7 熵奖励 ==========
                if entropy_coeff != 0:
                    entropy_loss = agg_loss(
                        loss_mat=entropy,
                        loss_mask=response_mask,
                        loss_agg_mode=loss_agg_mode
                    )
                    policy_loss = pg_loss - entropy_loss * entropy_coeff
                else:
                    policy_loss = pg_loss
                
                # ========== 4.8 KL 惩罚 ==========
                if self.config.use_kl_loss:
                    kl_loss = compute_kl_loss(log_prob, ref_log_prob)
                    policy_loss = policy_loss + kl_loss * kl_loss_coef
                
                # ========== 4.9 反向传播 ==========
                loss = policy_loss * loss_scale_factor
                loss.backward()
            
            # ========== 4.10 优化器步骤 ==========
            grad_norm = self._optimizer_step()
    
    return metrics
```

### Phase 4: 前向传播细节

```python
# dp_actor.py
def _forward_micro_batch(self, model_inputs, temperature, calculate_entropy=False):
    # 调用 actor 模型
    outputs = self.actor_module(
        input_ids=model_inputs["input_ids"],
        attention_mask=model_inputs["attention_mask"],
        position_ids=model_inputs["position_ids"],
    )
    
    # logits: (bsz, seq_len, vocab_size)
    logits = outputs.logits
    
    # 提取 response 部分的 logits
    response_logits = logits[:, prompt_len-1:-1, :]  # 偏移一位对齐
    
    # 计算 log_prob
    log_probs = F.log_softmax(response_logits / temperature, dim=-1)
    
    # 提取实际生成 token 的 log_prob
    response_log_probs = torch.gather(
        log_probs, 
        dim=-1, 
        index=responses.unsqueeze(-1)
    ).squeeze(-1)
    
    # 计算熵（如果需要）
    if calculate_entropy:
        probs = torch.softmax(response_logits / temperature, dim=-1)
        entropy = -torch.sum(probs * torch.log(probs + 1e-10), dim=-1)
    else:
        entropy = None
    
    return entropy, response_log_probs
```

# 关键数据流

```python
DataProto 数据结构在训练中的流转：

初始 batch:
├── input_ids: (bsz, prompt_len)
└── attention_mask: (bsz, prompt_len)

        ↓ Rollout (生成回答)

batch:
├── input_ids: (bsz, prompt_len + response_len)
├── attention_mask: (bsz, prompt_len + response_len)
├── responses: (bsz, response_len)
├── response_mask: (bsz, response_len)
└── old_log_probs: (bsz, response_len)  ← 旧策略的 log_prob

        ↓ Reward (计算奖励)

batch:
├── ... (同上)
└── token_level_scores: (bsz, response_len)  ← 每个 token 的奖励

        ↓ Advantage (计算优势)

batch:
├── ... (同上)
├── advantages: (bsz, response_len)  ← 每个 token 的 advantage
└── returns: (bsz, response_len)     ← 累积回报

        ↓ Policy Update (策略更新)

Actor Worker 内部:
├── entropy: (bsz, response_len)      ← 每个 token 的熵
├── log_prob: (bsz, response_len)     ← 当前策略的 log_prob
├── entropy_top_mask: (bsz, response_len)  ← 高熵 token mask
└── pg_loss: scalar                    ← 策略梯度损失
```

# 配置参数对应关系

| 配置参数                           | 代码变量                           | 论文对应        | 作用                   |
| ------------------------------ | ------------------------------ | ----------- | -------------------- |
| `entropy_top_ratio`            | `entropy_top_ratio`            | **ρ (rho)** | 保留的高熵 token 比例       |
| `entropy_coeff`                | `entropy_coeff`                | -           | 熵奖励系数                |
| `ppo_epochs`                   | `ppo_epochs`                   | -           | 每个 batch 的 PPO 更新轮数  |
| `ppo_mini_batch_size`          | `ppo_mini_batch_size`          | -           | mini-batch 大小        |
| `ppo_micro_batch_size_per_gpu` | `ppo_micro_batch_size_per_gpu` | -           | 每个 GPU 的 micro-batch |
| `temperature`                  | `temperature`                  | **T**       | 解码温度                 |
| `loss_agg_mode`                | `loss_agg_mode`                | -           | loss 聚合方式            |


# 总结
## verl 中 RL 训练的完整调用流程：




```python
┌─────────────────────────────────────────────────────────────┐
│  1. 初始化：创建 ActorRollout, RefPolicy, Reward, Actor    │
│                    Workers                                   │
├─────────────────────────────────────────────────────────────┤
│  2. 外层循环：遍历训练数据（epoch / batch）                   │
├─────────────────────────────────────────────────────────────┤
│  3. Rollout：ActorRollout Worker 生成回答                    │
│     └── 得到 responses, old_log_probs                       │
├─────────────────────────────────────────────────────────────┤
│  4. Reward：计算每个样本的奖励                               │
│     └── 得到 token_level_scores                             │
├─────────────────────────────────────────────────────────────┤
│  5. Advantage：使用 GAE/GRPO 计算每个 token 的 advantage     │
│     └── 得到 advantages                                     │
├─────────────────────────────────────────────────────────────┤
│  6. Policy Update：Actor Worker 更新策略                   │
│     ├── 6.1 前向传播：计算 log_prob + entropy               │
│     ├── 6.2 高熵筛选：get_global_entropy_top_mask()          │
│     ├── 6.3 计算 PPO loss（只更新高熵 token）               │
│     ├── 6.4 加熵奖励 + KL 惩罚                              │
│     └── 6.5 反向传播 + 优化器步骤                           │
├─────────────────────────────────────────────────────────────┤
│  7. 日志记录：metrics → wandb / tensorboard                 │
└─────────────────────────────────────────────────────────────┘
```
