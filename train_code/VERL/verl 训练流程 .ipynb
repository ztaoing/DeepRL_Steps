```python
┌─────────────────────────────────────────────────────────────┐
│                      verl 训练流程                            │
├─────────────────────────────────────────────────────────────┤
│                                                             │
│  1. 提示词 (Prompt)                                          │
│     ↓                                                       │
│  2. 生成阶段 (Rollout)                                       │
│     ├── 旧策略模型 π_old 生成 G 个回答（如 G=8）              │
│     ├── 每个回答有：文本 + 对数概率 log_prob                 │
│     └── 输出: {response_1, ..., response_G}                  │
│                                                             │
│  3. 奖励计算 (Reward)                                        │
│     ├── 规则奖励: 答案正确性、格式匹配、长度惩罚              │
│     ├── 模型奖励: RM 模型打分（可选）                         │
│     └── 输出: {r_1, ..., r_G}                                │
│                                                             │
│  4. 优势估计 (Advantage)  ← 关键比较步骤！                    │
│     ├── GRPO: 组内相对优势 = (r_i - mean(r)) / std(r)        │
│     └── PPO: GAE 广义优势估计                                 │
│                                                             │
│  5. 策略更新 (Update)                                        │
│     ├── 新策略 π_new 重新计算 log_prob                        │
│     ├── 计算比率 ratio = π_new / π_old                       │
│     ├── 裁剪目标: clip(ratio, 1-ε, 1+ε) × advantage         │
│     └── 最大化奖励同时保持策略稳定                             │
│                                                             │
└─────────────────────────────────────────────────────────────┘
```
