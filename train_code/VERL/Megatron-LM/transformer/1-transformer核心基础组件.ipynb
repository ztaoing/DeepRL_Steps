# 核心基础组件（积木块）
无论是 Encoder 还是 Decoder，它们内部都包含以下三个最核心的模块：

##  自注意力机制（Self-Attention / Multi-Head Attention）
- 作用：这是 Transformer 的灵魂。
  - 它允许模型在处理某个词时，同时“看”到输入序列中的所有其他词，<font color='red'>从而捕捉词与词之间的依赖关系（比如代词指代、长距离依赖等）。</font>
- 特点：通常使用“多头”（Multi-Head），
  - <font color='blue'>即把注意力机制并行运行多次，让模型能从不同的语义角度去理解文本。</font>
  
##  前馈神经网络（Feed-Forward Network, FFN）
- 作用：对经过注意力机制处理后的特征进行非线性的变换和提取。
- 特点：它是一个全连接的网络，<font color='red'>对序列中的每一个 token 独立进行相同的计算（也就是我们刚才聊到的 token-wise 操作）。</font>

##  残差连接与层归一化（Add & Norm）
- 作用：每个子层（Attention 或 FFN）的周围都会包裹着这两个结构。
- 特点：
  - 残差连接（Residual Connection）：将输入直接加到输出上（ x+Sublayer(x) ），防止网络太深导致梯度消失。
  - 层归一化（Layer Normalization）：稳定训练过程，加速收敛。
