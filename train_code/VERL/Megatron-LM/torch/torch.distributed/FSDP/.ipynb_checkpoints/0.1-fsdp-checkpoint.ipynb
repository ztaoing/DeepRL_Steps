# torch.distributed.fsdp.FullyShardedDataParallel（FSDP）
- 是 PyTorch 官方提供的全分片数据并行方案，是 DDP 的下一代替代者，专门用于超大模型训练（数百亿至数万亿参数）。
- 与 DDP 的"每 GPU 一个完整模型副本"不同，FSDP 将模型参数、梯度和优化器状态分片到所有 GPU，大幅降低显存占用。

## 一、核心原理对比
### DDP vs FSDP

```python
┌─────────────────────────────────────────────────────────────┐
│                    DDP（DistributedDataParallel）             │
│                                                              │
│   GPU 0: [完整模型] + [完整梯度] + [完整优化器状态]            │
│   GPU 1: [完整模型] + [完整梯度] + [完整优化器状态]            │
│   GPU 2: [完整模型] + [完整梯度] + [完整优化器状态]            │
│   GPU 3: [完整模型] + [完整梯度] + [完整优化器状态]            │
│                                                              │
│   显存 = 4 × 模型大小                                        │
│   可训练模型大小受限于单卡显存                                │
└─────────────────────────────────────────────────────────────┘

┌─────────────────────────────────────────────────────────────┐
│                    FSDP（FullyShardedDataParallel）           │
│                                                              │
│   GPU 0: [1/4 参数分片] + [1/4 梯度分片] + [1/4 优化器状态]  │
│   GPU 1: [1/4 参数分片] + [1/4 梯度分片] + [1/4 优化器状态]  │
│   GPU 2: [1/4 参数分片] + [1/4 梯度分片] + [1/4 优化器状态]  │
│   GPU 3: [1/4 参数分片] + [1/4 梯度分片] + [1/4 优化器状态]  │
│                                                              │
│   显存 ≈ 1 × 模型大小 / 4（分片后）+ 通信开销                │
│   可训练模型大小 = 总显存之和（线性扩展）                      │
└─────────────────────────────────────────────────────────────┘
```

### FSDP 的关键设计

| 组件        | DDP           | FSDP                   |
| --------- | ------------- | ---------------------- |
| **参数存储**  | 每卡完整副本        | 分片到所有卡                 |
| **梯度存储**  | 每卡完整副本        | 分片到所有卡                 |
| **优化器状态** | 每卡完整副本        | 分片到所有卡（配合 AdamW 8-bit） |
| **前向传播**  | 直接用本地参数       | All-Gather 收集所需参数      |
| **反向传播**  | All-Reduce 梯度 | Reduce-Scatter 分片梯度    |
| **显存效率**  | 低             | 高（可达 4× 提升）            |
| **通信量**   | 小（仅梯度同步）      | 大（参数收集 + 梯度分片）         |


## 核心机制详解
### 参数分片与 All-Gather

```python
# FSDP 前向传播时的参数管理

class FSDP(nn.Module):
    def forward(self, *args, **kwargs):
        # 1. All-Gather：从所有 GPU 收集完整参数
        #    之前每卡只存 1/N，现在临时拼出完整参数
        self._all_gather_params()
        
        # 2. 用完整参数做前向传播（与普通模型一样）
        output = self.module(*args, **kwargs)
        
        # 3. 释放完整参数，只保留分片（可选，看配置）
        self._free_full_params()
        
        return output
```

```python
All-Gather 过程（4 GPU）：

GPU 0: [P0] ──┐
GPU 1: [P1] ──┼── All-Gather ──→ GPU 0~3 都有 [P0|P1|P2|P3]
GPU 2: [P2] ──┤                  （完整参数，临时）
GPU 3: [P3] ──┘

计算完成后丢弃完整参数，只保留 [Pi]
```

### 反向传播与 Reduce-Scatter

| 操作          | 含义                    |
| ----------- | --------------------- |
| **Reduce**  | 将多个数据按元素进行归约（如求和、求平均） |
| **Scatter** | 将结果分片后分发给各进程          |


```python
# FSDP 反向传播时的梯度处理

def _post_backward_hook(self, param, grad):
    """
    梯度计算完成后自动触发
    """
    # 1. Reduce-Scatter：将完整梯度分片到各 GPU
    #    每个 GPU 只保留自己负责的那部分梯度
    sharded_grad = self._reduce_scatter_grad(grad)
    
    # 2. 释放完整梯度内存
    grad.data = sharded_grad  # 只保留分片
```

```python
Reduce-Scatter 过程：

GPU 0~3 都有完整梯度 [G0|G1|G2|G3]（临时）
        ↓
   Reduce-Scatter
        ↓
GPU 0: [G0_sum]  ← 各 GPU 的 G0 求和
GPU 1: [G1_sum]
GPU 2: [G2_sum]
GPU 3: [G3_sum]
```

步骤拆解：

| 步骤          | 描述                                                           |
| ----------- | ------------------------------------------------------------ |
| **初始状态**    | 4 个 GPU 各自计算完反向传播，都有**完整的梯度张量**（假设被逻辑切分为 G0, G1, G2, G3 四个块） |
| **Reduce**  | 各 GPU 的 G0 块互相通信求和 → G0\_sum；G1 块求和 → G1\_sum；以此类推           |
| **Scatter** | 求和完成后，GPU 0 只保留 G0\_sum，GPU 1 只保留 G1\_sum，...                |
| **最终状态**    | 每个 GPU 只持有**梯度的一个分片**，总显存占用降为 1/4                            |


```python
GPU 0: G = [a0, a1, a2 | b0, b1, b2 | c0, c1, c2 | d0, d1, d2]
GPU 1: G = [a3, a4, a5 | b3, b4, b5 | c3, c4, c5 | d3, d4, d5]
GPU 2: G = [a6, a7, a8 | b6, b7, b8 | c6, c7, c8 | d6, d7, d8]
GPU 3: G = [a9, a10, a11 | b9, b10, b11 | c9, c10, c11 | d9, d10, d11]

Reduce-Scatter 后：

GPU 0: [a0+a3+a6+a9, a1+a4+a7+a10, a2+a5+a8+a11]  = G0_sum
GPU 1: [b0+b3+b6+b9, b1+b4+b7+b10, b2+b5+b8+b11]  = G1_sum
GPU 2: [c0+c3+c6+c9, c1+c4+c7+c10, c2+c5+c8+c11]  = G2_sum
GPU 3: [d0+d3+d6+d9, d1+d4+d7+d10, d2+d5+d8+d11]  = G3_sum
```

## 完整使用示例
### 基础用法

```python
import torch
import torch.nn as nn
from torch.distributed.fsdp import FullyShardedDataParallel as FSDP
from torch.distributed.fsdp.wrap import (
    transformer_auto_wrap_policy,  # 自动包装策略
    size_based_auto_wrap_policy,    # 基于大小的策略
    enable_wrap,                     # 上下文管理器
    wrap,                            # 手动包装
)
import torch.distributed as dist

def setup():
    """初始化分布式环境"""
    dist.init_process_group(backend='nccl')
    local_rank = int(os.environ['LOCAL_RANK'])
    torch.cuda.set_device(local_rank)
    return local_rank

class TransformerBlock(nn.Module):
    """示例：Transformer 层"""
    def __init__(self, dim):
        super().__init__()
        self.attn = nn.MultiheadAttention(dim, num_heads=8, batch_first=True)
        self.ffn = nn.Sequential(
            nn.Linear(dim, dim * 4),
            nn.GELU(),
            nn.Linear(dim * 4, dim),
        )
        self.norm1 = nn.LayerNorm(dim)
        self.norm2 = nn.LayerNorm(dim)
    
    def forward(self, x):
        # 自注意力
        attn_out, _ = self.attn(x, x, x)
        x = self.norm1(x + attn_out)
        
        # 前馈网络
        ffn_out = self.ffn(x)
        x = self.norm2(x + ffn_out)
        return x

class GPTModel(nn.Module):
    def __init__(self, vocab_size, dim, num_layers):
        super().__init__()
        self.embedding = nn.Embedding(vocab_size, dim)
        self.layers = nn.ModuleList([
            TransformerBlock(dim) for _ in range(num_layers)
        ])
        self.head = nn.Linear(dim, vocab_size)
    
    def forward(self, input_ids):
        x = self.embedding(input_ids)
        for layer in self.layers:
            x = layer(x)
        return self.head(x)

def main():
    local_rank = setup()
    
    # ========== 1. 创建模型 ==========
    model = GPTModel(
        vocab_size=50257,
        dim=2048,
        num_layers=24
    ).cuda(local_rank)
    
    # ========== 2. 包装为 FSDP ==========
    fsdp_model = FSDP(
        model,
        # 自动包装策略：每个 TransformerBlock 作为一个 FSDP 单元
        auto_wrap_policy=transformer_auto_wrap_policy(
            module_cls={TransformerBlock}  # 遇到 TransformerBlock 就包装
        ),
        # 混合精度
        mixed_precision=torch.bfloat16,
        # 设备
        device_id=torch.cuda.current_device(),
        # 限制 All-Gather 的显存使用
        limit_all_gathers=True,
    )
    
    # ========== 3. 优化器（注意：优化器状态也会被分片！）==========
    optimizer = torch.optim.AdamW(fsdp_model.parameters(), lr=1e-4)
    
    # ========== 4. 训练循环 ==========
    for step in range(1000):
        batch = torch.randint(0, 50257, (4, 1024), device='cuda')
        
        optimizer.zero_grad()
        
        # 前向：内部自动 All-Gather 参数
        output = fsdp_model(batch)
        
        # 计算损失
        loss = nn.functional.cross_entropy(
            output.view(-1, 50257),
            batch.view(-1)
        )
        
        # 反向：内部自动 Reduce-Scatter 梯度
        loss.backward()
        
        # 优化器步骤：各 GPU 只更新自己负责的分片
        optimizer.step()
        
        if local_rank == 0 and step % 10 == 0:
            print(f"Step {step}, Loss: {loss.item():.4f}")

if __name__ == '__main__':
    main()
```

```python
torchrun --nproc_per_node=8 train.py
```

### 手动包装策略

```python
from torch.distributed.fsdp.wrap import enable_wrap, wrap

# 方式 1：自动包装（推荐）
fsdp_model = FSDP(
    model,
    auto_wrap_policy=size_based_auto_wrap_policy(
        min_num_params=1e6,  # 参数数 > 1M 的模块单独包装
    )
)

# 方式 2：手动精确控制
with enable_wrap(wrapper_cls=FSDP):
    # 只包装特定层
    for i, layer in enumerate(model.layers):
        if i % 2 == 0:  # 每两层包装一个 FSDP 单元
            model.layers[i] = wrap(layer)

# 方式 3：递归包装（最细粒度）
for name, child in model.named_children():
    if isinstance(child, (nn.Linear, nn.MultiheadAttention)):
        setattr(model, name, FSDP(child))
```

### 混合精度与 CPU 卸载

```python
from torch.distributed.fsdp import MixedPrecision, CPUOffload

# 混合精度配置
mp_policy = MixedPrecision(
    param_dtype=torch.bfloat16,      # 参数用 bf16
    reduce_dtype=torch.bfloat16,     # All-Reduce 通信用 bf16
    buffer_dtype=torch.float32,      # 缓冲区用 fp32
)

# CPU 卸载（显存极度不足时使用）
fsdp_model = FSDP(
    model,
    mixed_precision=mp_policy,
    cpu_offload=CPUOffload(offload_params=True),  # 参数 offload 到 CPU
    device_id=torch.cuda.current_device(),
)
```

## 关键参数详解

| 参数                  | 说明                                          | 建议                         |
| ------------------- | ------------------------------------------- | -------------------------- |
| `sharding_strategy` | `FULL_SHARD` / `SHARD_GRAD_OP` / `NO_SHARD` | 大模型用 `FULL_SHARD`          |
| `auto_wrap_policy`  | <font color='red'>决定哪些子模块单独包装</font>                                 | 每层 TransformerBlock 包装     |
| `backward_prefetch` | 反向时预取下一层参数                                  | `BACKWARD_PRE` 重叠通信        |
| `mixed_precision`   | BF16/FP16 配置                                | 配合 `torch.cuda.amp`        |
| `cpu_offload`       | 参数 offload 到 CPU                            | 显存不足时用，但慢                  |
| `limit_all_gathers` | 限制并发 All-Gather 数量                          | `True` 防止显存峰值              |
| `use_orig_params`   | 保留原始参数视图                                    | 需要 `model.parameters()` 时用 |


```python
FSDP(
    module,                          # 要包装的模型
    process_group=None,              # 进程组（默认全局）
    sharding_strategy="FULL_SHARD",  # 分片策略
    cpu_offload=None,                # CPU 卸载配置
    auto_wrap_policy=None,           # 自动包装策略
    backward_prefetch="BACKWARD_PRE", # 反向预取策略
    mixed_precision=None,            # 混合精度配置
    device_id=None,                  # 设备 ID
    limit_all_gathers=True,          # 限制并发 All-Gather
    use_orig_params=False,           # 是否保留原始参数
    param_init_fn=None,              # 参数初始化函数
    sync_module_states=False,        # 是否同步模块状态
)
```

## FSDP 的显存分析

```python
训练 70B 模型，8×A100 (80GB)，序列长度 4096

┌─────────────────────────────────────────────────────────────┐
│                    DDP 显存占用（估算）                        │
│                                                              │
│   参数: 70B × 2 (FP16) = 140 GB                              │
│   梯度: 70B × 2 = 140 GB                                     │
│   优化器状态: 70B × 4 (Adam FP32) = 280 GB                   │
│   激活值: ~200 GB (TP=8 后)                                  │
│   ─────────────────────────────────────                      │
│   总计: ~760 GB / 8 GPU = 95 GB/GPU > 80 GB ❌ OOM         │
└─────────────────────────────────────────────────────────────┘

┌─────────────────────────────────────────────────────────────┐
│                    FSDP 显存占用（估算）                       │
│                                                              │
│   参数分片: 70B × 2 / 8 = 17.5 GB                           │
│   梯度分片: 70B × 2 / 8 = 17.5 GB                           │
│   优化器状态分片: 70B × 4 / 8 = 35 GB                        │
│   激活值: ~200 GB (TP=8 后)                                  │
│   临时完整参数: ~35 GB (All-Gather 峰值)                     │
│   ─────────────────────────────────────                      │
│   总计: ~305 GB / 8 GPU = ~38 GB/GPU < 80 GB ✅ 可训练      │
└─────────────────────────────────────────────────────────────┘
```

## 总结

| 要点                | 说明                                        |
| ----------------- | ----------------------------------------- |
| **FSDP 是什么**      | 全分片数据并行，参数/梯度/优化器状态都分片                    |
| **vs DDP**        | DDP 每卡完整副本，FSDP 分片存储，显存效率更高               |
| **核心操作**          | All-Gather（前向收集参数）、Reduce-Scatter（反向分片梯度） |
| **适用场景**          | 单卡放不下的大模型（>10B 参数）                        |
| **与 Megatron 关系** | Megatron 可用 FSDP 替代自定义 DDP，但通常保留 TP/PP    |
| **性能权衡**          | 显存换通信，需要高速互联（NVLink/IB）                   |

- 一句话：
  - FSDP 是 PyTorch 官方的大模型训练方案，通过全分片将显存压力分散到所有 GPU，是 DDP 在超大模型场景下的自然演进。

### 对比：Megatron 自定义 DDP vs PyTorch FSDP

| 特性        | Megatron 自定义 DDP + TP + PP | PyTorch FSDP                |
| --------- | -------------------------- | --------------------------- |
| **并行维度**  | 3D 并行（TP × PP × DP）        | 数据并行（可组合 TP）                |
| **参数分片**  | 仅优化器状态（ZeRO-1）             | 参数 + 梯度 + 优化器状态             |
| **模型规模**  | 适合 1T+（配合 TP/PP）           | 适合 100B~1T                  |
| **显存效率**  | 中（TP 减少激活值）                | 高（全分片）                      |
| **通信开销**  | TP 大（All-Reduce），PP 小（P2P） | All-Gather + Reduce-Scatter |
| **实现复杂度** | 高（需手动配置 TP/PP）             | 低（自动分片）                     |
| **推荐场景**  | 超大规模（>1T）                  | 大规模（100B~1T）                |
