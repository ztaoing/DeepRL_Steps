| 写法 | 是否存在？ | 类型 | 作用 | 结果 |
| :--- | :---: | :--- | :--- | :--- |
| `x.requires_grad` | ✅ 存在 | 属性 (Property) | 读取状态 | 返回 `True` 或 `False` |
| `x.requires_grad_()` | ✅ 存在 | 方法 (Method) | 修改状态 | 原地修改  $ x $ ，返回  $ x $  本身 |
| `x.requires_grad()` | ❌ 不存在 | 错误写法 | 无 | 报错: `TypeError: 'bool' object is not callable` |



## 读取状态（使用属性）

```python
import torch

x = torch.tensor([1.0, 2.0], requires_grad=True)

# 正确：直接访问属性，不加括号
print(x.requires_grad)  # 输出: True

# 错误：加括号会报错
# print(x.requires_grad())  # <--- 崩溃！TypeError
```

## 修改状态（使用方法）

```python
y = torch.tensor([1.0, 2.0]) # 默认 False

# 正确：调用带下划线的方法
y.requires_grad_(True)   # 开启梯度
y.requires_grad_(False)  # 关闭梯度

print(y.requires_grad)   # 输出: False
```

##  如果你想“复制”一个张量并改变梯度状态怎么办？

既然没有 requires_grad() 来返回一个新张量，如果你想基于现有张量创建一个新张量并设置不同的 requires_grad 状态，应该使用 detach() 或 clone() 组合：

```python
x = torch.tensor([1.0, 2.0], requires_grad=True)

# 目标：创建一个新张量 y，数值同 x，但不需要梯度 (冻结副本)
# 方法：先 detach (切断梯度图)，再根据需要设置 (通常 detach 后默认为 False)
y = x.detach() 
print(y.requires_grad)  # False

# 目标：创建一个新张量 z，数值同 x，且明确需要梯度 (作为新的叶子节点)
z = x.detach().clone().requires_grad_(True)
# 或者更简单的：
z = torch.tensor(x.data, requires_grad=True) 

# 注意：z 和 x 现在完全独立，修改 z 不会影响 x 的计算图
```
