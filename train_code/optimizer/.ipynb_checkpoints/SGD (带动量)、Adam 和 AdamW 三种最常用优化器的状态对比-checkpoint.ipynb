| 特性 | SGD (with Momentum) | Adam | AdamW |
| :--- | :--- | :--- | :--- |
| 全称 | Stochastic Gradient Descent | Adaptive Moment Estimation | Adam with Decoupled Weight Decay |
| 核心思想 | 积累历史梯度方向，加速收敛并抑制震荡 | 同时利用梯度的一阶矩（均值）和二阶矩（未中心方差）自适应调整学习率 | 同 Adam，但将权重衰减从梯度更新中解耦出来 |
| 状态数量 | 1 个 / 参数 | 2 个 / 参数 | 2 个 / 参数 |
| 状态名称 | `momentum_buffer` | `exp_avg` (一阶矩), `exp_avg_sq` (二阶矩) | 同 Adam |
| 显存占用 | 低 (约 1x 参数量) | 高 (约 2x 参数量 + 额外计算) | 高 (同 Adam) |
| 适用场景 | 泛化性要求极高、需精细调参的经典 CV 任务 | 默认首选，NLP、生成模型、快速原型验证 | 大模型预训练 (LLM)、Transformer 架构的标准配置 |
| 关键区别 | 无自适应学习率，依赖手动调 LR | 权重衰减与梯度更新耦合，可能导致正则化失效 | 修复了 Adam 权重衰减的逻辑缺陷，在大模型上表现更好 |



在 PyTorch 中，不同优化器的核心区别不仅在于更新公式，更在于它们维护的内部状态（State）不同。这些状态决定了优化器如何“记忆”过去的梯度，从而调整未来的更新步长。
- 当你调用 optimizer.state_dict() 时，你会看到一个字典，其中 'state' 键包含了每个参数的优化器状态。

## add_param_group - 动态添加参数组


```python
def add_param_group_demo():
    '''
    动态添加参数组
    '''
    print('=== add_param_group ===')
    
    # 创建优化器
    model = nn.Linear(10,5)
    optimizer = optim.SGD(model.parameters(),lr=0.01)
    
    print('初始的参数组')
    for i,group in enumerate(optimizer.param_groups):
        print(f"组{i}: lr={group['lr']},参数数量={len(group['params'])}")
        print(group['params'])
        
    # 创建新的参数组
    new_params = [nn.Parameter(torch.randn(10,3))]
    new_param_group = {
        'params':new_params, # 不同的学习率
        'lr':0.001, # 不同的动量
        'weight_decay':0.01 # 不同权重衰减
    }
    
    # 动态添加参数组
    optimizer.add_param_group(new_param_group)
    print("\n添加新参数组后:")
    for i, group in enumerate(optimizer.param_groups):
        print(f"  组{i}: lr={group['lr']}, momentum={group.get('momentum', 'N/A')}, "
              f"weight_decay={group.get('weight_decay', 'N/A')}, "
              f"参数数量={len(group['params'])}")
    
    # === 实际应用场景：冻结部分层后添加新层 ===
    print("\n实际应用场景 - 迁移学习:")
    
    class TransferModel(nn.Module):
        def __init__(self):
            super().__init__()
            self.backbone = nn.Linear(100,50)
            self.head = nn.Linear(50,10)
        def forward(self,x):
            return self.head(self.backbone(x))
        
    model = TransferModel()
    
    # 冻结backbone
    for param in model.backbone.parameters():
        param.requires_grad = False
    
    # 创建优化器，只优化head
    optimizer = optim.Adam(model.head.parameters(),lr=0.001)
    print(f"初始优化器：{len(optimizer.param_groups)}个参数组")
    
    # 后续解冻backbone并添加回优化器
    for param in model.backbone.parameters():
        param.requires_grad = True
        
    optimizer.add_param_group({'params':model.backbone.parameters(),'lr':0.0001})
    print(f"添加backbone后: {len(optimizer.param_groups)}个参数组")
    print(f"  Group 0 (head): lr={optimizer.param_groups[0]['lr']}")
    print(f"  Group 1 (backbone): lr={optimizer.param_groups[1]['lr']}")
add_param_group_demo()
```

## state_dict 和 load_state_dict - 状态保存与恢复

```python
def state_dict_demo():
    """
    优化器状态保存与恢复
    """
    print("=== state_dict 和 load_state_dict 详解 ===\n")
    
    # 创建模型和优化器
    model = nn.Linear(10, 5)
    optimizer = optim.Adam(model.parameters(), lr=0.001, betas=(0.9, 0.999))
    
    # 模拟训练几步
    for step in range(3):
        x = torch.randn(32, 10)
        y = torch.randn(32, 5)
        output = model(x)
        loss = nn.MSELoss()(output, y)
        
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
        
        print(f"Step {step+1}: loss={loss.item():.4f}")
        
    # 保存优化器状态
    optimizer_state = optimizer.state_dict()
    
    print("\n优化器状态字典结构:")
    print(f"  state: {len(optimizer_state['state'])}个参数状态")
    print(f"  param_groups: {len(optimizer_state['param_groups'])}个参数组")
    
    # 查看状态详情
    for param_group in optimizer_state['param_groups']:
        print(f"\n参数组配置")
        
        # param_group中包含哪些项
        for key,value in param_group.items():
            if key != 'params':
                print(f"{key}:{value}")
                
    # 查看第一个参数的状态
    first_param_id = list(optimizer_state['state'].keys())[0]
    first_state = optimizer_state['state'][first_param_id]
    print(f"\n参数状态示例 (参数ID: {first_param_id}):")
    print(f"  step: {first_state['step']}")
    print(f"  exp_avg shape: {first_state['exp_avg'].shape}")
    print(f"  exp_avg_sq shape: {first_state['exp_avg_sq'].shape}")
    
    
    # 保存到文件
    torch.save(optimizer_state,'optimizer_state.pt')
    
    # === 创建新优化器并加载状态 === 
    new_model = nn.Linear(10, 5)
    new_optimizer = optim.Adam(new_model.parameters(), lr=0.001)
    new_optimizer.load_state_dict(torch.load('optimizer_state.pt')) 
    
    print("\n状态恢复验证:")
    print(f"  原优化器学习率: {optimizer.param_groups[0]['lr']}")
    print(f"  新优化器学习率: {new_optimizer.param_groups[0]['lr']}")
    
    # 验证参数状态
    old_state = optimizer.state[list(model.parameters())[0]]
    new_state = new_optimizer.state[list(new_model.parameters())[0]]
    
    print(f"  原优化器step: {old_state['step']}")
    print(f"  新优化器step: {new_state['step']}")
    print(f"  exp_avg一致: {torch.allclose(old_state['exp_avg'], new_state['exp_avg'])}")
    
    # 清理文件
    import os
    os.remove('optimizer_state.pt')
    
    print("\n实际应用场景 - 断点续训:")
    print("  checkpoint = {")
    print("      'epoch': epoch,")
    print("      'model_state_dict': model.state_dict(),")
    print("      'optimizer_state_dict': optimizer.state_dict(),")
    print("      'loss': loss,")
    print("  }")
    print("  torch.save(checkpoint, 'checkpoint.pt')")

    
state_dict_demo()    
```

## 钩子机制（Hooks）-  状态字典钩子

```python
def state_dict_hooks_demo():
    """
    状态字典钩子详解
    """
    print("=== state_dict 钩子详解 ===\n")
    
    class CustomOptimizer(optim.Adam):
        def __init__(self, params, lr=0.001):
            super().__init__(params, lr=lr)
            self.custom_info = {'version': '1.0', 'timestamp': None}
        
        def register_custom_hooks(self):
            # 保存前的钩子
            def pre_hook(optimizer, state_dict, *args, **kwargs):
                print("  [Pre-hook] 正在准备保存优化器状态...")
                # 添加自定义信息
                state_dict['custom_info'] = optimizer.custom_info
                return state_dict
            
            # 保存后的钩子
            def post_hook(optimizer, state_dict, *args, **kwargs):
                print("  [Post-hook] 状态已保存，可执行清理工作...")
                return state_dict
            
            self.register_state_dict_pre_hook(pre_hook)
            self.register_state_dict_post_hook(post_hook)
    
    # 创建模型和优化器
    model = nn.Linear(10, 5)
    optimizer = CustomOptimizer(model.parameters(), lr=0.001)
    optimizer.register_custom_hooks()
    
    # 模拟训练
    x = torch.randn(32, 10)
    y = torch.randn(32, 5)
    output = model(x)
    loss = nn.MSELoss()(output, y)
    
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
    
    # 保存状态（会触发钩子）
    print("\n保存优化器状态:")
    optimizer.custom_info['timestamp'] = '2024-03-25 15:00:00'
    state_dict = optimizer.state_dict()
    
    print(f"\n状态字典包含自定义信息: {'custom_info' in state_dict}")
    print(f"  custom_info: {state_dict['custom_info']}")
    
    # 加载状态的钩子
    def load_pre_hook(optimizer, state_dict, *args, **kwargs):
        print("\n  [Load Pre-hook] 准备加载状态...")
        # 验证版本兼容性
        if 'custom_info' in state_dict:
            version = state_dict['custom_info'].get('version', 'unknown')
            print(f"    检测到版本: {version}")
        return state_dict
    
    def load_post_hook(optimizer, state_dict, *args, **kwargs):
        print("  [Load Post-hook] 状态加载完成，执行后处理...")
        # 恢复自定义信息
        if 'custom_info' in state_dict:
            optimizer.custom_info = state_dict['custom_info']
        return state_dict
    
    optimizer.register_load_state_dict_pre_hook(load_pre_hook)
    optimizer.register_load_state_dict_post_hook(load_post_hook)
    
    # 创建新优化器并加载
    new_optimizer = CustomOptimizer(model.parameters(), lr=0.001)
    new_optimizer.register_custom_hooks()
    
    print("\n加载优化器状态:")
    new_optimizer.load_state_dict(state_dict)
    print(f"  加载后custom_info: {new_optimizer.custom_info}")

state_dict_hooks_demo()
```

## Step 钩子

```python
def step_hooks_demo():
    """
    Step钩子详解
    """
    print("=== step 钩子详解 ===\n")
    
    class MonitoredOptimizer(optim.Adam):
        def __init__(self, params, lr=0.001):
            super().__init__(params, lr=lr)
            self.gradient_stats = []
            self.register_hooks()
        
        def register_hooks(self):
            # step前钩子
            def pre_hook(optimizer, *args, **kwargs):
                print(f"\n  [Step Pre-hook] 准备更新参数...")
                
                # 记录梯度统计
                grad_norms = []
                for param_group in optimizer.param_groups:
                    for param in param_group['params']:
                        if param.grad is not None:
                            grad_norms.append(param.grad.norm().item())
                
                if grad_norms:
                    avg_grad_norm = sum(grad_norms) / len(grad_norms)
                    optimizer.gradient_stats.append({
                        'step': len(optimizer.gradient_stats),
                        'avg_grad_norm': avg_grad_norm,
                        'max_grad_norm': max(grad_norms)
                    })
                    print(f"    平均梯度范数: {avg_grad_norm:.6f}")
                
                return args, kwargs
            
            # step后钩子
            def post_hook(optimizer, *args, **kwargs):
                print(f"  [Step Post-hook] 参数更新完成")
                
                # 记录参数变化
                for param_group in optimizer.param_groups:
                    for param in param_group['params']:
                        if hasattr(param, 'prev_value'):
                            param_change = (param.data - param.prev_value).norm().item()
                            print(f"    参数变化: {param_change:.6f}")
                        param.prev_value = param.data.clone()
                
                return args, kwargs
            
            self.register_step_pre_hook(pre_hook)
            self.register_step_post_hook(post_hook)
    
    # 创建模型和优化器
    model = nn.Linear(10, 5)
    optimizer = MonitoredOptimizer(model.parameters(), lr=0.01)
    
    # 训练几步
    x = torch.randn(32, 10)
    y = torch.randn(32, 5)
    
    for step in range(3):
        print(f"\n{'='*40}")
        print(f"Step {step + 1}")
        print(f"{'='*40}")
        
        output = model(x)
        loss = nn.MSELoss()(output, y)
        
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
        
        print(f"Loss: {loss.item():.6f}")
    
    # 打印统计信息
    print(f"\n梯度统计历史:")
    for stat in optimizer.gradient_stats:
        print(f"  Step {stat['step']}: avg_norm={stat['avg_grad_norm']:.6f}, "
              f"max_norm={stat['max_grad_norm']:.6f}")

step_hooks_demo()
```

## step - 参数更新（内部机制）
### param_groups
- 是一个列表（list），其中每个元素是一个字典（dict）。它是优化器管理模型参数的核心数据结构

```python
[
    {
        'params': [tensor1, tensor2, ...],  # 该组包含的参数张量列表
        'lr': 0.01,                         # 该组的学习率
        'momentum': 0.9,                    # 该组的动量
        'weight_decay': 0.0001,             # 该组的权重衰减
        'dampeni
        ng': 0.0,                   # 该组的阻尼系数
        'nesterov': False,                  # 是否使用 Nesterov 动量
        ...                                 # 其他优化器特定的参数
    },
    {
        'params': [tensor3, tensor4, ...],  # 第二组参数
        'lr': 0.001,                        # 不同的学习率
        ...
    },
    ...
]
```

```python
def step_method_demo():
    '''
    step方法内部机制
    '''
    class CustomSGD(optim.SGD):
        '''
        自定义SGD，演示step的内部机制
        '''
        def __init__(self,params,lr=0.01,momentum=0,dampening=0,weight_decay=0):
            super().__init__(params,lr=lr,momentum=momentum,dampening=dampening,weight_decay=weight_decay)
            
        @torch.no_grad()
        def step(self,closure=None):
            '''
            手动实现SGD更新逻辑
            '''
            loss = None
            if closuref is not None:
                loss = closure()
            for param_group in self.param_groups:
                lr = param_group['lr']
                momentum = param_group['momentum']
                weight_decay = param_group['weight_decay'] # 权重衰减
                dampening = param_group['dampening'] # 它的作用是减弱当前梯度对动量累积的影响
                
                for param in param_group['params']:
                    if param.grad is None:
                        continue
                    grad = param.grad
                    
                    # 权重衰减
                    if weight_decay !=0:
                        grad +=weight_decay * param.data
                        
                    # 动量更新 
                    if momentum !=0:
                        param_state = self.state[param]
                        if 'momentum_buffer' not in param_state:
                            buf = param_state['momentum_buffer'] = torch.clone(grad).detach()
                            
                        else:
                            buf = param_state['momentum_buffer']
                            buf.mul_(momentum).add_(grad,alpha=1 - dampening)
                        grad = buf
                        
                    param.add_(grad,alpha=-rl)
            return loss
        
       # 比较标准SGD和自定义SGD
    model1 = nn.Linear(10, 5)
    model2 = nn.Linear(10, 5)
    
    # 确保初始参数相同
    model2.load_state_dict(model1.state_dict())
    
    sgd_standard = optim.SGD(model1.parameters(), lr=0.01, momentum=0.9)
    sgd_custom = CustomSGD(model2.parameters(), lr=0.01, momentum=0.9)
    
    x = torch.randn(32, 10)
    y = torch.randn(32, 5)
    
    for step in range(5):
        # 标准SGD
        output1 = model1(x)
        loss1 = nn.MSELoss()(output1, y)
        sgd_standard.zero_grad()
        loss1.backward()
        sgd_standard.step()
        
        # 自定义SGD
        output2 = model2(x)
        loss2 = nn.MSELoss()(output2, y)
        sgd_custom.zero_grad()
        loss2.backward()
        sgd_custom.step()
        
        print(f"Step {step+1}: loss_standard={loss1.item():.6f}, "
              f"loss_custom={loss2.item():.6f}")
        
        # 验证参数一致
        param_diff = 0
        for p1, p2 in zip(model1.parameters(), model2.parameters()):
            param_diff += (p1 - p2).abs().sum().item()
        print(f"  参数差异: {param_diff:.2e}")

step_method_demo()
 
                
```

## zero_grad - 梯度清零

```python
def zero_grad_demo():
    '''
    zero_grad方法详解
    '''
    print("=== zero_grad 方法详解 ===\n")
    
    model = nn.Linear(10, 5)
    optimizer = optim.SGD(model.parameters(), lr=0.01)
    
    x = torch.randn(32, 10)
    y = torch.randn(32, 5)
    
    print("=== zero_grad的作用: 清空累积的梯度=== \n")
    
    # 演示梯度累积
    print("==1. 不调用zero_grad（梯度累积）:==")
    for step in range(3):
        output = model(x)
        loss = nn.MSELoss()(output, y)
        loss.backward()  # 梯度会累积
        
        # 查看梯度范数
        grad_norm = 0
        for param in model.parameters():
            if param.grad is not None:
                grad_norm += param.grad.norm().item() ** 2
        grad_norm = grad_norm ** 0.5
        
        print(f"  Step {step+1}: loss={loss.item():.4f}, grad_norm={grad_norm:.4f}")
        
    
    # 重置模型
    model = nn.Linear(10, 5)
    optimizer = optim.SGD(model.parameters(), lr=0.01)
    
    print("\n==2. 调用zero_grad（正常训练）:==")
    
    for step in range(3):
        optimizer.zero_grad()  # 清空梯度
        
        output = model(x)
        loss = nn.MSELoss()(output, y)
        loss.backward()
        
        # 查看梯度范数
        grad_norm = 0
        for param in model.parameters():
            if param.grad is not None:
                grad_norm += param.grad.norm().item() ** 2
        grad_norm = grad_norm ** 0.5
        
        print(f"  Step {step+1}: loss={loss.item():.4f}, grad_norm={grad_norm:.4f}")
        
    # zero_grad的参数
    print("\nzero_grad参数: set_to_none")
    print("  set_to_none=True: 将梯度设为None（节省内存）")
    print("  set_to_none=False: 将梯度设为0（默认）")
    
    # 性能对比
    import time
    
    def test_zero_grad_performance(model, optimizer, use_none):
        start = time.time()
        for _ in range(1000):
            if use_none:
                optimizer.zero_grad(set_to_none=True)
            else:
                optimizer.zero_grad()
        return time.time() - start
    
    time_zero = test_zero_grad_performance(model, optimizer, False)
    time_none = test_zero_grad_performance(model, optimizer, True)
    
    print(f"\n性能对比 (1000次调用):")
    print(f"  zero_grad(set_to_none=False): {time_zero*1000:.2f}ms")
    print(f"  zero_grad(set_to_none=True): {time_none*1000:.2f}ms")
    print(f"  加速比: {time_zero/time_none:.2f}x")

zero_grad_demo()
```

## 详细状态解析与代码示例
- 假设我们有一个简单的模型参数 weight (形状 [2, 2])

### A. SGD with Momentum (动量随机梯度下降)
