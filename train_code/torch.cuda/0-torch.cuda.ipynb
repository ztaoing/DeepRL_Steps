- torch.cuda 是 PyTorch 中用于管理和操作 NVIDIA GPU（CUDA 设备）的核心模块。它提供了与 CPU 张量完全相同的功能接口，但底层利用 GPU 进行并行计算加速。

以下是 torch.cuda 提供的核心方法分类详解及常用代码示例：
## 🔍 设备信息查询与检查
这些方法用于在代码运行前确认硬件环境和状态：
- torch.cuda.is_available()：检查当前系统是否安装了支持 CUDA 的 NVIDIA 显卡，以及 PyTorch 是否已编译为支持 CUDA。这是写代码时最常用的判断逻辑。
- torch.cuda.device_count()：返回当前机器上可用的 GPU 数量。
- torch.cuda.get_device_name(device=None)：获取指定 GPU 的名称（如 "NVIDIA GeForce RTX 3090"）。
- torch.cuda.get_device_capability(device=None)：获取指定 GPU 的 CUDA 计算能力（例如返回元组 (7, 5)），常用于判断显卡是否支持某些高级特性。
- torch.version.cuda：查看当前 PyTorch 编译时使用的 CUDA 版本（如 "11.8"）。

## ⚙️ 设备管理与上下文控制
用于在多卡环境下指定和切换当前操作的 GPU：
- torch.cuda.current_device()：返回当前正在使用的 GPU 索引。
- torch.cuda.set_device(device)：设置当前默认的 GPU 设备索引。
- class torch.cuda.device(device)：一个上下文管理器，用于在特定的代码块内临时切换 GPU 设备

## 🧠 内存管理与性能监控
GPU 显存（VRAM）非常宝贵，这些方法帮助你监控和优化显存使用：
- torch.cuda.empty_cache()：释放缓存分配器当前持有的所有未占用的显存。注意，这不会增加 PyTorch 可用的总显存，但可以让其他进程或 nvidia-smi 看到更多空闲显存。
- torch.cuda.memory_allocated(device=None)：返回当前指定设备上已被张量占用的显存总量。
- torch.cuda.max_memory_allocated(device=None)：返回指定设备上曾经被张量占用的最大显存总量，常用于排查显存峰值。
- torch.cuda.memory_summary(device=None)：以表格形式打印出当前设备的详细显存分配和使用摘要

## ⚡ 张量操作与同步
- torch.cuda.synchronize(device=None)：由于 GPU 计算是异步的，调用此方法会阻塞 CPU，直到指定设备上的所有 CUDA 操作（如流中的计算）全部完成。常用于精确测量 GPU 程序的运行时间。
- torch.cuda.manual_seed(seed) / torch.cuda.manual_seed_all(seed)：为当前或所有 GPU 设置随机数种子，以保证实验结果的可复现性。


## 🚀 流与事件 (Streams & Events)
用于更细粒度的 GPU 并发控制：
- class torch.cuda.Stream：创建一个 CUDA 流。流是一系列按顺序在 GPU 上执行的命令，使用多个流可以实现操作的并行化。
- class torch.cuda.Event：用于记录流中的时间点，常配合 elapsed_time() 方法来精确测量两段代码在 GPU 上的执行耗时。
