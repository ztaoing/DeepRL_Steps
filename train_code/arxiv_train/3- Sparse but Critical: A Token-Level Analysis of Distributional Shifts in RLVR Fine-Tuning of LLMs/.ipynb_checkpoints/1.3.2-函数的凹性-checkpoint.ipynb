#  什么是函数的凹性？
## 直观理解

|          | 凸函数（Convex）                                 | 凹函数（Concave）                                |
| -------- | ------------------------------------------- | ------------------------------------------- |
| **形状**   | 开口向上 ∪                                      | 开口向下 ∩                                      |
| **比喻**   | 碗状，水不会流出去                                   | 穹顶状，水会流出去                                   |
| **不等式**  | $f(\frac{a+b}{2}) \leq \frac{f(a)+f(b)}{2}$ | $f(\frac{a+b}{2}) \geq \frac{f(a)+f(b)}{2}$ |
| **二阶导数** | $f''(x) \geq 0$                             | $f''(x) \leq 0$                             |

| 特征 | 凸函数 (Convex) | 凹函数 (Concave) |
| :--- | :---: | :---: |
| 开口方向 | 向上 ⬆️ | 向下 ⬇️ |
| 形状 | U 型 🥣 | ∩ 型 ⛰️ |
| 二阶导数 |  $ f''(x) \ge 0 $  |  $ f''(x) \le 0 $  |
| 弦的位置 | 弦在曲线上方 | 弦在曲线下方 |
| 极值性质 | 局部极小 = 全局极小 | 局部极大 = 全局极大 |



# 对数函数为什么是凹的？
- 曲线上任意两点间的弦在曲线下方

## 图形验证
对数函数 f(x)=lnx （或 $ log_2x $ ）的图像：

```python
y
↑
│    ╭────╮
│   ╱      ╲
│  ╱   ●    ╲   ← 曲线上任意两点间的弦在曲线下方
│ ╱  ╱    ╲   ╲
│╱ ╱        ╲   ╲
├──────────────────→ x
0
```

**关键特征**：连接曲线上任意两点的直线段（弦）始终位于曲线下方。

## 数学证明

对 f(x)=lnx  求导：
- 一阶导：f′(x)= $\frac{}1{x}$
  
- 二阶导：f′′(x)=$−\frac{1}{x^2}<0 $  （对所有 x>0 ）
因为二阶导数严格小于0，所以 lnx  是严格凹函数。

# 一句话总结
对数函数的凹性是信息论的"基石"——它让 Jensen 不等式得以应用，从而保证了 KL散度、互信息、最大熵等核心概念的良好性质。没有凹性，整个信息论的数学大厦将失去根基。
