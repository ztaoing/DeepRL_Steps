# 研究背景与动机
## 1.1 RLVR 的成功与黑箱
RLVR (Reinforcement Learning with Verifiable Rewards) 通过可验证奖励（如数学问题的正确性）进行强化学习微调，已在推理任务上取得显著突破（如 GRPO、DAPO 等方法）。然而，<font color='red'>现有评估主要关注聚合层面的指标</font>：
- 准确率 (Accuracy)
- 奖励 (Reward)
- 响应长度 (Response Length)

这些指标无法回答一个根本问题：<font color='red'>RLVR 究竟在 token 级别做了什么？</font>

## 1.2 核心研究问题
- "How does RLVR reshape the token-level predictive distributions of a base model, and which of these changes actually drive downstream reasoning gains?"

# 核心发现：Sparse but Critical
论文提出了一个统一的视角：RLVR 不是全局重写模型行为，而是在少量关键 token 位置进行稀疏、结构化的概率重分配，从而"引导"生成轨迹走向更有效的推理路径。
## 2.1 发现一：分布偏移极度稀疏 (Sparse)
### 量化指标：JS 散度分布
论文使用 Jensen-Shannon (JS) 散度 衡量 base 模型与 RL 微调后模型在每个 token 位置的分布差异：
$JS(x <t) = D_{JS}(π_{base}(⋅|x<t)||π_{RL}（⋅|x<t）)$
### 关键结果（图2）：
| 方法           | 近零散度比例    | 说明                          |
| ------------ | --------- | --------------------------- |
| **DAPO**     | > **83%** | 即使无 KL 正则化，大部分 token 分布几乎不变 |
| **SimpleRL** | > **98%** | 更严格的约束导致更稀疏的更新              |

### 核心洞察：
- 直方图在 JS=0 处有尖锐峰值
- 百分位曲线在 90% 处才开始明显上升
- 只有极少数 token 位置经历了实质性的分布变化

## 2.2 发现二：位置集中性 (Positional Concentration)
### 图3 展示了 JS 散度沿序列位置的分布：
- 序列开头：散度最高 → 修改初始高层分支决策
- 序列中间：散度较低且平稳 → 保持推理连贯性
- 序列末尾：散度再次上升 → 调整答案格式和终止行为
注意：这是聚合趋势，单个序列中高分歧 token 可能稀疏地散布在整个序列中。

## 2.3 发现三：与熵的关系 (Divergence-Entropy Relationship)
图4 对比了低/高散度 token 的熵分布：

| 类别            | 特征                           |
| ------------- | ---------------------------- |
| **低散度 token** | 大多是**低熵**（模型 confident），少量高熵 |
| **高散度 token** | 跨越**广泛的熵范围**                 |

### 方法差异：
- DAPO：能修改低熵和高熵预测，甚至可以覆盖 base 模型的 confident 输出
- SimpleRL：更保守，主要集中在高熵区域
这与 clip-higher 机制有关：更高的 clip 值允许更大的探索，导致更广泛的熵分布。

# Cross-Sampling 干预实验：验证稀疏变化的功能重要性
这是论文最精巧的实验设计——通过选择性替换 token 来证明稀疏变化的因果作用。
## 3.1 实验设计
### Forward Cross-Sampling（正向干预）：
- 在 base 模型生成的序列中，选择性替换某些 token 为 RL 模型的采样结果
- 观察性能如何随替换预算变化
### Reverse Cross-Sampling（反向干预）：
- 在 RL 模型生成的序列中，注入 base 模型的 token 选择
- 观察性能如何退化

## 3.2 核心结果
"Inserting only a small fraction of RL-sampled tokens into base generations progressively recovers RL performance gains"
"Injecting a similarly small number of base token choices into otherwise RL-generated sequences collapses performance to base levels"
### 关键结论：
- 只需替换少量 token（与稀疏性一致），就能将 base 性能提升到 RL 水平
- 反之，只需注入少量 base token，就能将 RL 性能崩溃到 base 水平
- 这孤立出了一小撮直接负责 RLVR 性能增益的 token 级决策

# 细粒度分布机制 (Fine-Grained Mechanics)
## 4.1 Top-k 重叠与重排序
图9：高散度位置的 top-k 候选集重叠

| 方法       | k≥2 重叠率            | 说明   |
| -------- | ------------------ | ---- |
| SimpleRL | > **80%** (常 >85%) | 高度保守 |
| DAPO     | 略低但仍显著             | 更探索性 |

### 关键发现：
- 从 k=1 到 k=2，重叠率急剧上升
- 这意味着：虽然 top-1 token 经常变化，但替代者通常已在 base 模型的 top-3 中
## 4.2 重排序而非替换 (Re-ranking over Replacement)
图22：RL 的 top-3 token 在 base 模型中的排名分布
- ~30% 的 RL top-1 在 base 中已是第 1 名
- 80% (DAPO) 和 >90% (SimpleRL) 的 RL top-1 在 base top-3 内
- RL top-2 通常在 base top-3~4 内

### 核心洞察：RLVR 主要是"选择"而非"发明"
- "RLVR predominantly amplifies underweighted but plausible alternatives rather than effectively inventing new ones"
## 4.3 尾部概率分析
- RLVR 很少显著提升 base 模型中概率极低的 token
- 大多数高散度位置的 RL top-1 在 base 中已有非平凡概率
## 4.4 训练演化 (Training Evolution)
图11：使用中间 checkpoint 追踪
- JS 散度单调递增 throughout training
- 高分位（95th, 99th）增长更快 → 变化越来越集中在稀疏子集
- Jaccard 重叠度：早期逐渐增长，训练末期急剧上升 → 散度 token 集合趋于稳定

# 与 SFT 的对比
关键问题：这种稀疏性是 RLVR 特有，还是微调的普遍特征？
图12 & 图14：SFT vs RLVR 对比

| 维度       | SFT            | RLVR (DAPO) |
| -------- | -------------- | ----------- |
| 高散度集合大小  | **显著更大**       | 更稀疏         |
| 散度分布     | 更宽、更全局         | 更集中         |
| 位置模式     | 整个响应都 elevated | 主要在开头和结尾    |
| top-k 重叠 | **更低**         | 更高          |
| 重排序强度    | 更激进            | 更保守         |

## 结论：
- RLVR 的稀疏性是 distinctive 的，不是微调的通用特征。
SFT 直接学习目标输出，导致全局性、更尖锐的分布更新；而 RLVR 通过奖励信号进行选择性精炼。

# 六、Divergence-Weighted Advantage
## 6.1 动机
- 既然 token 级散度能指示"关键决策位置"，是否可以在训练时利用这一信息？

## 6.2 方法
提出散度加权的优势信号变体：$\hat{A}{~}_t = w(JS_t)⋅A_t$)其中$w(JS_t)$  是基于 token 级 JS 散度的权重函数。

## 6.3 结果
- 这种简单修改可以影响学习动态
- 在某些设置下能超越基线
- 但稳定性能可能需要模型特定的选择

# 核心洞察与意义
## 8.1 "手术刀而非锤子" (Scalpel, not Hammer)
"RLVR acts primarily as a sparse and targeted refinement process"
RLVR 的成功不在于广泛重写模型，而在于精准地识别并放大 base 模型中已存在但被低估的合理推理路径。
## 8.2 对遗忘的启示
这与 "RL's Razor" (Shenfeld et al., 2025) 的发现一致：
- RL 遗忘更少，因为它变化更稀疏、更靠近 base 分布
- 这篇论文在 token 分布层面 提供了直接证据
## 8.3 对未来方法的启示
- 设计更高效的 RLVR：专注于高散度 token，可能减少计算开销
- 可解释性：识别"关键决策 token"有助于理解模型推理
- 可控性：通过干预这些稀疏关键点，可能实现更精细的模型编辑

# 一句话总结
- RLVR 不是让模型学会全新的推理方式，而是在模型已有的知识中，精准地找到并放大那些通往正确答案的"被低估的合理路径"——这一切只需要修改极少数关键的 token 决策。
