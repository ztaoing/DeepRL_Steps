# JSD

## 定义
### 数学公式
<font color='red'>对于两个概率分布 P  和 Q </font>，Jensen-Shannon 散度定义为：

$D_{JS}(P||Q) = \frac{1}{2}D_{KL}(P||M) + \frac{1}{2}D_{KL}(Q||M)$

其中：
- $M = \frac{1}{2}(P+Q)$ <font color='red'>两个分布的平均分布（中点分布）</font>
- $D_{KL}(⋅||⋅)$

#### 第一步：构造平均分布 M

- M  代表"如果我不知道来自哪个分布，最公平的猜测是什么"。

```python
P = [0.7, 0.2, 0.1]      (分布1：比如 base 模型的 token 概率)
Q = [0.1, 0.6, 0.3]      (分布2：比如 RL 模型的 token 概率)
        ↓
M = ½(P + Q) = [0.4, 0.4, 0.2]   (中间立场)
```

#### 第二步：<font color='red'>计算两个 KL 散度</font>

| 项                | 含义                             |
| ---------------- | ------------------------------ |
| D_{KL}(P \| M) | **从 P 的角度看**："用 M 来近似 P，有多糟糕？" |
| D_{KL}(Q \| M) | **从 Q 的角度看**："用 M 来近似 Q，有多糟糕？" |


-  KL 散度  $D_{KL}(X∥Y)=\sum_iX(i)⋅log\frac{Y(i)}{X(i)}$
- <font color='red'> 衡量：用分布 Y  来编码来自分布 X  的数据，多浪费多少比特</font>

#### 第三步：取平均
$\frac{1}{2}[P的损失] + \frac{1}{2}[Q的损失]$
##### 为什么要平均？
- <font color='red'>对称性</font>：$D_{JS}(P∥Q)=D_{JS}(Q∥P)$  —— 不偏向任何一方
- 公平性：两个分布都是"信息源"，同等重要

#### 直观理解
##### 类比：两个人的"意见距离"

```python
小明认为：A题选①(70%), ②(20%), ③(10%)
小红认为：A题选①(10%), ②(60%), ③(30%)

路人平均：      ①(40%), ②(40%), ③(20%)

小明有多不认同路人？ → D_KL(小明||路人)
小红有多不认同路人？ → D_KL(小红||路人)

JS散度 = 两人不认同程度的平均
        = "小明和小红的分歧有多大"
```

#### 为什么不用直接用 KL？

| 问题                               | KL 散度          | JS 散度               |
| -------------------------------- | -------------- | ------------------- |
| $D_{KL}(P\|Q)$ vs $D_{KL}(Q\|P)$ | <font color='red'>**不相等**，方向敏感  </font> | <font color='blue'>**相等**，无方向  </font>        |
| P 和 Q 支撑集不同                  | 可能 **∞**（未定义）  | **始终有定义**           |
| 值域                               | $[0, +\infty)$ | $[0, \ln 2]$ **有界** |

- <font color='blue'>在 "Sparse but Critical" 论文中，正是利用 JS 的有界性和定义良好性，才能在 top-p 截断分布上稳定计算。</font>

#### 与熵的关系（另一种表达）

### 对称形式
$D_{JS}(P||Q) = H(M) - \frac{1}{2}H(P) -\frac{1}{2}H(Q)$

| 项                                   | 含义            |
| ----------------------------------- | ------------- |
| H(M)                              | 混合后的**不确定性**  |
|  \frac{1}{2}H(P) + \frac{1}{2}H(Q)  | 各自不确定性的**平均** |


- 其中 H(⋅)  是香农熵。
##### 这揭示了 JSD 的本质：<font color='red'>平均分布的熵 减去 两个分布熵的平均</font>

解读：混合后比"各管各"多了多少混乱？

- 若 P≈Q ：混合不增加混乱 → $D_{JS}≈0 $
- 若 P⊥Q ：混合很混乱 → $D_{JS}=ln2 （最大）$


## 核心性质

| 性质        | 说明                                        | 与 KL 散度的对比                                |
| --------- | ----------------------------------------- | -------------------------------- |
| <font color='red'>**有界性** </font>  | $0 \leq D_{JS}(P \| Q) \leq \ln 2$（对离散分布） | KL 散度无上界，可能为 $+\infty$                    |
| **对称性**   | $D_{JS}(P \| Q) = D_{JS}(Q\|P)$         | KL 散度不对称：$D_{KL}(P\|Q) \neq D_{KL}(Q\|P)$ |
| **非负性**   | $D_{JS}(P \| Q) \geq 0$，等号当且仅当 $P=Q$      | KL 同样非负                                   |
| <font color='red'>**定义良好性**</font> | 即使 $P$ 和 $Q$ 的支撑集不同，JSD 仍有定义              | KL 散度要求 $P \ll Q$（绝对连续）                   |
| **度量性**   | 其平方根 $\sqrt{D_{JS}}$ 是一个**度量（metric）**    | KL 不是度量                                   |



## 与 KL 散度的关系
### 直观对比

```python
KL 散度：D_KL(P||Q) = E_P[log(P/Q)]  ——  有方向性，"从 Q 看 P 的惊讶程度"
JSD：    D_JS(P||Q)                  ——  无方向性，"P 和 Q 的平均混乱程度"
```

### 为什么 JSD 更好？
#### 场景：比较 top-p 截断分布
在 LLM 中，由于内存限制，我们通常只能获取 top-p（如前 90% 累积概率）的 token 分布，而非完整词汇表分布。

| 问题           | KL 散度                                    | JSD                 |
| ------------ | ---------------------------------------- | ------------------- |
| 截断分布的支撑集可能不同 | $D_{KL}$ 可能 **无定义**（出现 $0/0$ 或 $\log 0$） | **始终有定义**           |
| 极端值主导统计      | 可能因小概率事件产生巨大值                            | 被 $\ln 2$ 上界**稳定化** |
| 方向性歧义        | 需要选择 $D_{KL}(P\|Q)$ 还是 $D_{KL}(Q\|P)$    | 对称，无需选择             |

- 这正是 "Sparse but Critical" 论文选择 JSD 而非 KL 的核心原因（见论文 Section 2.1）。

## 信息论解释
### 从熵的角度
$D_{JS}(P||Q) = H(\frac{P+Q}{2}) - \frac{H(P) + H(Q)}{2}$

#### 这可以解读为：
<font color='red'>"两个分布混合后的不确定性" 减去 "各自不确定性的平均"</font>
- 如果 P≈Q ：混合不会增加太多不确定性 → JSD ≈ 0
- 如果 P  和 Q  差异大：混合后更"混乱" → JSD 大

### 与互信息的联系

JSD 与互信息有深刻联系。

设 Z  是一个二元随机变量，以 1/2 概率从 P  或 Q  中采样，
则：$D_{JS}(P∥Q)=I(Z;X) $

- 即 JSD 等于 Z  和采样结果 X  之间的互信息——衡量"知道来自哪个分布"能减少多少不确定性。

## 在 "Sparse but Critical" 论文中的应用
###  具体计算方式

对于每个 token 位置 t ，给定前缀 $x_{<t}$：

$JS(x_{<t}) = D_{JS}(π_{base}(⋅∣x_{<t}) || π_{RL}(⋅∣x_{<t}))$
= $\frac{1}{2}D_{KL}(π_{base}||M_t) + \frac{1}{2}D_{KL}(π_{RL}||M_t)$

其中$M_t = \frac{1}{2}(π_{base}(⋅∣x_{<t}) + π_{RL}(⋅∣x_{<t}))$

### 为什么这个选择是关键的？

| 论文需求                     | JSD 的优势     |
| ------------------------ | ----------- |
| 比较 base 和 RL 模型的**条件分布** | 对称性：不偏向任一模型 |
| 处理 top-p 截断分布            | 定义良好：无需绝对连续 |
| 聚合统计（如直方图、百分位）           | 有界性：极端值不会主导 |
| 跨方法比较（DAPO vs SimpleRL）  | 标准化：值在同一尺度上 |

### 实际计算中的简化
由于内存限制，论文在** top-p 截断分布**上计算 JSD：

```python
# 概念性伪代码
def compute_js_divergence(logits_base, logits_rl, top_p=0.9):
    # 1. 分别获取 top-p 截断分布
    dist_base = top_p_truncated_softmax(logits_base, top_p)
    dist_rl = top_p_truncated_softmax(logits_rl, top_p)
    
    # 2. 取平均分布
    M = 0.5 * (dist_base + dist_rl)
    
    # 3. 计算两个 KL 项
    kl_base_to_M = sum(dist_base[i] * log(dist_base[i] / M[i]) for i in tokens)
    kl_rl_to_M = sum(dist_rl[i] * log(dist_rl[i] / M[i]) for i in tokens)
    
    # 4. JSD
    jsd = 0.5 * kl_base_to_M + 0.5 * kl_rl_to_M
    
    return jsd
```

## 可视化理解
### 两个相同分布

```python
P = [0.5, 0.5]      Q = [0.5, 0.5]
M = [0.5, 0.5]

D_JS(P||Q) = 0      ← 完全重合，无散度
```

### 两个完全不同分布

```python
P = [1, 0]          Q = [0, 1]
M = [0.5, 0.5]

D_KL(P||M) = 1*log(1/0.5) + 0 = log(2)
D_KL(Q||M) = log(2)
D_JS(P||Q) = 0.5*log(2) + 0.5*log(2) = log(2) ≈ 0.693

← 达到最大值 ln(2)
```

### 部分重叠分布

```python
P = [0.7, 0.3]      Q = [0.3, 0.7]
M = [0.5, 0.5]

D_JS ≈ 0.1          ← 中等散度
```

## 与其他距离/散度的对比

```python
| 度量                 | 对称 | 有界 | 三角不等式       | 要求绝对连续 | 适用场景      |
| ------------------ | -- | -- | ----------- | ------ | --------- |
| **KL 散度**          | ✗  | ✗  | ✗           | ✓      | 变分推断、最大似然 |
| **Jensen-Shannon** | ✓  | ✓  | ✗（但√JSD 满足） | ✗      | 分布比较、GAN  |
| **总变差 (TV)**       | ✓  | ✓  | ✓           | ✗      | 概率收敛      |
| **Wasserstein**    | ✓  | ✗  | ✓           | ✗      | 最优传输、GAN  |
| **Hellinger**      | ✓  | ✓  | ✓           | ✗      | 假设检验      |
```

## 历史与命名
Jensen-Shannon 散度由 Jianhua Lin 于 1991 年提出，以两位数学家的名字命名：
- Johan Jensen (1859–1925)：丹麦数学家，Jensen 不等式
- Claude Shannon (1916–2001)：美国数学家，信息论之父
其命名反映了公式中** Jensen 不等式的应用和 Shannon 熵**的核心角色。


## 一句话总结
- Jensen-Shannon 散度是 KL 散度的"对称、有界、定义良好"版本，通过在两个分布的中点分布上取平均 KL 散度，既保留了信息论的深度，又克服了 KL 散度的方向性和无界性缺陷——这正是 "Sparse but Critical" 论文选择它来精确度量 token 级分布差异的根本原因。
