# Extended Convex Lifting for Policy Optimization of Optimal and Robust Control

- 由 Yang Zheng、Chih-Fan Pai、Yujie Tang 三位作者撰写，于2026年发表在控制领域顶级期刊 IEEE Transactions on Automatic Control (TAC) 上，是该期刊系列文章的第16篇。
- https://proceedings.mlr.press/v283/zheng25a.html

## 研究背景与动机
- 经典的最优与鲁棒控制问题（如LQR、LQG、H∞控制）在策略优化形式下通常是非凸甚至非光滑的。
- 虽然已有不少工作通过变量变换（凸重构）证明了某些控制问题具有"良性"的优化景观，但这些工作大多是针对单个问题的个案分析，缺乏一个统一的理论框架来解释这些现象。

### 核心贡献：扩展凸提升（ECL）框架
论文提出了 Extended Convex Lifting（ECL） 这一统一理论框架，主要贡献包括：
  - 统一视角：
    - 通过引入提升变量（lifting variables）和辅助集，<font color='red'>将原始非凸函数的上图集（epigraph）与高维空间中的"部分凸"集合通过光滑微分同胚联系起来，从而揭示出经典控制问题中隐藏的凸性结构。</font>
  - 全局最优性保证：
    - 证明了在ECL框架下，对于一大类次微分正则函数，任何"非退化"的Clarke稳定点都是全局极小点（定理2）。这为分析各种控制问题的优化景观提供了统一的理论工具。
  - 覆盖多个基准控制问题：
    - 为 LQR（命题1）和状态反馈 H∞ 控制（定理4）构造了ECL，证明所有稳定化策略都是非退化的，其任何稳定点都全局最优。
    - 克服了非严格LMI带来的技术困难，成功为 LQG（命题3）和输出反馈 H∞ 控制（命题4）构造了ECL，严谨地证明了"非退化"策略的稳定点具有全局最优性。
    
## 关键技术创新
  - 上图集提升策略：
    - 不同于传统方法直接研究非凸函数本身，ECL将视角提升到其上图集，通过引入额外的提升变量（通常对应Lyapunov变量），在高维空间中揭示隐藏的凸性结构。
  - 链式包含关系设计：
    - ECL定义中采用严格的链式包含关系，深刻洞察了控制理论中严格LMI与非严格LMI之间的微妙差异，解决了最优解落在LMI可行域边界上时无法恢复的问题。
  - 非退化点概念：
    - 自然引出了"非退化点"的定义，完美区分了全局最优策略和可能是次优鞍点的退化策略（如LQG中由相似变换导致的鞍点）。
    
## 研究意义
该工作为理解策略梯度等学习型控制方法在复杂控制问题中的成功提供了新的理论视角，表明其收敛到的稳定点（若非退化）具有全局最优性保证。作者还认为，ECL框架对控制领域之外的非凸问题分析也具有独立的学术价值。
