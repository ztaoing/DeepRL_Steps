# 熵的基本定义
## 信息论中的熵
对于离散概率分布 P ：
$H(P)=−∑_xP(x)logP(x)$
 
| 分布            | 熵值               | 含义         |
| ------------- | ---------------- | ---------- |
| 确定性（P(a)=1） | H=0            | 完全确定，无不确定性 |
| 均匀分布（n个等概率） | H=logn       | 最大不确定性     |
| 偏斜分布          | 0 < H < logn | 部分确定，部分不确定 |
 
## 策略熵（Policy Entropy）
在RL中，**策略 $π_θ(a∣s) $ 是在状态 s  下选择动作 a  的概率分布：**
-  $H(π_θ(⋅∣s))=−∑_aπ_θ(a∣s)logπ_θ(a∣s)$
 
### 直观理解：
- 高熵 → 策略"犹豫"，动作选择分散 → 探索性强
- 低熵 → 策略"自信"，倾向于某个动作 → 利用性强

# 为什么RL需要熵？
## 探索-利用的困境（Exploration-Exploitation Dilemma）

```python
┌─────────────────────────────────────────────────────────┐
│  利用（Exploitation）        探索（Exploration）          │
│  ───────────────────        ───────────────────          │
│  选择当前认为最好的动作       尝试未知或不确定的动作        │
│  → 短期收益最大化            → 长期发现更优策略            │
│  → 风险：陷入局部最优         → 风险：短期收益波动         │
│                                                         │
│  熵的角色：量化探索潜力                                     │
│  → 高熵 = 还有探索空间                                     │
│  → 低熵 = 探索耗尽，可能陷入局部最优                        │
└─────────────────────────────────────────────────────────┘
```

|          | 高熵           | 低熵            |
| -------- | ------------ | ------------- |
| **概率分布** | 分散（<font color='red'>多个选项概率相近</font>） | 集中（某个选项概率接近1） |
| **模型行为** | "犹豫"、探索、多样   | "自信"、确定、单一    |
| **直观比喻** | 抛均匀硬币（各50%）  | 抛作弊硬币（99%正面）  |


## 高熵（分散）：

```python
动作A: 33%    log(0.33) = -1.10
动作B: 33%    log(0.33) = -1.10
动作C: 34%    log(0.34) = -1.08

H = -(0.33×(-1.10) + 0.33×(-1.10) + 0.34×(-1.08)) ≈ 1.10

→ 模型"不知道"该选哪个，在探索
```

## 低熵（集中）

```python
动作A: 95%    log(0.95) = -0.05
动作B: 3%     log(0.03) = -3.51
动作C: 2%     log(0.02) = -3.91

H = -(0.95×(-0.05) + 0.03×(-3.51) + 0.02×(-3.91)) ≈ 0.20

→ 模型"确信"选A，几乎不探索
```

## 多臂老虎机
假设有3个拉杆，真实奖励分别为 [1, 2, 3]：

| 策略            | 行为           | 熵                     | 结果               |
| ------------- | ------------ | --------------------- | ---------------- |
| 纯利用           | 总是拉杆3        | H=0                 | 错过验证，如果估计错误就永远错了 |
| ϵ-贪心 | 90%拉杆3，10%随机 | H ≈ 0.1 log 3 | 偶尔探索，可能发现更好的     |
| 高熵策略          | 三个拉杆概率相近     | H ≈ log 3    | 充分探索，但短期收益低      |


# 熵在RL算法中的具体应用
## 最大熵RL（Maximum Entropy RL）

目标函数加入熵奖励：

$max_θE[∑_tr_t+βH(π_θ(⋅∣s_t))]$
 
| 算法                         | 熵的作用                  |
| -------------------------- | --------------------- |
| **SAC（Soft Actor-Critic）** | 自动调整 β，使熵保持在目标值 |
| **SQL（Soft Q-Learning）**   | 通过softmax获得随机策略       |
| **MaxEnt RL**              | 在约束下最大化轨迹熵            |
 
## SAC的自动温度调整：
- $min_βE[−βlogπ(a∣s)−βH^ˉ]$
其中 $ H^ˉ$ 是目标熵（如动作维度的负数）。
