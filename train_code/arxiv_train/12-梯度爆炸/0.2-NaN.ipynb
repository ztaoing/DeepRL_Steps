# NaN（全称 Not a Number，即"非数字"）
- 是计算机科学中浮点数数据类型的一个特殊值，用于表示未定义或不可表示的数值结果。它最早由 1985 年的 IEEE 754 浮点数标准引入。

## 核心特性
NaN 有几个非常"反直觉"的特性：
- 类型是 number：typeof NaN 的结果是 "number"，<font color='red'>虽然它表示"非数字"，但它属于数值类型</font>
- **不等于自身**：NaN === NaN 的结果是 false，这是所有值中唯一不与自身相等的特例
- 具有传染性：<font color='red'>任何涉及 NaN 的数学运算，结果仍然是 NaN</font>，例如 <font color='blue'>NaN + 5 → NaN</font>

## 常见产生场景
### 无效数学运算

```python
0 / 0              // NaN（注意：非零数/0 是 Infinity，不是 NaN）
Infinity - Infinity // NaN
Infinity / Infinity // NaN
0 * Infinity       // NaN
Math.sqrt(-1)      // NaN（负数开方，实数范围内无解）
```

### 类型转换失败

```python
Number("abc")      // NaN
parseInt("hello")  // NaN
"abc" * 3          // NaN
```

### 深度学习中的 NaN
结合我们之前讨论的内容，在模型训练中 loss 变成 NaN 通常是因为：
- 梯度爆炸：梯度值指数级增长，<font color='red'>溢出浮点数表示范围</font>
- 数值溢出：exp(x) 中 <font color='red'>x 过大</font>、log(0) 等
- 数据异常：输入数据本身包含 NaN 或 Inf

## 如何正确检测 NaN
由于 NaN !== NaN，不能用 === 来判断，正确方式如下：

| 方法 | 示例 | 推荐度 |
|---|---|---|
| `Number.isNaN(x)` | `Number.isNaN(NaN)` → `true` | ✅ 最推荐，严格判断 |
| `Object.is(x, NaN)` | `Object.is(NaN, NaN)` → `true` | ✅ 可靠 |
| `x !== x` | 利用 NaN 不等于自身的特性 | ✅ 简洁的 trick |
| `isNaN(x)` | `isNaN("abc")` → `true`（会误判） | ⚠️ 不推荐，有隐式转换 |


## 在 IEEE 754 中的底层表示
在 32 位单精度浮点数中，NaN 的二进制表示规则是：
- 指数位全为 1（8 位全 1）
- 尾数位不全为 0（如果尾数全为 0，则表示 Infinity）
这意味着 NaN 实际上不是"一个"值，而是一族值，尾数位的不同组合可以表示不同的 NaN。

### 一句话总结
- NaN 是浮点数系统中用来标记"这个计算结果没有意义"的哨兵值。它类型是 number，但不等于任何值（包括自己），且具有传染性——一旦参与运算，结果就是 NaN。
