# 为什么SFT用正向KL，RLHF用反向KL？

- 在 SFT（监督微调）和 RLHF（基于人类反馈的强化学习）中，<font color='red'>虽然都涉及 KL 散度（Kullback-Leibler Divergence）作为正则化项来防止模型偏离参考策略</font>（Reference Policy），<font color='blue'>但它们采用的方向确实不同</font>：
  - SFT 通常使用**正向 KL**（Forward KL），
  - 而 RLHF 通常使用**反向 KL**（Reverse KL）。

- 这种选择并非随意，而<font color='blue'>是由优化目标、数学性质、采样机制以及防止模式崩溃（Mode Collapse）的需求共同决定的</font>。

## 为什么 SFT 用正向 KL？
在 SFT 中，
- 目标分布 P 是人类标注的 Ground Truth 数据分布，
- 模型 Q 需要尽可能拟合这些数据。

### A. 避免“漏掉”正确答案（Zero-avoiding）
<font color='red'>SFT 的核心任务是模仿</font>。
   - 如果人类标注了某个答案 $x^∗$是正确的（即 $P(x^∗)>0 $），模型绝对不能说“这个答案不可能”（即 $Q(x^∗)≠0 $）。
   
- <font color='red'>正向 KL 具有 Zero-avoiding 性质</font>：只要 P(x)>0 ，**惩罚项 $log\frac{P(x)}{Q(x)}$就会迫使 Q(x) 保持非零。**
- 如果用反向 KL，**模型可能会为了安全或简单，直接忽略某些低频但正确的标注模式（Mode-seeking），导致漏答或遗忘**，<font color='red'>这在 SFT 中是不可接受的。</font>

### B. 数学上的等价性

```python
Image(filename='./Image/kl1.png')
```

### C. 采样来源
SFT 的数据是从 $P_{data}$中采样的。在正向 KL 的梯度估计中，我们需要从 P 采样，这与 SFT 的训练方式完全一致。

## 总结对比

| 特性 | SFT (正向 KL:   P → Q  ) | RLHF (反向 KL:   Q → P  ) |
| :--- | :--- | :--- |
| 行为模式 | <font color='red'>Mean-seeking (均值寻求)</font> | <font color='red'>Mode-seeking (模式寻求)</font> |
| 对   P>0, Q→ 0   的惩罚 | 无穷大 (Zero-avoiding) | 无惩罚 |
| 对   Q>0, P→ 0   的惩罚 | 无惩罚 | 无穷大 (Zero-forcing) |
| 核心目的 | <font color='red'>不遗漏</font>：必须覆盖所有人类标注的正确答案 | <font color='red'>不越界</font>：必须避开低奖励或荒谬区域，防止 Reward Hacking |
| 采样方式 | <font color='red'>Off-policy (从数据   P   采样)</font> | <font color='red'>On-policy (从模型   Q   采样)</font> |
| 梯度估计 |  E_{x ∼ P}   (稳定) |  E_{x ∼ Q}   (稳定，无需 IS) |
| 失败后果 | 模型漏答、遗忘知识 | 模型输出乱码、陷入奖励陷阱、熵坍塌 |

### 一句话解释：
- **SFT 用正向 KL** 是因为“<font color='red'>宁可多学，不可漏学</font>”（**必须覆盖所有正确答案**）；
- **RLHF 用反向 KL** 是因为“<font color='red'>宁可保守，不可乱来</font>”（**必须避开奖励陷阱，且受限于 On-policy 采样**）。

## 补充：最新的趋势（KL 惩罚 vs KL 约束）
值得注意的是，在最新的 RLHF 算法（如 KTO, DPO, GRPO）中，显式的 KL 散度计算正在被简化：

- DPO: **将 KL 约束隐式地融入到了 Loss 中，不再显式计算 KL。**
- GRPO: 使用 Group Relative Policy Optimization，通过组内相对优势来替代 KL 惩罚，利用组内方差自然起到正则化作用。
- Truncated KL: 在 PPO 中，为了防止 KL 爆炸，常使用 clip 后的 KL 或仅用 KL 作为 early stopping 信号，而不是直接加在 Loss 里。

**但归根结底，这些方法的理论基石依然是**：
- 我们希望模型在参考分布附近探索（**Reverse KL 的思想**），
- 而不是被迫去拟合一个可能包含噪声的全局目标分布（**Forward KL 的风险**）。

# 展开讲讲 PPO 中 KL 散度具体是怎么加到 loss 里的吗？这是理解这个机制最直接的入口。

在 PPO（Proximal Policy Optimization）中，KL 散度并不是像 SFT 那样直接作为一个普通的 Loss 项加进去，**而是扮演着一个“安全阀”和“自适应调节器”的角色。**

要理解这一点，我们需要先回顾 PPO 的核心目标函数。PPO 的核心是最大化一个“截断的代理目标（Clipped Surrogate Objective）”，<font color='red'>为了防止模型在单步更新中偏离参考策略（Reference Policy）太远，通常会引入一个基于反向 KL 散度的惩罚项。</font>

**PPO 的完整 Loss 函数通常长这样：**

```python
Image(filename='./Image/ppo_loss.png')
```


这里面的 <font color='red'>$−β⋅D_{KL}[π_θ∣∣π_ref] $就是我们要讨论的核心</font>。

下面我**分三个层次为你详细拆解它在 PPO 中的具体运作机制**：

## 为什么是反向 KL？（从采样角度）
注意看 KL 散度的形式： 
$D_{KL}[π_θ∣∣π_{ref}] $。
- <font color='red'>期望是在当前模型 $π_θ$下采样的（即 E_{x∼$π_θ$} ）</font>。
  - **这意味着我们在计算当前模型和参考模型的差异时，只关心当前模型实际生成了哪些 Token**。
  - 如果当前模型生成了一个参考模型认为概率为 0 的 Token（即 $P_{ref}→0$ ），反向 KL 会给出无穷大的惩罚。<font color='red'>这完美契合了 RLHF 中“防止模型乱造奖励模型没见过的荒谬内容”的需求</font>。


## KL 在 PPO 中的两种具体“加法”（核心机制）
在实际的工程实现（如 OpenRLHF、TRL 等库）中，KL 散度并不是简单地乘以一个固定的 β 就完事了。

**为了防止训练崩溃，通常采用以下两种机制之一，或者结合使用：**

### 机制 A：自适应 KL 系数（Adaptive KL Penalty）
如果 β 是一个固定常数，很容易出问题：
- <font color='red'>β 太小</font>：模型会为了刷高奖励，<font color='blue'>疯狂偏离参考模型，导致奖励黑客（Reward Hacking）或输出乱码</font>。

- <font color='red'>β 太大</font>：<font color='blue'>模型被死死按在参考模型附近，不敢探索，RLHF 退化成普通的 SFT，学不到新东西。</font>

#### 解决方案：把 β 变成一个动态变量。
在每一个 Epoch 或 Step 结束后，<font color='blue'>计算当前 batch 的平均 KL 散度 $D^ˉ_{KL}$ </font>，然后与一个目标阈值 $D_{target}$进行比较：


```python
Image(filename='./Image/β.png')
```

这种机制让 KL 惩罚变成了一个PID 控制器，<font color='blue'>始终将模型的探索范围控制在安全边界内</font>。

### 机制 B：KL 截断 / 早期停止（KL Truncation / Early Stopping）
- <font color='blue'>这是目前更主流、更稳妥的做法</font>。
- **在 PPO 的每一个 Epoch 内部，模型会进行多次 mini-batch 的梯度更新。随着更新的进行，当前模型 $π_θ$会不可避免地越来越偏离初始的参考模型 $π_{ref}$。**
  
#### 解决方案：
- 在每次计算 Loss 时，实时监控 KL 散度。一旦某个 mini-batch 的 KL 散度超过了一个预设的硬上限（比如 target_kl = 0.05），直接停止当前 Step 的所有后续更新，进入下一个 Step 重新采样。
  - 通俗比喻：
    - PPO 的 Clip 机制是“限速带”（防止梯度太大导致车毁人亡）；
    - 而 KL 散度是“偏离导航的警报器”。一旦你偏离原定路线太远，
      - 自适应 KL 会慢慢踩刹车，
      - <font color='blue'>而 KL 截断则是直接拉手刹，让你回到起点重新规划路线。</font>

## KL 散度在代码中是如何计算的？
在 LLM 的 PPO 中，我们不会真的去算完整的概率分布积分，而是基于采样的 Token 序列进行近似计算。
对于一个生成的 Token 序列 x ，反向 KL 散度的单步计算极其简单：

$D_{KL}(π_θ∣∣π_{ref})≈logπ_θ(x_t)−logπ_{ref}(x_t)  $
 
代码层面的逻辑（伪代码）： 

```python
# 1. 在当前策略下采样，并计算当前策略和参考策略的 log_probs
current_log_probs = model(input_ids).log_probs
ref_log_probs = ref_model(input_ids).log_probs  # ref_model 通常是冻结的，不参与梯度计算

# 2. 计算 Token 级别的 KL 散度 (即 log ratio)
token_kl = current_log_probs - ref_log_probs

# 3. 在序列维度求和或求平均，得到整个句子的 KL
sequence_kl = token_kl.sum(dim=-1) 

# 4. 将 KL 加入总 Loss (通常乘以动态的 beta)
total_loss = -ppo_ratio_loss + beta * sequence_kl.mean()
```

## 总结
- 在 PPO 中，KL 散度不是用来“拟合”什么的，<font color='blue'>而是用来“限制”什么的。</font>
- 它利用了反向 KL 的 Zero-forcing 特性，配合自适应系数或截断机制，确保模型在利用 Reward Model 提升分数的同时，始终保持在人类语言的自然流形（Manifold）内，不至于变成只会刷分的“复读机”或“乱码生成器”。

# 价值函数（Value Function）
- 它用来估计每个 Token 的累积回报

在 PPO（Proximal Policy Optimization）中，价值函数（Value Function / Critic）扮演着“基准线（Baseline）”和“预言家”的角色。
- <font color='blue'>它的核心任务是估算当前状态下，未来还能获得多少累积奖励（即 Value）。</font>

**在 LLM 的 RLHF 场景中，价值函数的训练和 SFT 或 Actor（策略模型）的训练有显著的不同。**
下面我为你详细拆解它的训练方式：
##  价值函数的目标是什么？（训练标签）
- 价值函数 V(x) 不直接预测最终的奖励（Reward），而是预测状态价值（State Value）。
- 它的训练标签（Target）是通过 GAE（Generalized Advantage Estimation，广义优势估计） 计算出来的。

假设我们在时间步 t ，价值函数的训练目标 V_{target}是：
$V_target=R_t + γV(x_{t+1})$

其中 $R_t$包含了**即时奖励**和 **KL 惩罚**。


### 关键细节：LLM 中的“稀疏奖励”问题
- 在 LLM 生成中，奖励模型（Reward Model）只在整个句子生成结束时（End of Sentence）给出一个分数，中间所有 Token 的即时奖励都是 0。

- <font color='blue'>为了让中间的 Token 也能得到训练信号，PPO 必须使用 GAE 算法，将句末的奖励沿着时间步反向传播（Bootstrap） 到每一个 Token 上。</font>
  - GAE 巧妙地结合了多步预测的方差和偏差，是目前 RL 中最标准的做法。

## 价值函数的 Loss 怎么算？
价值函数的 Loss 通常由两部分组成：

```python
Image(filename='./Image/value_loss.png')
```

### A. MSE Loss（均方误差）
这是最基础的回归 Loss，**强迫当前的价值网络 $V_θ$去逼近 GAE 算出来的 
$V_{target}$**

### B. Value Clipping（价值截断）
和 Actor 的 PPO Clip 机制非常类似，为了防止价值网络在单步更新中发生剧烈震荡，通常会限制价值函数的更新幅度：

$Clipped Value=V_{old} + clip(V_θ − V_{old},−ϵ,ϵ)$

然后计算 Clipped Value 和 $V_{target}$的 MSE，并与原始 MSE 取最小值。
- <font color='blue'>这保证了即使 GAE 算出来的 Target 波动很大，Critic 也不会“步子迈得太大”。</font>
  
## LLM 价值函数的特殊架构设计
在传统的 RL（如玩 Atari 游戏）中，Critic 和 Actor 往往是共享底层网络参数的。   - 但在 LLM 的 RLHF 中，Critic 和 Actor 通常是完全解耦（独立）的，原因如下：
- 输入输出不对等：
  - Actor 是生成模型（自回归），输出是下一个 Token 的概率分布；
  - Critic 是判别模型，输出是一个标量（Value）。
- 工程实现：
  - Critic 通常复用 Actor 的预训练权重进行初始化，但会删掉最后的 LM Head（词表投影层），替换成一个输出维度为 1 的 Linear 层。
- 显存优化：
  - 因为 Critic 不需要输出整个词表（Vocab Size 通常高达 32k~150k）的 Logits，它的显存占用远小于 Actor。
  
## 训练过程中的“坑”与优化技巧
训练 LLM 的 Critic 是非常脆弱的，实际工程中通常会采用以下手段：
- 梯度惩罚（Gradient Penalty）：为了防止 Critic 过拟合导致梯度爆炸，通常会在 Loss 中加入对 Vθ梯度的惩罚项。
- EMA（指数滑动平均）：很多框架（如 DeepSpeed-Chat）会使用一个 EMA Critic 来平滑价值估计，减少训练过程中的方差。
- KL 惩罚的加入：Critic 看到的 Reward 实际上是 $R_{model}−β⋅KL $。这意味着 Critic 不仅在拟合奖励模型，还在拟合“不要偏离参考模型太远”的约束。

## 总结：Critic 在 PPO 中的角色
如果把 PPO 比作一次“寻宝探险”：
  - Actor（策略模型）是探险家，负责决定下一步往哪走。
  - Reward Model（奖励模型）是终点处的宝藏，只有走到终点才知道有没有宝藏。
  - Critic（价值函数）则是探险家脑海中的“GPS 导航仪”。它通过过去的经验，不断评估“我现在站的位置，距离宝藏还有多远/希望有多大”。

**训练 Critic 的本质:**
  - 就是让这个 GPS 导航仪的预测越来越准。只有导航仪准了，Actor 才能根据$ Advantage=R+V(x_{t+1})−V(x_t) $准确地判断出：
    - “我刚才这一步是走对了（优势为正），还是走错了（优势为负）”，从而进行有效的梯度更新。

# 为什么 OPD、RLHF 在线强化学习统一使用反向 KL，SFT 离线蒸馏却用正向 KL？仅调整顺序差异巨大？

因为<font color='red'>两个方向的 KL 具有不同的分布拟合特性</font>。训练时需要根据具体目标，判断更需要哪种特性，再选择相应的损失函数。
  - 注：下文提到的“多模态” 不是指图像、文本模态这个多模态，而是指概率分布上的多个峰。这篇文章 1 从统一视角分析了不同散度及其对应的消息传递算法。
  
简单来说，在用 student 拟合 teacher 分布时，**正向 KL 和反向 KL 会作出不同的取舍：**
- 正向 KL 倾向于<font color='red'>覆盖</font> teacher 的各个主要模态；
- 反向 KL 倾向于集中拟合其中一部分模态，甚至收缩到单个主导模态。 这个现象也叫 <font color='red'>mode seeking</font>；
