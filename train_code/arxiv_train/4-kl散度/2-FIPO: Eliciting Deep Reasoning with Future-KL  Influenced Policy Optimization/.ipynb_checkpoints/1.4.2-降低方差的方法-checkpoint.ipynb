# 降低方差：五种核心方法详解
##### 先理解"方差"是什么
方差（Variance） 描述的是模型**对训练数据微小波动的敏感程度**。

**高方差意味着：**
- 模型过度拟合训练数据中的噪声
- 换一批训练数据，学到的模型参数变化很大
- <font color='red'>训练集表现好，测试集表现差（过拟合）</font>
  - 核心思路：降低方差 = <font color='blue'>让模型更"稳定"、更"简单"、更"鲁棒"</font>
  
## 正则化（Regularization）：L1 / L2

###  L2 正则化（Ridge / 权重衰减）
核心思想：<font color='red'>惩罚大的权重，让权重趋向于小且分散。</font>

### L1 vs L2 对比

| 特性    | L2 (Ridge)   | L1 (Lasso)         |
| ----- | ------------ | ------------------ |
| 惩罚项   | $w^2$        | $\|w\|$            |
| 解的稀疏性 | 权重趋近小但不精确为0  | 产生精确为0的权重          |
| 几何直观  | 圆形约束区域，解在边界上 | 菱形约束区域，解容易落在顶点（稀疏） |
| 适用场景  | 特征都重要，需要平滑   | 特征很多，需要自动筛选        |
| 计算    | 可导，优化方便      | 在0点不可导，需特殊处理       |


## Dropout
###  核心思想
- 训练时：以概率 p （如 0.5）随机"关闭"一部分神经元（将其输出置为 0）
- 测试时：所有神经元都参与，但输出乘以 p （或训练时除以 p ，即 Inverted Dropout）

```python
训练时某层：
  输入 → [神经元A] ─┐
       [神经元B] ──┼→ 输出（但B被随机关闭了）
       [神经元C] ─┘
```

### 为什么能降低方差？
- （1）隐式集成学习（Implicit Ensemble）
  - 每次随机丢弃不同子集，相当于在训练大量不同的"子网络"
  - 最终模型是这些子网络的平均
  - 集成平均天然降低方差（参考后面的 Bagging）
- （2）打破神经元共适应（Co-adaptation）
  - 没有 Dropout 时，神经元可能形成"小团体"：A 依赖 B，B 依赖 C
  - 有了 Dropout，每个神经元必须学会独立且鲁棒的特征表示
  - 不能"偷懒"依赖其他神经元，必须自己掌握核心信息
- （3）数据增强的等效视角
  - 随机丢弃等价于对网络结构做"数据增强"
  - 同一个样本在不同训练步骤中"看到"的是不同的网络，增加了有效训练数据的多样性
### 注意事项

| 要点      | 说明                                        |
| ------- | ----------------------------------------- |
| 丢弃率 $p$ | 通常隐藏层 0.2~0.5，输入层 0.1~0.2                 |
| 测试时行为   | 必须缩放（Inverted Dropout 在训练时除 $p$，测试时不做操作）  |
| 不适合层    | 输出层通常不用 Dropout；BatchNorm + Dropout 组合需谨慎 |


## 集成学习：Bagging（Bootstrap Aggregating）
### 核心思想
"三个臭皮匠，顶个诸葛亮"
- Bootstrap 采样：从原始训练集中有放回地随机抽取多个子数据集
- 独立训练：每个子数据集训练一个基学习器（如决策树）
- 聚合预测：分类用投票，回归用平均

### 为什么能降低方差？
#### 关键数学原理：
假设有 n  个独立的基学习器，每个的预测方差为 $σ^2$，则它们的平均预测的方差为：

```python
Image(filename='./Image/Bagging.png')
```

- 理想情况：n  个完全独立的模型，方差降为原来的 1/n 。
- 实际情况：模型之间不可能完全独立（都在同一数据集上训练），但 Bootstrap 采样增加了数据的差异性，使得模型之间的相关性降低，从而有效降低方差。

### 以随机森林（Random Forest）为例
随机森林是 Bagging 的升级版，做了两件事降低相关性：

| 技术               | 作用                   |
| ---------------- | -------------------- |
| **Bootstrap 采样** | 每个树用不同的训练子集          |
| **随机特征子集**       | 每个节点分裂时，只考虑随机选取的部分特征 |

两者结合使得每棵树都"长得不一样"，聚合时方差大幅降低。

###  Bagging vs Boosting

|            | Bagging     | Boosting          |
| ---------- | ----------- | ----------------- |
| **目标**     | 降低方差        | 降低偏差              |
| **基学习器关系** | 并行、独立       | 串行、依赖             |
| **代表算法**   | 随机森林        | XGBoost, AdaBoost |
| **数据采样**   | 有放回随机采样     | 调整样本权重            |
| **适用基学习器** | 强学习器（容易过拟合） | 弱学习器（容易欠拟合）       |


## 增加训练数据
###  核心思想
过拟合的本质是：模型记住了训练数据中的噪声。数据越多，噪声越难被"记住"。
###  为什么能降低方差？
- （1）噪声被"稀释"
  - 真实信号在更多数据中会一致地重复出现
  - 随机噪声不会一致重复，模型难以将其拟合为"规律"
- （2）决策边界更稳定
  - 数据点稀疏时，决策边界很容易被个别异常点"拉偏"
  - 数据密集时，边界由大量样本共同决定，更稳定
- （3）覆盖更完整的分布
  - 训练集更能代表真实数据分布
  - 测试集上的泛化误差更接近训练误差
###  实际限制与解决

| 问题     | 解决方案                        |
| ------ | --------------------------- |
| 标注数据昂贵 | 半监督学习、主动学习                  |
| 数据难以获取 | **数据增强**（Data Augmentation） |
| 类别不平衡  | 过采样（SMOTE）、欠采样              |

数据增强是增加训练数据的"廉价版"：
- 图像：翻转、旋转、裁剪、颜色抖动
- 文本：同义词替换、回译、随机删除
- 语音：加噪、变速、时移

## 早停（Early Stopping）
### 核心思想
训练不是越久越好。在验证集误差开始上升之前停止训练。

```python
训练过程：
训练误差 ───────────────→ 持续下降
验证误差 ───╲__╱───────→ 先降后升（过拟合开始）
              ↑
           最佳停止点
```

### 为什么能降低方差？
- （1）防止记忆噪声
  - 训练初期：模型学习的是通用模式（大权重、主要特征）
  - 训练后期：模型开始"死记硬背"训练集中的噪声和特例（小权重、细微差异）
  - 早停截断了"死记硬背"的阶段
- （2）等价于 L2 正则化（理论联系）
  - 有理论证明：早停等价于在某种意义下的 L2 正则化。
  - 训练步数少 → 权重更新次数少 → 权重不会变得特别大
  - 训练步数多 → 权重有机会增长到很大以拟合噪声
### 实现方式

```python
# 伪代码
best_val_loss = float('inf')
patience = 10  # 容忍验证集不改善的轮数
counter = 0

for epoch in range(max_epochs):
    train(model, train_loader)
    val_loss = evaluate(model, val_loader)
    
    if val_loss < best_val_loss:
        best_val_loss = val_loss
        save_checkpoint(model)  # 保存最佳模型
        counter = 0
    else:
        counter += 1
        if counter >= patience:
            print(f"Early stopping at epoch {epoch}")
            break

# 加载最佳模型
load_checkpoint(model)
```

### 注意事项

| 要点          | 说明                        |
| ----------- | ------------------------- |
| 验证集选择       | 必须与训练集、测试集都独立             |
| Patience 设置 | 太小容易欠拟合，太大失去早停意义          |
| 恢复最佳权重      | 不要直接用最后一步的权重，要恢复验证集最优时的权重 |
| 学习率调度       | 常与 ReduceLROnPlateau 配合使用 |

## 五种方法总结对比

| 方法          | 核心机制      | 降低方差原理       | 额外效果   | 主要代价          |
| ----------- | --------- | ------------ | ------ | ------------- |
| **L2 正则化**  | 惩罚大权重的平方  | 权重收缩，模型简化    | 数值稳定性好 | 需要调 $\lambda$ |
| **L1 正则化**  | 惩罚权重的绝对值  | 产生稀疏解，自动特征选择 | 可解释性强  | 优化稍复杂         |
| **Dropout** | 随机关闭神经元   | 隐式集成 + 打破共适应 | 计算开销小  | 训练变慢，需调丢弃率    |
| **Bagging** | 多模型平均     | 方差按模型数量衰减    | 天然并行   | 训练/推理成本×N     |
| **增加数据**    | 扩大训练集     | 噪声稀释，边界稳定    | 提升泛化上限 | 数据获取成本高       |
| **早停**      | 监控验证集提前终止 | 阻止记忆噪声       | 节省训练时间 | 需要验证集，可能欠拟合   |

## 实际使用建议

### 深度学习场景：
  - 必用：L2 正则化（Weight Decay，通常设 1e-4 ~ 1e-2）
  - 必用：Dropout（全连接层 0.5，CNN 中较少用）
  - 必用：早停 + 验证集监控
  - 辅助：数据增强（图像任务几乎必做）
  
### 传统机器学习（如树模型）：
  - 高方差问题 → 随机森林（Bagging）
  - 数据不足 → SMOTE 或过采样
  - 特征过多 → L1 正则化（如 Lasso、带 L1 的线性模型）
  
### 组合使用：
这些方法互不冲突，通常组合使用效果最佳。例如：
   - L2 正则化 + Dropout + 早停 + 数据增强，是现代深度学习模型的标准配置。
