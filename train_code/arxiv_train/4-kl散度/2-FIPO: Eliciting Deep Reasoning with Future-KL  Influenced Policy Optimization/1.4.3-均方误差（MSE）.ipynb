# 均方误差（MSE）
- 是衡量估计量或模型预测精度的核心指标，
- 其本质是“**预测值与真实值之差**的平方的期望”；
- 在平方损失下，MSE可被严格分解为: <font color='red'>偏差平方、方差与不可约误差三项之和</font>，这一分解为理解模型误差来源及进行偏差-方差权衡提供了统一的数学框架。

## 📌 基本定义
均方误差定义为：

MSE=$E[(y − \hat{y})^2]$
- 真实值为 y
- 模型预测值为 $\hat{y}$

即预测误差平方的期望值。

## 🔍 三项的物理含义
- 不可约误差（ $σ^2$）：数据本身的噪声，任何模型都无法消除，决定了误差的理论下限。
- 偏差平方（ $Bias^2$）：模型平均预测与真实函数的差距，反映模型的系统性偏离程度。
- 方差（ Variance ）：模型在不同训练集上预测值的波动程度，反映模型对数据扰动的敏感度。

## ⚖️ <font color='red'>偏差-方差权衡</font>
模型复杂度变化时，三项呈现此消彼长的关系：
- 模型过于简单：偏差高（欠拟合），方差低，总误差大。
- 模型过于复杂：偏差低，方差高（过拟合），总误差大。
- 最优复杂度：<font color='red'>偏差² + 方差之和最小，达到最佳泛化性能</font>



## ⚠️ 重要前提与适用范围
- 严格成立条件：该分解仅在平方损失下严格成立。对于交叉熵等其他损失函数，分解形式不同，但偏差-方差权衡的直觉依然适用。
- 估计量视角：若用于评估单一参数估计量 $\hat{θ}$（无噪声项），分解简化为 
$MSE( \hat{θ})=Bias^2 ( \hat{θ})+Var( \hat{θ} ) $。

## 📝 总结
- <font color='red'>MSE分解的核心价值在于将总误差拆解为**可解释的三个独立来源**</font>，为模型选择、正则化强度调节等提供了清晰的理论指导。
