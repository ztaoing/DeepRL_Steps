## <font color='red'>两者其实是在估计同一个梯度</font>

更严格地说，PPO 的 “**KL in reward**” 和 GRPO 的 “**KL in loss**” 都对应 <font color='red'>Reverse KL 正则，只是实现路径不同</font>。

| 形式 | 梯度来源 | 关键条件 |
|---|---|---|
| PPO: KL in reward | 通过  $ \nabla \log \pi_\theta \cdot \mathrm{KL} $  **隐式传播 **| 采样策略与优化策略一致时更干净 |
| GRPO: KL in loss | 直接对  $ \mathrm{KL}(\pi_\theta\|\pi_{\text{ref}}) $  求导 | 需要正确的 token 级估计器 |
| 等价性 | **on-policy 下两者梯度等价** | <font color='red'>off-policy 下若不补重要性采样会有偏</font> |

也就是说，<font color='red'>位置差异本质是估计量差异：</font>
  - PPO 把 KL 当作 <font color='red'>reward</font> 的一部分，让策略梯度估计器“间接”承担 KL 约束；
  - GRPO 把 KL 当作 <font color='red'>loss</font> 的一部分，让 KL 正则“直接”产生梯度；
  - 在 on-policy 条件下，两者可以证明是梯度等价的；
  - 但在 GRPO 这种 off-policy 采样、当前策略更新的设置下，<font color='red'>如果只把采样轨迹上的 KL 放进 loss，而不做重要性采样修正，就会引入偏差。</font>

## 为什么 GRPO 选择放在 loss？
工程上主要是三点：
1. 避免 reward 尺度被 KL 污染
  PPO 里 β 调不好，reward 信号会被 KL 压变形。
2. KL 梯度更稳定
  GRPO 用 token 级无偏估计，KL 波动更小，策略更新更可控。
3. 不需要 value model
  GRPO 用组内相对优势替代 critic，KL 放进 loss 后，整个目标函数更自洽。

但理论上的代价是：
  - GRPO 当前常用的 KL-in-loss 形式，严格来说只是 Reverse KL 目标的一阶近似；
  - 在 off-policy 场景下，如果不补重要性采样，它并不是完全无偏的。

所以一句话概括：
  - PPO 把 KL 放进 reward，是让 KL 通过策略梯度估计器间接影响更新；
  - GRPO 把 KL 放进 loss，是让 KL 直接产生梯度。
  - 两者在 on-policy 下等价，但在 GRPO 的 off-policy 采样下，必须依赖正确的 token 级无偏估计，否则位置变化会改变估计量本身。

# 从策略梯度的两条路径说起

如果同时读过 InstructGPT 和 DeepSeekMath 的实现，大概率注意到过一件小事：PPO-RLHF 把 KL 惩罚拌进 reward 里，用的是 k1；GRPO 把 KL 单列成一个 loss 项，用的是 k3。这两处差异经常被当成"实现风格不同"一笔带过，但把它们放在一起看会发现：位置和估计量并不是两个独立的选择——放置位置决定了梯度从哪条路径流出，而梯度路径反过来锁死了哪个估计量是合法的。 两者是绑死的，不能自由组合。本文从策略梯度本身出发，把这条因果链完整梳理一遍。00 | 拿走不送策略梯度只有两条路：score function 项和 pathwise 项。KL 放 reward 等于只走前者，放 loss 等于只走后者。走 score function 路 → 必须用 k1（无偏、token 级可加）。走 pathwise 路 → 不能用 k1，因为它沿这条路的期望梯度恒为零。k2 不是不能用。恰恰相反，它的 pathwise 梯度才是反向 KL（RLHF 的实际目标）的精确梯度。它输在读数无意义、且梯度双向无界。通用的 k3-in-loss，梯度实际对应的是正向 KL。GRPO 把 KL 放 loss 里，主因不是"省掉 critic"，而是组内均值会把 reward 里的 KL 直接减没。在 RLVR 这类可验证奖励的任务上，KL 项正在被整个删掉。01 | KL 约束在约束什么RLHF 里的奖励模型是人类偏好的代理指标，不是偏好本身。策略一旦被允许自由优化这个代理，就会去找它的漏洞——输出变得谄媚、重复、格式化，奖励分数一路走高而实际质量崩塌。这就是 reward hacking。标准对策是给策略拴一根绳：选一个可信的参考模型 （通常就是 RL 之前的 SFT 模型），要求  不要跑得太远。度量"跑多远"用 KL 散度，于是优化目标变成β 控制绳子的松紧。这个形式还有个漂亮的性质：它的解析最优解是 ，也就是参考模型的一次奖励加权重塑——DPO 那一支就是从这里推出来的。KL的方向不要搞混。 后文会反复用到（适用大模型领域）： ，反向 KL（reverse KL， 在第一位，期望也在  下）。它惩罚" 有概率而  没有"的区域，逼着  缩进参考模型的支撑集里，行为上是 mode-seeking。 ，正向 KL（forward KL），惩罚  漏掉了  的高概率区域，行为上是 mass-covering。上面那个 RLHF 标准目标用的是反向 KL。这也解释了 RLHF 常被诟病的多样性下降：mode-seeking 天然收窄分布。记住这个方向——第 5 节会得出一个略反直觉的结论。接下来的问题很具体： 上面那个目标里的 KL 项，写进代码时可以塞在两个不同的位置。它们看起来等价，实际不是。

## 位置之差：五个维度

① 梯度方差

(I) 是 score function 估计，(II) 是 pathwise 估计。同样的量，后者方差小得多——这是蒙特卡洛梯度估计里的老结论，也是两种写法最实际的一条差别。

② 信用分配放 

reward 里，第 t 步的 KL 惩罚会通过 return / GAE 传播到之前所有 token 的优势里。它的含义是"避免进入将来不得不偏离 ref 的状态"——有前瞻性，但也意味着序列开头的 token 承担了整条序列的累计偏离，量级严重失衡；而且 r < 1  时 KL 还被折扣了。

放 loss 里是逐 token 局部的，谁偏离罚谁，不传播。

同一条序列，两种方式下惩罚强度的分布是这样的：

```python
Image(filename='./Image/grpo_reward_8.jpg')
```

③ critic 的负担放

reward 里，价值网络必须去拟合这部分惩罚。麻烦在于 KL 惩罚随策略更新而变——同一个 state 的价值目标是非平稳的，会给 critic 引入额外噪声。放 loss 里完全绕开 critic。

④ β

放 reward 里，KL 会和任务奖励混在一起，被后续对奖励做的任何变换（尺度归一化、白化、baseline 相减）一并处理。这些变换是为任务奖励设计的，却会连带改变 KL 的实际强度——用户设置的 β 不等于生效的 β。放 loss 里，β 是相对于策略梯度损失的直接系数，量纲稳定、可调。这一条在 GRPO 里会变得致命问题。

⑤ off-policy 时效性

PPO 一次 rollout 要做多个 inner epoch。KL 在 reward 里是 rollout 时就算好的，衡量的是 $\pi_{old}$ 对 $\pi_{ref}$ 的偏离，内循环里不再更新，策略在 epoch 内的漂移它感知不到。KL 在 loss 里每步用当前 $\pi_\theta$ 重算，约束是实时的。

### 小结：一个略反直觉的对称性

把上述讨论汇总，出现了一个不太符合直觉的图景：

想优化第 1 节说的那个反向 KL（RLHF 的标准目标），有两条等价路径——reward 里放 k1（走 I），或 loss 里放 k2（走 II）。

而全行业实际在用的 k3-in-loss，梯度对应的是正向 KL——方向恰好相反。

需要说明的是，k3 的数值仍然是反向 KL 的无偏估计——数值和梯度对应到了两个不同方向的散度上，这本身就挺有意思。

实践中这不构成错误：两者都在  处取零、符号方向一致，作为"别跑太远"的软约束都能工作。但如果我们在论文里写"我们最小化 "而代码里是 k3-in-loss，严格说是对不上的。

```python
Image(filename='./Image/grpo_reward_17.png')
```

```python
Image(filename='./Image/grpo_reward_18.jpg')
```

KL 于是从"锚定在  附近"退化成"别比你的兄弟样本更出格"，失去了绝对锚点。策略可以整体性地一路漂走而感受不到任何回拉力。放在 loss 里就没这个问题——那是逐 token 对  的绝对度量，不经过任何 baseline 相减。

② 除以  让 β 彻底失控

是几百上千个 log-ratio 的求和，量级和方差都随长度增长，很容易远大于任务奖励本身。RLVR 场景下 ，任务奖励的方差本就很小，于是： 被 KL 的方差主导 → 有效强度变成 ，随 batch 剧烈波动；任务信号  在归一化后被压缩，训练信噪比下降。

③ 信用分配彻底消失

PPO 里 KL 进 reward 至少还经过 GAE，能得到一定的时序结构（§4 ②）。GRPO 把一个标量优势广播到序列所有 token：偏离极大的 token 和完全贴合 ref 的 token 拿到完全相同的惩罚。约束退化成纯粹的序列级平均压力。

④ 隐式长度惩罚 

是求和不是平均，长回答天然累积更多惩罚，等于往 reward 里偷偷塞了个与长度耦合的项。GRPO 本身就有长度偏置问题（Dr. GRPO 讨论过），这会叠加放大。

⑤ inner epoch 内陈旧

这条与 critic 无关，依然成立：rollout 时算好的  在多个 inner epoch 里不更新。

所以 GRPO 把 KL 放 loss 里，并不只是"省掉 critic"，更主要是避开组内归一化对约束的破坏。DeepSeekMath 原文明确提到这一点：这样 KL 不参与优势的计算，从而不干扰  的尺度。

如果确实想在 GRPO 里保留绝对锚定又不想动 loss，可选的替代是不做 std 归一化（Dr. GRPO 的做法），或者干脆去掉 KL（DAPO，见第 9 节）。

## 尾声：KL 项正在被删掉

在 RLVR（可验证奖励的推理任务）这条线上，DAPO、Dr. GRPO 等工作直接把 KL 项去掉了。

KL 约束的初衷是防 reward hacking——本质上是在给一个不可靠的代理奖励打补丁。而当奖励来自数学答案验证或代码单元测试时，它就不是代理指标了，而是真实目标，hacking 空间很小；与此同时，我们期望模型在推理模式上大幅偏离初始策略（这正是长 CoT 涌现的地方）。此时那根绳子从保险变成了拖累。

换句话说：KL 约束的必要性，与奖励信号的可靠性成反比。

这也从另一个角度说明了为什么"放哪儿、用哪个估计量"值得较真——在 KL 项真正承担约束职责的场景里，这些细节直接决定了它是在起作用，还是只是在日志里打印一个好看的数字
