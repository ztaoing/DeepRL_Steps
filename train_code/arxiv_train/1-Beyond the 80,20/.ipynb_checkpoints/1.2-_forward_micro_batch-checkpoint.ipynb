- https://github.com/Shenzhi-Wang/Beyond-the-80-20-Rule-RLVR/blob/main/verl/workers/actor/dp_actor.py#L100

- 它的主要作用是针对一个微批次（micro-batch）的数据，计算模型生成响应的对数概率（log_probs）和熵（entropy）。

```python
    def _forward_micro_batch(
        self, micro_batch, temperature, calculate_entropy=False
    ) -> tuple[torch.Tensor, torch.Tensor]:
        """
        Returns:
            entropy: # (bs, response_len)
            log_probs: # (bs, response_len)
        """
        response_length = micro_batch["responses"].size(-1)
        multi_modal_inputs = {}
        '''
        多模态输入处理：如果输入中包含 multi_modal_inputs（如图像），则调用工具函数提取并处理这些多模态数据。
        '''
        if "multi_modal_inputs" in micro_batch.keys():
            from verl.utils.model import extract_multi_modal_inputs

            multi_modal_inputs = extract_multi_modal_inputs(micro_batch["multi_modal_inputs"])

        '''
        混合精度计算：使用 torch.autocast(dtype=torch.bfloat16) 开启混合精度计算，这可以显著降低显存占用并加速计算。
        '''
        with torch.autocast(device_type=self.device_name, dtype=torch.bfloat16):
            input_ids = micro_batch["input_ids"]
            batch_size, seqlen = input_ids.shape
            attention_mask = micro_batch["attention_mask"]
            position_ids = micro_batch["position_ids"]
            entropy = None
            if position_ids.dim() == 3:  # qwen2vl mrope
                position_ids = position_ids.transpose(0, 1)  # (bsz, 4, seqlen) -> (4, bsz, seqlen)

            '''
            当 self.use_remove_padding 为 True 时，为了消除变长序列中的 padding 带来的无效计算，代码执行了以下高级优化
            
            '''
            if self.use_remove_padding:
                '''
                Unpad 操作：调用 unpad_input 将带有 padding 的输入压缩为紧凑格式（total_nnz 表示非零元素总数），
                同时记录原始索引 indices 和累积序列长度 cu_seqlens，以便后续恢复形状。
                '''
                input_ids_rmpad, indices, cu_seqlens, *_ = unpad_input(
                    input_ids.unsqueeze(-1), attention_mask
                )  # input_ids_rmpad (total_nnz, ...)
                input_ids_rmpad = input_ids_rmpad.transpose(0, 1)  # (1, total_nnz)

                # unpad the position_ids to align the rotary
                if position_ids.dim() == 3:
                    position_ids_rmpad = (
                        index_first_axis(rearrange(position_ids, "c b s ... -> (b s) c ..."), indices)
                        .transpose(0, 1)
                        .unsqueeze(1)
                    )  # (4, bsz, seqlen) -> (4, 1, bsz * seqlen)
                else:
                    position_ids_rmpad = index_first_axis(
                        rearrange(position_ids.unsqueeze(-1), "b s ... -> (b s) ..."), indices
                    ).transpose(0, 1)

                if "image_bound" in multi_modal_inputs:
                    from verl.utils.dataset.vision_utils import (
                        process_multi_modal_inputs_for_minicpmo,
                    )

                    multi_modal_inputs = process_multi_modal_inputs_for_minicpmo(
                        input_ids, attention_mask, position_ids, cu_seqlens, multi_modal_inputs
                    )

                # for compute the log_prob
                input_ids_rmpad_rolled = torch.roll(input_ids_rmpad, shifts=-1, dims=1)  # (1, total_nnz)

                '''
                Ulysses 序列并行（SP）：如果开启了 use_ulysses_sp，
                会将紧凑后的输入在序列维度上进行切分和填充（ulysses_pad_and_slice_inputs），
                从而将长序列的计算分摊到多个设备上。
                '''
                # pad and slice the inputs if sp > 1
                if self.use_ulysses_sp:
                    is_vlm_model = hasattr(
                        getattr(self.actor_module, "module", self.actor_module).config, "vision_config"
                    )
                    if is_vlm_model:
                        # vlm model's inputs will be sliced after embedding
                        input_ids_rmpad, position_ids_rmpad, pad_size = ulysses_pad(
                            input_ids_rmpad,
                            position_ids_rmpad=position_ids_rmpad,
                            sp_size=self.ulysses_sequence_parallel_size,
                        )
                    else:
                        input_ids_rmpad, position_ids_rmpad, pad_size = ulysses_pad_and_slice_inputs(
                            input_ids_rmpad,
                            position_ids_rmpad=position_ids_rmpad,
                            sp_size=self.ulysses_sequence_parallel_size,
                        )
                    input_ids_rmpad_rolled, _, _ = ulysses_pad_and_slice_inputs(
                        input_ids_rmpad_rolled,
                        position_ids_rmpad=None,
                        sp_size=self.ulysses_sequence_parallel_size,
                    )

                input_ids_rmpad_rolled = input_ids_rmpad_rolled.squeeze(0)  # ((total_nnz / sp) + pad)

                # only pass input_ids and position_ids to enable flash_attn_varlen
                extra_args = {}
                if self.use_fused_kernels:
                    extra_args["temperature"] = temperature
                    extra_args["return_dict"] = True

                '''
                模型前向传播：
                将处理后的 input_ids_rmpad 传入模型。这里传入了 attention_mask=None，
                因为紧凑格式配合 flash_attn_varlen（变长 Flash Attention）不需要传统的 mask。
                
                '''
                output = self.actor_module(
                    input_ids=input_ids_rmpad,
                    attention_mask=None,
                    position_ids=position_ids_rmpad,
                    **multi_modal_inputs,
                    use_cache=False,
                    **extra_args,
                )  # prevent model thinks we are generating

                if self.use_fused_kernels:
                    '''
                    如果使用了融合算子（use_fused_kernels），模型会直接输出 log_probs 和 entropy。
                    '''
                    log_probs = output.log_probs.squeeze(0)  # (total_nnz,)
                    entropy_rmpad = output.entropy.squeeze(0)  # (total_nnz,)

                else:
                    '''
                    否则，取出 logits 并除以 temperature，然后调用 logprobs_from_logits 计算对数概率。如果要求计算熵，还会通过 compute_entropy_from_logits 计算熵值。
                    '''
                    logits_rmpad = output.logits.squeeze(0)  # (total_nnz, vocab_size)
                    logits_rmpad.div_(temperature)

                    # if use_sp: ((total_nnz / sp) + pad) ; if not use_sp: (batch, seqlen)
                    inplace_backward = True
                    if calculate_entropy:
                        inplace_backward = False
                    log_probs = logprobs_from_logits(
                        logits=logits_rmpad,
                        labels=input_ids_rmpad_rolled,
                        inplace_backward=inplace_backward,
                    )

                    # compute entropy
                    if calculate_entropy:
                        if not self.config.entropy_checkpointing:
                            entropy_rmpad = self.compute_entropy_from_logits(logits_rmpad)  # ((total_nnz / sp) + pad)
                        else:
                            entropy_rmpad = torch.utils.checkpoint.checkpoint(
                                self.compute_entropy_from_logits, logits_rmpad
                            )

                # gather log_prob if sp > 1
                if self.use_ulysses_sp:
                    # gather and unpad for the ulysses sp
                    log_probs = gather_outputs_and_unpad(
                        log_probs,
                        gather_dim=0,
                        unpad_dim=0,
                        padding_size=pad_size,
                    )
                    if calculate_entropy:
                        entropy_rmpad = gather_outputs_and_unpad(
                            entropy_rmpad,
                            gather_dim=0,
                            unpad_dim=0,
                            padding_size=pad_size,
                        )
                # pad back to (bsz, seqlen)
                if calculate_entropy:
                    full_entropy = pad_input(
                        hidden_states=entropy_rmpad.unsqueeze(-1),
                        indices=indices,
                        batch=batch_size,
                        seqlen=seqlen,
                    )
                full_log_probs = pad_input(
                    hidden_states=log_probs.unsqueeze(-1),
                    indices=indices,
                    batch=batch_size,
                    seqlen=seqlen,
                )

                # only return response part:
                if calculate_entropy:
                    entropy = full_entropy.squeeze(-1)[:, -response_length - 1 : -1]  # (bsz, response_length)
                log_probs = full_log_probs.squeeze(-1)[:, -response_length - 1 : -1]  # (bsz, response_length)

            else:  # not using rmpad and no ulysses sp
                extra_args = {}
                if self.use_fused_kernels:
                    extra_args["temperature"] = temperature
                    extra_args["return_dict"] = True

                output = self.actor_module(
                    input_ids=input_ids,
                    attention_mask=attention_mask,
                    position_ids=position_ids,
                    **multi_modal_inputs,
                    use_cache=False,
                    **extra_args,
                )  # prevent model thinks we are generating

                if self.use_fused_kernels:
                    log_probs = output.log_probs[:, -response_length - 1 : -1]
                    entropy = output.entropy[:, -response_length - 1 : -1]  # (bsz, response_length)

                else:
                    logits = output.logits

                    logits.div_(temperature)
                    logits = logits[:, -response_length - 1 : -1, :]  # (bsz, response_length, vocab_size)
                    log_probs = logprobs_from_logits(logits, micro_batch["responses"])
                    if calculate_entropy:
                        if not self.config.entropy_checkpointing:
                            entropy = verl_F.entropy_from_logits(logits)  # (bsz, response_length)
                        else:
                            entropy = torch.utils.checkpoint.checkpoint(verl_F.entropy_from_logits, logits)

            return entropy, log_probs
```

# DataParallelPPOActor._forward_micro_batch()
- 负责计算每个 token 的 log_prob 和 entropy。这是 verl 框架中非常关键的前向传播函数

## 函数签名与返回值

```python
def _forward_micro_batch(
    self, 
    micro_batch,           # 输入数据
    temperature,            # 解码温度
    calculate_entropy=False # 是否计算熵
) -> tuple[torch.Tensor, torch.Tensor]:
    """
    Returns:
        entropy:   # (bs, response_len)  每个 response token 的熵
        log_probs: # (bs, response_len)  每个 response token 的 log_prob
    """
```

## 整体流程

```python
┌─────────────────────────────────────────────────────────────┐
│              _forward_micro_batch 整体流程                    │
├─────────────────────────────────────────────────────────────┤
│                                                             │
│  1. 提取输入数据                                            │
│     ├── input_ids: (bsz, seqlen)                           │
│     ├── attention_mask: (bsz, seqlen)                      │
│     ├── position_ids: (bsz, seqlen) 或 (bsz, 4, seqlen)   │
│     └── responses: (bsz, response_len)                     │
│                                                             │
│  2. 条件分支                                                │
│     ├── 使用 remove_padding (rmpad) + Ulysses SP            │
│     │   └── 复杂路径：去填充 → 并行计算 → 重填充            │
│     └── 不使用 rmpad                                        │
│         └── 简单路径：直接前向传播                          │
│                                                             │
│  3. 计算 log_prob 和 entropy                               │
│     ├── 使用 fused kernels（高效路径）                      │
│     └── 不使用 fused kernels（通用路径）                    │
│                                                             │
│  4. 切片：只返回 response 部分                              │
│     └── [:, -response_length-1 : -1]                      │
│                                                             │
└─────────────────────────────────────────────────────────────┘
```

## 详细解析
### 输入数据准备

```python
response_length = micro_batch["responses"].size(-1)  # response 长度

# 多模态输入处理（如 VLM）
multi_modal_inputs = {}
if "multi_modal_inputs" in micro_batch.keys():
    multi_modal_inputs = extract_multi_modal_inputs(micro_batch["multi_modal_inputs"])
```

### 主分支：使用 Remove Padding（rmpad）
这是高效训练路径，用于处理变长序列：

#### 去填充（Unpad）

```python
# 原始 input_ids: (bsz, seqlen)，包含 padding
input_ids_rmpad, indices, cu_seqlens, *_ = unpad_input(
    input_ids.unsqueeze(-1), attention_mask
)
# input_ids_rmpad: (1, total_nnz)  total_nnz = 非 padding token 总数
```

##### 为什么去填充？
- 变长序列直接 batch 会有大量 padding
- Flash Attention 等 kernel 对变长序列支持不好
- 去填充后可以用 flash_attn_varlen 高效计算

#### 位置编码处理

```python
# 标准位置编码: (bsz, seqlen)
# Qwen2VL mrope: (bsz, 4, seqlen) → 转置为 (4, bsz, seqlen)
if position_ids.dim() == 3:
    position_ids = position_ids.transpose(0, 1)
```

#### 标签偏移（用于计算 log_prob）

```python
# 将 input_ids 左移一位，得到预测目标
input_ids_rmpad_rolled = torch.roll(input_ids_rmpad, shifts=-1, dims=1)
# 例如: [A, B, C, D, pad] → [B, C, D, pad, A]
# 这样 logits[i] 预测的是 input_ids[i+1]
```

#### Ulysses 序列并行（可选）
- Ulysses SP：将长序列切分到多个 GPU，每个 GPU 处理一部分，最后 gather 结果。


```python
if self.use_ulysses_sp:
    # 将序列切分到多个 GPU 上并行计算
    input_ids_rmpad, position_ids_rmpad, pad_size = ulysses_pad_and_slice_inputs(
        input_ids_rmpad,
        position_ids_rmpad=position_ids_rmpad,
        sp_size=self.ulysses_sequence_parallel_size,
    )
```

#### 模型前向传播

```python
output = self.actor_module(
    input_ids=input_ids_rmpad,        # 去填充后的输入
    attention_mask=None,              # Flash Attention 不需要 mask
    position_ids=position_ids_rmpad, # 位置编码
    use_cache=False,                  # 训练模式，不使用 KV cache
    **extra_args,                     # 可能包含 temperature
)
```

#### 计算 log_prob 和 entropy
##### 路径 A：使用 Fused Kernels（最高效）
- 这里模型直接返回 log_probs 和 entropy，省去了手动计算。

```python
if self.use_fused_kernels:
    log_probs = output.log_probs.squeeze(0)   # (total_nnz,)
    entropy_rmpad = output.entropy.squeeze(0) # (total_nnz,)
```

##### 通用计算

```python
else:
    logits_rmpad = output.logits.squeeze(0)  # (total_nnz, vocab_size)
    logits_rmpad.div_(temperature)           # 应用温度缩放
    
    # 计算 log_prob: log softmax + gather
    log_probs = logprobs_from_logits(
        logits=logits_rmpad,
        labels=input_ids_rmpad_rolled,  # 目标 token
    )
    
    # 计算熵（如果需要）
    if calculate_entropy:
        entropy_rmpad = self.compute_entropy_from_logits(logits_rmpad)
```

###### logprobs_from_logits 的实现：

```python
def logprobs_from_logits(logits, labels):
    """
    logits: (total_nnz, vocab_size)
    labels: (total_nnz,) 目标 token IDs
    
    Returns: (total_nnz,) 每个位置的 log_prob
    """
    log_probs = F.log_softmax(logits, dim=-1)           # (total_nnz, vocab_size)
    token_log_probs = torch.gather(                     # 提取目标 token 的 log_prob
        log_probs, 
        dim=-1, 
        index=labels.unsqueeze(-1)
    ).squeeze(-1)                                       # (total_nnz,)
    return token_log_probs
```

###### compute_entropy_from_logits 的实现：

```python
def compute_entropy_from_logits(logits):
    """
    logits: (total_nnz, vocab_size)
    
    Returns: (total_nnz,) 每个位置的熵
    """
    probs = F.softmax(logits, dim=-1)                   # (total_nnz, vocab_size)
    log_probs = F.log_softmax(logits, dim=-1)           # (total_nnz, vocab_size)
    entropy = -torch.sum(probs * log_probs, dim=-1)     # (total_nnz,)
    return entropy
```

#### Gather 结果（如果用了 SP）

```python
if self.use_ulysses_sp:
    # 从多个 GPU gather 结果
    log_probs = gather_outputs_and_unpad(log_probs, ...)
    if calculate_entropy:
        entropy_rmpad = gather_outputs_and_unpad(entropy_rmpad, ...)
```

#### 重填充（Pad Back）

```python
# 将去填充的结果重新填充回 (bsz, seqlen)
full_log_probs = pad_input(
    hidden_states=log_probs.unsqueeze(-1),
    indices=indices,           # 去填充时的索引
    batch=batch_size,
    seqlen=seqlen,
)  # (bsz, seqlen, 1)
```

#### 切片：只取 response 部分

```python
# 只返回 response 部分的 log_prob 和 entropy
# -response_length-1 : -1 表示倒数 response_length+1 到倒数第1个
log_probs = full_log_probs.squeeze(-1)[:, -response_length - 1 : -1]
# (bsz, response_length)

if calculate_entropy:
    entropy = full_entropy.squeeze(-1)[:, -response_length - 1 : -1]
    # (bsz, response_length)
```

##### 为什么切片范围是 [-response_length-1 : -1]？

```python
序列: [prompt_token1, prompt_token2, ..., prompt_tokenN, response_token1, response_token2, ..., response_tokenM]

索引:  0              1              ...  N-1            N              N+1            ...  N+M-1

logits 预测的是下一个 token:
  logits[0] 预测 token1 → 实际是 token1
  logits[N-1] 预测 tokenN → 实际是 response_token1
  logits[N] 预测 response_token1 → 实际是 response_token2
  ...
  logits[N+M-1] 预测 response_tokenM → 实际是 <EOS>

我们需要的是 response_token1 到 response_tokenM 的 log_prob:
  对应 logits 的索引: N-1 到 N+M-2
  
  即: [-response_length-1 : -1] = [N-1 : N+M-1-1] = [N-1 : N+M-2]
```

### 分支 B：不使用 Remove Padding
这是简单路径，直接前向传播：

```python
else:  # not using rmpad and no ulysses sp
    output = self.actor_module(
        input_ids=input_ids,           # (bsz, seqlen)
        attention_mask=attention_mask,  # (bsz, seqlen)
        position_ids=position_ids,
        use_cache=False,
        **extra_args,
    )
    
    if self.use_fused_kernels:
        log_probs = output.log_probs[:, -response_length - 1 : -1]
        entropy = output.entropy[:, -response_length - 1 : -1]
    else:
        logits = output.logits  # (bsz, seqlen, vocab_size)
        logits.div_(temperature)
        
        # 切片到 response 部分
        logits = logits[:, -response_length - 1 : -1, :]
        # (bsz, response_length, vocab_size)
        
        # 计算 log_prob
        log_probs = logprobs_from_logits(logits, micro_batch["responses"])
        # (bsz, response_length)
        
        # 计算熵
        if calculate_entropy:
            entropy = verl_F.entropy_from_logits(logits)
            # (bsz, response_length)
```

## 关键设计点
###  Remove Padding（rmpad）

#### 优势：
- 避免 padding token 的无效计算
- Flash Attention Varlen 对变长序列更高效
- 减少内存占用

```python
原始序列（变长）:
  Batch 1: [A, B, C, pad, pad]      seqlen=5, actual=3
  Batch 2: [D, E, F, G, H]          seqlen=5, actual=5
  Batch 3: [I, J, pad, pad, pad]    seqlen=5, actual=2

去填充后:
  [A, B, C, D, E, F, G, H, I, J]   total_nnz=10

Flash Attention Varlen 计算:
  使用 cu_seqlens = [0, 3, 8, 10] 标记每个序列的边界

重填充后:
  恢复原始 (bsz, seqlen) 形状
```

### Ulysses 序列并行

```python
序列长度=8192, 4个GPU:

不使用 SP:
  每个 GPU 处理完整的 8192 长度
  → 内存压力大

使用 SP:
  每个 GPU 处理 8192/4 = 2048 长度
  → 内存压力小，可以处理更长序列

计算后 gather:
  从4个GPU收集结果，拼接回完整序列
```

### Fused Kernels

```python
不使用 Fused Kernel:
  1. 模型输出 logits
  2. 手动计算 log_softmax
  3. 手动 gather 目标 token 的 log_prob
  4. 手动计算 entropy

使用 Fused Kernel:
  模型直接返回 log_probs 和 entropy
  → 减少内存搬运，提高速度
```

## 与论文《Beyond the 80/20 Rule》的关联
这段代码是论文核心机制的关键支撑：

### 关键洞察：
- _forward_micro_batch 计算的 entropy 是 token 级别 的
- 每个位置 (b,t)  的 entropy[b, t] 代表该 token 的生成不确定性
- <font color='red'>高熵 token 就是论文中的 forking tokens（决策点）</font>
  - 通过 entropy_top_mask 筛选，实现"只更新关键决策点"

```python
_forward_micro_batch 返回:
├── log_probs: (bsz, response_len)  ← 用于 PPO ratio 计算
└── entropy: (bsz, response_len)      ← 用于 forking tokens 筛选

                    ↓

update_policy 中:
├── entropy_top_mask = get_global_entropy_top_mask(entropy, response_mask, top_ratio=0.2)
│   └── 只保留 top-20% 高熵 token
│
└── policy_loss_fn(..., entropy_top_mask=entropy_top_mask)
    └── 只更新高熵 token 的策略梯度
```

## 完整数据流

```python
输入:
  input_ids: (bsz, seqlen) = [prompt + response + padding]
  
        ↓ 模型前向传播

输出 logits: (bsz, seqlen, vocab_size)
  
        ↓ 计算 log_prob 和 entropy

log_probs: (bsz, seqlen)  每个位置的 log_prob
entropy: (bsz, seqlen)    每个位置的熵（如果 calculate_entropy=True）

        ↓ 切片（只取 response 部分）

log_probs: (bsz, response_len)
entropy: (bsz, response_len)

        ↓ 返回给 update_policy

在 update_policy 中:
  ├── 用 log_probs 计算 PPO ratio
  ├── 用 entropy 筛选 forking tokens
  └── 只更新高熵 token 的策略
```

## 总结

| 组件                | 作用                                 |
| ----------------- | ---------------------------------- |
| **rmpad**         | 去填充，高效处理变长序列                       |
| **Ulysses SP**    | 序列并行，支持超长序列                        |
| **Fused Kernels** | 融合计算，减少内存搬运                        |
| **Temperature**   | 控制生成随机性                            |
| **Entropy**       | 衡量 token 级不确定性，用于筛选 forking tokens |
| **Log\_prob**     | 计算 PPO ratio，用于策略梯度                |
