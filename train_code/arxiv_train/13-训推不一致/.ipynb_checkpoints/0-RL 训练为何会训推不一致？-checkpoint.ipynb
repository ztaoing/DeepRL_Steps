# 先搞清楚：大模型 RL 到底在训什么

大模型 RL 训练三步循环：Rollout → Reward → Training，<font color='red'>其中 Rollout 往往占总时间的 60–80%</font>

```python
Image(filename='./Image/rl_02.png')
```

大多数人熟悉的"预训练"是拿大量文本让模型学会预测下一个 Token。强化学习（RL）干的事不一样：<font color='red'>它要让模型学会在特定任务上产出更好的结果</font>——比如解数学题、写代码、按人类偏好回答问题。


RL 的核心循环可以理解成三个步骤不断重复：
1. Rollout（推理生成）：<font color='red'>用当前版本的模型（Policy）生成一批答案</font>。比如给 1000 道数学题，每题让模型生成多个解法。
2. Reward（评分）：对每个答案打分。数学题可以自动验证是否正确；聊天问答则可能用奖励模型或人工打分。
3. Training（训练更新）：<font color='red'>根据 Reward 信号调整模型参数</font>——让高分答案的生成概率上升，低分答案的概率下降。听起来简单。但当模型是 300B 参数的庞然大物，Rollout 要生成 8000 个 Token 的长答案，整个循环每一秒都跑在成百上千张 GPU 上时，每一个环节都会变成一道工程难题。

反直觉事实： RL 训练中，<font color='blue'>真正的时间杀手往往不是"训练"本身，而是"推理生成"阶段（Rollout）</font>。
 - DeepSeek-R1 论文中的实验数据显示，在某些配置下 Rollout 占据了整个训练时间的 60%–80%。
   - <font color='red'>训练节点在等推理节点出结果，推理节点在等长尾样本结束，长尾样本在等模型一个 Token 一个 Token 地解码——时间就这样一层层地浪费掉了。</font>

# RL 的时间为什么耗在推理侧

大模型 RL 的一个反直觉事实是：训练并不一定是最慢的部分。大量时间花在 Rollout，也就是让当前 Policy 生成 Response、调用环境、获得 Reward，再把数据送回训练。<font color='red'>长尾样本会拖住整个 Batch。短 Response 已经结束，少数长 Response 仍在逐 Token Decode</font>；同步系统只能等最慢样本结束，才能开始下一轮更新。

想象一个 100 人的接力跑：99 个人已经跑完，只有 1 个人还在路上，全队只能站在终点等。这就是同步 Rollout 的代价。

图 1：同步 Rollout 中，1 个长尾样本（~3000 tokens）让其余 7 个已完成的短样本 GPU 全程空转等待。

```python
Image(filename='./Image/rl_03.png')
```

RL 系统要同时处理模型接入、资源管理、参数传输和 Agent Control，典型架构可以分成七层：
  - 上层框架（PyTorch/Ray）、
  - 通信层（NCCL/Gloo）、
  - 训练后端（FSDP 系列）、
  - 算法层（PPO/GRPO）、
  - 推理引擎（vLLM/SGLang）、
  - 权重同步层和模型层。

任何一层的接口不稳定，都会让新模型接入变成一次全栈改造。
- 以 veRL 为代表的开源框架，方向是把训练、推理、权重转换和数据传输**"引擎化"**，通过接口按需装配，让上层算法研究者不需要关心底层拼接逻辑。

# FSDP Turbo 如何压长序列成本

训练长序列时，<font color='red'>内存是第一道关卡</font>。一个简单的等式：序列越长，Attention 的激活内存占用越大
  - 8K 序列的激活比 2K 序列多 4 倍。

FSDP（Fully Sharded Data Parallel）是 PyTorch 的分布式训练核心机制：
 - 把模型参数切分到多张 GPU 上，需要时再临时 All-Gather 到本地做计算。
 - FSDP Turbo 在其基础上提供统一入口，<font color='blue'>**一次调用完成**</font><font color='red'>运行状态初始化、量化转换、多策略并行和 FSDP 分片，兼容 TP、CP、重计算、Chunked Loss、Activation Swap、MXFP8 和 FP8</font>

图 2：统一入口在运行时装配并行、量化、内存优化和 FSDP。图源：会议现场实拍。

```python
Image(filename='./Image/rl_04.jpeg')
```

长序列的第一个问题是**激活显存**。完全重计算可以省显存，却会重复执行昂贵的 FlashAttention。
- 现场方案把前向的输入输出暂存到 CPU，反向前再 Prefetch 回 Device，从而跳过这部分重计算。

在 FSDP + EP（Expert Parallelism）场景，还需要把参数 Prefetch、Activation Offload 和通信重叠起来。只把张量搬到 CPU 并不会自动变快；若 PCIe 传输挡在关键路径上，节省的显存会变成新的等待。

关键技巧：Micro Chunk 流水线。 
- <font color='red'>把一个 Micro Batch 切成多个 Micro Chunk</font>，当前 Layer 计算一个 Chunk 时，异步处理其他 Chunk 的激活和通信，把峰值控制在单个微块。现场记录的收益约为 4%–5%，代价是代码复杂度提升。

一个具体例子
假设 mini-batch 有 8 个样本，先切成 2 个 micro-batch，每个 micro-batch 再切成 4 个 chunk：

- 这样相比粗粒度 micro-batch，通信更早开始、反向更早触发、激活缓存窗口更短

```python
mini-batch: 8 samples
 ├─ micro-batch A: 4 samples
 │    ├─ chunk A1: 1 sample
 │    ├─ chunk A2: 1 sample
 │    ├─ chunk A3: 1 sample
 │    └─ chunk A4: 1 sample
 └─ micro-batch B: 4 samples
      ├─ chunk B1: 1 sample
      ├─ chunk B2: 1 sample
      ├─ chunk B3: 1 sample
      └─ chunk B4: 1 sample
```

# 异步 Flow 怎样消灭等待

同步 RL 的时间线很规整，也很浪费：
  - 所有推理节点用最新权重生成数据，训练等待推理结束；
  - 训练更新完成后，所有推理节点又一起等待权重同步。

就好比餐厅服务员做法是：<font color='red'>所有桌子都点完菜，厨房才开始做；所有菜都上完，才通知下一桌点菜。显然效率极低。</font>

<font color='red'>异步 Flow 在中间增加 CPU 权重暂存层（TransferQueue）。训练完成更新后，先把新权重放进 CPU；某个推理节点完成当前生成后，再加载新版本，不必让所有节点同时停下来。</font>


图 4：权重通过 TransferQueue 异步流动，训练与推理不再严格串行。图源：会议现场实拍。

```python
Image(filename='./Image/rl_05.jpeg')
```

下一步还可以感知各推理节点的生产速率，轮换更新，而不是整组切换。对长尾样本，则把它们集中到少数节点，释放其他节点进入下一批 Rollout。

图 5：同步时间线与异步时间线的等待对比。图源：会议现场实拍。


```python
Image(filename='./Image/rl_06.jpeg')
```

现场测试中，Flow 相对同步方案获得约 3.8 倍提升。这个数字的含义不是单个 Kernel 快了 3.8 倍，而是训练、权重传输和推理之间的空闲被压缩。

异步的代价： 
- 不同样本由不同版本 Policy 产生，这会引入"数据分布漂移"。<font color='red'>系统必须记录每个样本对应的 Policy Version（权重版本号）</font>，并限定 Staleness（<font color='blue'>最大允许的版本落后程度</font>），否则吞吐提升可能换来训练不稳定。根据 veRL AMD 博客的实验数据，适当限制 Staleness 后，异步模式的 KL Divergence 约为 0.001，**属于可控范围**。

# 投机推理如何缩短单样本 Decode

<font color='red'>异步 Flow 减少了系统级等待，但没有缩短一个长 Response 本身的生成时间</font>。
- 如果让模型回答一道复杂推理题，输出 3000 个 Token，每个 Token 都要跑一遍完整的大模型——这笔时间账异步机制无法抹去。

投机推理（Speculative Decoding）针对的正是这个问题。基本思路是"先猜后验证"：

- ①用一个轻量草稿模型（Draft Model，参数量是主模型的几十分之一）先快速连续生成 4–8 个 Token。

- ②主模型一次并行验证这些草稿 Token，计算量和验证 1 个 Token 差不多，但可以一次接受多个正确的 Token。

- ③遇到第一个错误 Token，由主模型修正，然后继续让草稿模型猜下一段。

<font color='red'>按照论文 DAS（Draft-then-Accept Speculative Decoding）的实验结果，使用后缀树（Suffix Tree）作为非参数化草稿器，可以将 Rollout 时间减少最多 50%，且不改变主模型的输出分布（理论上与贪心解码等价）。</font>

图 6：草稿模型快速生成多个 Token，主模型并行验证，接受正确前缀并在第一个错误位置修正——DAS 论文实测最多节省 50% Rollout 时间。

```python
Image(filename='./Image/rl_07.png')
```

## <font color='red'>现场讨论了几条草稿来源路径：</font>

- 历史文本检索（Retrieval）：用 Hash Table 匹配历史生成过的上下文，无需草稿模型，但 IO 和查表是固定成本。主模型越快，这笔固定成本占比越高。

- MTP（Multi-Token Prediction）：在模型结构中加入多 Token 预测 Head，训练时额外学习预测后续几步，推理时直接当草稿用。

- Eagle 3：需要专门预训练的轻量草稿模型，接收率取决于草稿质量与主模型的契合程度。

- DeepFlash：用扩散方式一次产生多个 Token 草稿，再由多个轻量草稿头修正。

草稿模型是否参与 RL 训练存在争议。
- 参与训练可以跟上主模型更新、提高接收率，却占用额外显存和计算；
- 不参与训练成本低，但 Policy 变化后可能迅速过时。
- 所以投机推理必须看三个指标：接收率、草稿开销、主模型验证节省——只看 Decode Token/s 很容易把查表或草稿成本藏起来。

# 训推一致性为什么是最后一关

假设性能问题全部解决了，训练跑得很快，Rollout 也很快——但最终发现 Reward 曲线迟迟不涨，或者涨一段时间后莫名崩溃。这时候最有可能的嫌疑犯，是训推不一致。

## 什么是训推不一致？
- 简单说，就是同一段文本，在训练时计算出的 Log Probability（每个 Token 的生成概率）和在推理时计算出的结果不一样。**而 RL 算法（如 PPO、GRPO）的梯度更新依赖这个概率**，若训推两侧用的不是同一个概率，更新方向就会出错。

### 根本原因：浮点不结合律。 
- 数学上，(a+b)+c = a+(b+c)，<font color='red'>但计算机浮点数不满足这条规律</font>。
  - 举个例子：把同样的 100 个数相加，先加前 50 个再加后 50 个，和一次性全加，结果可能在小数点后几位有细微差异。这个差异听起来很小，但在大模型里，相同的输入放进不同 Batch Size，MatMul 会因为矩阵切分方式不同（K 轴切分数量变化），导致浮点累加顺序变化，最终产生不同的概率值。


训练和推理经常使用不同算子实现：训练端用了 Hadamard 变换，推理端没有；训练用 SwiGLU 融合算子，推理用多个小算子——<font color='red'>现场观察到差异可能达到约 10%</font>

MoE（混合专家模型）更敏感。<font color='red'>因为微小的数值差异就可能改变 Router 对专家的选择</font>，随后误差被整条生成轨迹放大，**因此 MoE 需要 Routing Replay**——<font color='blue'>推理时记录下路由决策，训练时严格重放，保持路由一致。</font>
- 根据 LMSYS Miles 博客的实测数据，Dense 模型的训推 KL Divergence 约为 10⁻⁵～10⁻³，MoE 模型则可能高达 10⁻³～10⁻¹。

做 RL Infra 优化时，我会按这个顺序验收：
1. 先比较训练和推理的 Log Prob，而不是先看 Reward。
2. <font color='red'>固定一个样本，跨 Batch Size、精度和算子实现重放，看结果是否一致</font>
3. 再测 Rollout、权重同步和训练的时间线，找真正等待点。
4. 最后引入异步和投机推理，并记录 Policy Version 与 Staleness。

性能优化像加快发牌。一致性决定了你发出的，还是不是同一副牌。
