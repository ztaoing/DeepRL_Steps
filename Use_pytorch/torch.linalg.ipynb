### torch.linalg

torch.linalg 是 PyTorch 中的一个模块，提供了一系列用于线性代数运算的函数。
- 这些函数涵盖了矩阵分解、矩阵求解、特征值和特征向量计算、行列式计算、矩阵的迹计算等常见的线性代数操作。
- 使用 torch.linalg 可以方便地进行各种矩阵和张量的线性代数运算。

以下是一些常用的 torch.linalg 函数及其简要说明：
- torch.linalg.eig：
  - 计算方阵的特征值和特征向量。
  - 返回值：特征值和特征向量。
- torch.linalg.eigvals：
  - 计算方阵的特征值。
  - 返回值：特征值。
- torch.linalg.svd：
  - 计算矩阵的奇异值分解（SVD）。
  - 返回值：U, S, Vh。
- torch.linalg.qr：
  - 计算矩阵的 QR 分解。
  - 返回值：Q, R。
- torch.linalg.solve：
  - 解线性方程组 AX=B。
  - 返回值：解矩阵 X。
- torch.linalg.inv：
  - 计算方阵的逆矩阵。
  - 返回值：逆矩阵。
- torch.linalg.det：
  - 计算方阵的行列式。
  - 返回值：行列式的值。
- torch.linalg.trace：
  - 计算方阵的迹。
  - 返回值：迹的值。
- torch.linalg.norm：
  - 计算矩阵或向量的范数。
  - 返回值：范数的值。
- torch.linalg.matrix_rank：
  - 计算矩阵的秩。
  - 返回值：秩的值。
