torch.apply 是 PyTorch 中的一个非常有用的方法，它允许你对 nn.Module 的所有子模块递归地应用一个函数。这在深度学习中非常有用，尤其是在需要对模型的每个子模块进行操作时，例如初始化参数、修改参数、打印模块信息等

- torch.apply 是对模块的参数和属性进行操作，而不是对输入数据进行操作。如果你需要对输入数据进行操作，应该在 forward 方法中实现

torch.apply 的主要作用是：
- 递归应用函数：对 nn.Module 的所有子模块递归地应用一个函数。
- 灵活性：可以对每个子模块进行自定义操作，例如初始化参数、修改参数、打印模块信息等

```python
torch.nn.Module.apply(fn)
# fn：一个函数，将被递归地应用到每个子模块上。
```

```python
import torch
import torch.nn as nn

def init_weights(m):
    if isinstance(m, nn.Linear):
        torch.nn.init.xavier_uniform_(m.weight)
        m.bias.data.fill_(0.01)

# 定义一个简单的神经网络
model = nn.Sequential(
    nn.Linear(10, 5),
    nn.ReLU(),
    nn.Linear(5, 2)
)

# 应用初始化函数
model.apply(init_weights)
```

### 打印模块信息
- 假设你想要打印模型中每个子模块的信息，可以使用 torch.apply：

```python
def print_module_info(m):
    print(m)

# 应用打印函数
model.apply(print_module_info)
```

### 修改参数
- 假设你想要将模型中所有权重的值乘以一个常数，可以使用 torch.apply

```python
def scale_weights(m):
    if isinstance(m, nn.Linear):
        m.weight.data *= 2.0

# 应用修改函数
model.apply(scale_weights)
```

torch.apply 会返回模块本身，这使得你可以链式调用多个 apply 方法

```python
model.apply(init_weights).apply(print_module_info)
```
