- 是PyTorch中将张量注册为模型可训练参数的核心操作。

```python
import torch
import torch.nn as nn


class MyModel(nn.Module):
    def __init__(self):
        super().__init__()
        
        # 普通张量 - 不会被优化器更新
        self.regular_tensor = torch.randn(10, 20)
        
        # Parameter - 会被优化器更新
        self.weight_param = nn.Parameter(torch.randn(10, 20))
        
        # 普通张量但requires_grad=True - 仍不会被优化器更新
        self.requires_grad_tensor = torch.randn(10, 20, requires_grad=True)

model = MyModel()
optimizer = torch.optim.SGD(model.parameters(), lr=0.01)

# 查看哪些参数被优化器管理
print(list(model.parameters()))


param = nn.Parameter(torch.randn(3, 4))

# 1. 自动设置 requires_grad=True
print(param.requires_grad)  # True

# 2. 在模型参数列表中
# print(param in model.parameters())  # True

# 3. 可以被优化器更新
optimizer = torch.optim.SGD([param], lr=0.01)

# 4. 支持梯度计算
loss = param.sum()
loss.backward()
print(param.grad)  # 有梯度值
```
