```python
import torch
ce = torch.nn.CrossEntropyLoss(ignore_index=-100)
logits = torch.randn(2,5,50256)

labels = torch.full((2,5),-100,dtype=torch.long)
print(labels)

# 在第 0 条样本、第 2~3 号位置（共 2 个 token）上，把原来被设成 -100 的占位符改成真正需要模型去预测的词 ID。
# 它把“只有这两个位置才参与 loss 计算”的示例具体化，其余位置仍保持 -100，从而验证 CrossEntropyLoss(ignore_index=-100) 是否真的只统计了这俩 token 的误差
labels[0,2:4] = torch.tensor([12345,23456])
print(labels)
```

```python
print("logits.shape:",logits.shape)
print("labels.shape:",labels.shape)
print(logits.view(-1,50256).shape,labels.view(-1).shape)

loss = ce(logits.view(-1,50256),labels.view(-1))
print(loss)
```
