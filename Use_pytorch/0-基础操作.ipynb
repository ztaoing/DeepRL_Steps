# tensor维度变换

##  view 和reshape维度变换
- view： 要求张量是连续的（contiguous）。如果张量不是连续的（例如，经过某些操作如 transpose 后），view 会报错。
  - 数据的总元素数量必须保持不变。
  
- reshape： 不要求张量是连续的。即使张量不是连续的，reshape 也可以正常工作。
  - 数据的总元素数量必须保持不变。


```python
# 4张图片，一个通道，28*28像素
a = torch.randn(4,1,28,28)
print(a.shape)
```

```python
# 转换为4,784维度，相当于打平
print(a.reshape(4,784).shape)
```

```python
# 再转换为4,28,28维度
print((a.reshape(4,784)).reshape(4,28,28).shape)
```

```python
# view的用法和reshape的用法完全一样
print(a.view(4,784).shape)
```

```python
# 检查张量是否是连续的
x = torch.randn(2,3)
print(x.is_contiguous())
```

## unsqueeze插入维度


```python
# 2*2的tensor
a = torch.randn(2,3)
print(a.shape)
```

```python
# 在第0维插入一个维度
print(a.unsqueeze(0).shape)
```

```python
# 在倒数第一维插入维度
print(a.unsqueeze(-1).shape)
```

## squeeze删除维度，只能删除为1的维度


```python
a = torch.randn(1,2,3,1)
print(a.shape)
```

```python
# 删除第0维
print(a.squeeze(0).shape)
```

```python
# 删除最后一维
print(a.squeeze(-1).shape)
```

```python
# 删除所有为1的维度
print(a.squeeze().shape)
```

## 维度交换


```python
# t转置，只能操作2维tensor
print(torch.randn(1,2).t().shape)
```

```python
# ranspose维度交换，指定要转换的维度，只能两两交换
print(torch.randn(1,2,3).transpose(0,1).shape)
```

```python
# permute维度交换，输入维度的顺序
print(torch.randn(1,2,3).permute(2,1,0).shape)
```

# broadcat 广播

```python
# broadcast

a = torch.randn(2,3)
b = torch.randn(1,3)
c = torch.randn(1)

print((a+b).shape)
print((a+c).shape)
```

# tensor的拼接和拆分
- torch.cat
- torch.stack
- torch.split
- torch.chunk

```python
a = torch.rand(4,32,8)
b = torch.rand(5,32,8)
```

```python
# cat拼接,dim=0是指定要拼接的维度
torch.cat([a,b],dim=0).shape
```

```python
# stack组合，会创建一个新的维度，用来区分组合后的两个tensor

a = torch.rand(5,32,8)
b = torch.rand(5,32,8)
# a和b 的size必须是相同
torch.stack([a,b],dim=0).shape
```

```python
a = torch.rand(4,32,8)

# split拆分，在0维度上拆分，每2个元素1拆分
_1,_2 = a.split(2,dim=0)
print(_1.shape)
print(_2.shape)
```

```python
_1,_2,_3 = a.split([1,2,1],dim=0)
print(_1.shape)
print(_2.shape)
print(_3.shape)
```

```python
a = torch.rand(7,32,8)

# chunk拆分，在0维度上拆分，拆成2个
_1,_2 = a.chunk(2,dim=0)
print(_1.shape)
print(_2.shape)
```

# 数学运算

```python
#测试数据
a = torch.FloatTensor([[0, 1, 2], [3, 4, 5]])
b = torch.FloatTensor([0, 1, 2])

a, b
```

```python
#四则运算
print(a + b)
print(a - b)
print(a * b)
print(a / b)
```

```python
#矩阵乘法
print(a @ b)
print(a.matmul(b))

#计算过程
0 * 0 + 1 * 1 + 2 * 2, 0 * 3 + 1 * 4 + 2 * 5
```

```python
# 指数、对数运算

# 求指数
print(a**2)

# 开根号
print(a**0.5)
```

```python
# 求e的n次方
print(a.exp())
```

```python
# 以e为底，求对数
print(a.log())
```

```python
# 以2为底，求对数
print(a.log2())
```

```python
# 裁剪，限制数据的上下限制
a.clamp(2,4)
```

```python
#逻辑运算

#大于
print(a > b)

#小于
print(a < b)

#等于
print(a == b)

#不等于
print(a != b)
```
