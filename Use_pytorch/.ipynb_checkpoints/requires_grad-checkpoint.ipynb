在 PyTorch 中，model.parameters() 返回的是模型中所有参数的迭代器。
- 每个参数都有一个 requires_grad 属性，这个属性是一个布尔值，用于指示该参数是否需要计算梯度。

作用
- requires_grad=True：表示该参数在反向传播时会计算梯度，并且可以被优化器更新。这通常用于模型的可训练参数，例如权重和偏置。
- requires_grad=False：表示该参数在反向传播时不会计算梯度，也不会被优化器更新。这通常用于冻结某些层的参数，或者用于不参与训练的参数（例如预训练模型的参数）。

```python
import torch
import torch.nn as nn

# 定义一个简单的线性模型
model = nn.Linear(10, 5)

# 查看模型的参数
for param in model.parameters():
    print(param.requires_grad)  # 默认情况下，所有参数的 requires_grad 都是 True
    
```

```python
# 冻结第一个参数
for param in model.parameters():
    param.requires_grad = False

# 再次查看参数
for param in model.parameters():
    print(param.requires_grad)  # 现在所有参数的 requires_grad 都是 False
```
