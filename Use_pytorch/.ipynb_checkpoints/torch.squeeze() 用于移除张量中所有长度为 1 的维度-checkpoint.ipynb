在 PyTorch 中，torch.squeeze() 函数用于移除张量中所有长度为 1 的维度。这个操作对于减少张量的维度特别有用，尤其是在处理具有单一维度的张量时
- torch.squeeze() 返回一个新的张量，不会修改原始张量。如果需要修改原始张量，可以使用 torch.Tensor.squeeze_() 方法。

```python
from IPython.display import Image
```

```python
Image(filename='./Image/squeeze.png',width=600)
```

```python
import torch

# 创建一个具有单一维度的张量
x = torch.tensor([[[2]]])

# 使用squeeze()函数移除单一维度
y = torch.squeeze(x)

print("Original Tensor:", x)
print("Squeezed Tensor:", y)
```

```python
# 创建一个具有多个单一维度的张量
x = torch.tensor([[[2], [3]]])

# 使用squeeze()函数指定维度移除单一维度
y = torch.squeeze(x, dim=0)

print("Original Tensor:", x)
print("Squeezed Tensor:", y)
```
