```python
import torch
import torch.distributions as dist

# 定义一个分类分布
probs = torch.tensor([0.1, 0.2, 0.3, 0.4])
categorical_dist = dist.Categorical(probs=probs)

# 打印分类分布的参数
print("Categorical Distribution:")
print(f"  Probs: {categorical_dist.probs}")
print(f"  Logits: {categorical_dist.logits}")

# 采样一个动作
action = categorical_dist.sample()
print("采样的动作索引:", action.item())

# 计算给定动作的对数概率
log_prob = categorical_dist.log_prob(action)
print("采样动作的对数概率:", log_prob.item())

# 计算熵
entropy = categorical_dist.entropy()
print("分布的熵:", entropy.item())
```

```python
# 定义一个分类分布，使用对数概率
logits = torch.tensor([-2.0, -1.0, 0.0, 1.0])
categorical_dist = dist.Categorical(logits=logits)

# 采样一个动作
action = categorical_dist.sample()
print("采样的动作索引:", action.item())

# 计算给定动作的对数概率
log_prob = categorical_dist.log_prob(action)
print("采样动作的对数概率:", log_prob.item())

# 计算熵
entropy = categorical_dist.entropy()
print("分布的熵:", entropy.item())
```

# Normal 分布，也称为高斯分布，
- 是一种以均值（mean）和标准差（standard deviation）为参数的对称分布。

以下是如何在 PyTorch 中使用 Normal 分布的一些基本示例：

1. 创建一个正态分布对象
- 你可以使用均值（loc）和标准差（scale）来创建一个正态分布对象。

- 标准差也可以通过方差（variance）来指定。

```python
import torch

# 使用均值和标准差创建正态分布
normal_dist = torch.distributions.Normal(loc=0.0, scale=1.0)
print(normal_dist)
```
