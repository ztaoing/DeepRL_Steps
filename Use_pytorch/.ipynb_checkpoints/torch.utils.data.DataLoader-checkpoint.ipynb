torch.utils.data.DataLoader 是 PyTorch 中用于批量加载数据的工具。它提供了一种高效、灵活的方式来迭代数据集，支持多线程加载、数据打乱、批量处理等功能。通过使用 DataLoader，你可以轻松地将数据集中的数据分批加载到模型中进行训练和测试。

DataLoader 的主要作用是：
- 批量加载数据：将数据集中的数据分批加载，方便模型进行批量训练。
- 支持多线程加载：通过多线程加载数据，提高数据加载的效率，减少训练过程中的等待时间。
- 数据打乱：在每个 epoch 开始时，可以随机打乱数据，避免模型对数据的顺序产生依赖。
- 自定义数据加载方式：通过自定义 Dataset 类，可以灵活地加载各种格式的数据。


DataLoader 的构造函数有多个参数，常用的参数包括：
- dataset：Dataset 对象，表示要加载的数据集。
- batch_size：每个批次的样本数量。默认值为 1。
- shuffle：是否在每个 epoch 开始时随机打乱数据。默认值为 False。
- num_workers：加载数据时使用的子进程数量。默认值为 0，表示在主进程中加载数据。设置为正整数可以启用多线程加载。
- collate_fn：用于将多个样本组合成一个批次的函数。默认情况下，DataLoader 会将多个样本组合成一个批次，但你可以通过自定义 collate_fn 来实现更复杂的组合逻辑。
- drop_last：是否丢弃最后一个不完整的批次。默认值为 False。如果设置为 True，则在数据集大小不能被 batch_size 整除时，最后一个不完整的批次会被丢弃。

```python
import torch
from torch.utils.data import Dataset, DataLoader

class MyDataset(Dataset):
    def __init__(self, data, labels):
        self.data = data
        self.labels = labels

    def __len__(self):
        return len(self.data)

    def __getitem__(self, idx):
        return self.data[idx], self.labels[idx]

# 创建数据集实例
data = torch.randn(100, 10)  # 假设有100个样本，每个样本有10个特征
labels = torch.randint(0, 2, (100,))  # 假设有100个标签，每个标签是0或1
dataset = MyDataset(data, labels)
```

### 创建 DataLoader
- 使用 DataLoader 来批量加载数据：

```python
# 创建 DataLoader
dataloader = DataLoader(dataset, batch_size=10, shuffle=True, num_workers=2)

# 遍历数据集
for batch_idx, (data, labels) in enumerate(dataloader):
    print(f"Batch {batch_idx}")
    print("Data shape:", data.shape)
    print("Labels shape:", labels.shape)
```

### 自定义 collate_fn
- 默认情况下，DataLoader 会将多个样本组合成一个批次。如果你的数据集中的样本是不同类型的对象，或者需要更复杂的组合逻辑，可以通过自定义 collate_fn 来实现。例如：

```python
def my_collate_fn(batch):
    # 自定义组合逻辑
    data = [item[0] for item in batch]
    labels = [item[1] for item in batch]
    return torch.stack(data), torch.tensor(labels)

# 创建 DataLoader
dataloader = DataLoader(dataset, batch_size=10, shuffle=True, collate_fn=my_collate_fn)
```
