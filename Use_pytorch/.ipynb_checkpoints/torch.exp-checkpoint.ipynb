torch.exp 是 PyTorch 中的一个函数，用于计算输入张量中每个元素的指数值。

- 具体来说，它对输入张量的每个元素 x 计算 $e^x$
- 其中 e 是自然对数的底数，约等于 2.71828。

```python
import torch

# 创建一个张量
x = torch.tensor([0.0, 1.0, 2.0, 3.0])

# 计算指数值
exp_x = torch.exp(x)

# 返回值
# 返回值是一个新的张量，其形状与输入张量相同，每个元素是输入张量对应元素的指数值。

print(exp_x)
```

### 注意事项:
- 数值稳定性：当输入值非常大时，torch.exp 的结果可能会非常大，导致数值溢出。为了避免这种情况，可以在计算前对输入值进行适当的缩放或偏移。
- 对数和指数的组合：在某些情况下，你可能需要同时使用 torch.log 和 torch.exp。例如，torch.log(torch.exp(x)) 等价于 x，但直接使用 x 更高效且数值更稳定。

```python
import torch

x = torch.tensor([[0.0, 1.0], [2.0, 3.0]])
exp_x = torch.exp(x)
print(exp_x)
```
