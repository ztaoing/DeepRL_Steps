## 断点续训练完整模板

```python
import torch,os,tqdm
from model import Net        # 模型
from data import get_loader  # 数据

def train(resume=False):
    device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    model = Net().to(device)
    optimizer = torch.optim.Adam(model.parameters(),lr=1e-3)
    start_epoch = 0
    
    # 恢复保存的checkpoint
    if resume and os.path.isfile('checkpoint.pt'):
        ckpt = torch.load('checkpoint.pt',map_location=device)
        model.load_state_dict(ckpt['model'])
        opimizer.load_state_dict(ckpt['optimizer'])
        start_epoch = ckpt['epoch'] + 1
        print(f"Resume from epoch {start_epoch}")
        
    # 加载数据
    loader = get_loader()
    for epoch in tqdm.trange(start_epoch,100):
        for x,y in loader:
            x,y = x.to(device),y.to(device)
            loss = model(x,y)
            optimizer.zero_grad()
            loss.backward()
            optimizer.step()
        # 每个epoch保存一次
        torch.save({
            'model':model.state_dict(),
            'optimizer':optimizer.state_dict(),
            'epoch':epoch,
            'loss':loss.item()
        },'checkpoint.pt')

if __name__ = "__main__":
    train(resume=True) # 第一次调用resume是False
```
