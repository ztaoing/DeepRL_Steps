nn.ModuleList 是 PyTorch 中的一个容器类，用于存储多个 nn.Module 对象。
- 它类似于 Python 中的列表，但专门用于存储 PyTorch 的模块。
- nn.ModuleList 是一个非常有用的工具，可以方便地管理多个子模块，尤其是在构建复杂的神经网络时。

nn.ModuleList 的主要作用是：
- 管理多个子模块：可以将多个 nn.Module 对象存储在一个列表中，方便管理和访问。
- 动态添加和删除模块：可以在运行时动态地添加或删除子模块。
- 自动注册参数：nn.ModuleList 会自动将子模块的参数注册到父模块中，使得这些参数可以被优化器访问和更新。

```python
import torch
import torch.nn as nn

class MyModel(nn.Module):
    def __init__(self, input_size, hidden_size, output_size, num_layers):
        super(MyModel, self).__init__()
        self.input_layer = nn.Linear(input_size, hidden_size)
        # 使用 nn.ModuleList 来存储多个子模块
        self.hidden_layers = nn.ModuleList([nn.Linear(hidden_size, hidden_size) for _ in range(num_layers)])
        self.output_layer = nn.Linear(hidden_size, output_size)

    def forward(self, x):
        x = self.input_layer(x)
        for layer in self.hidden_layers:
            x = torch.relu(layer(x))
        x = self.output_layer(x)
        return x

# 创建模型实例
model = MyModel(input_size=10, hidden_size=5, output_size=2, num_layers=3)

# 查看模型的参数
for param in model.parameters():
    print(param)

# 前向传播
inputs = torch.randn(3, 10)  # 假设有3个样本，每个样本有10个特征
outputs = model(inputs)
print(outputs)
```

### 动态添加和删除子模块
- nn.ModuleList 支持在运行时动态添加和删除子模块。例如：

```python
# 动态添加一个子模块
model.hidden_layers.append(nn.Linear(5, 5))

# 动态删除一个子模块
del model.hidden_layers[-1]
```

#### 不要使用普通的 Python 列表来存储子模块，因为普通列表不会自动注册子模块的参数。例如：

```python
self.hidden_layers = [nn.Linear(hidden_size, hidden_size) for _ in range(num_layers)]  # 错误
```

#### 正确的做法是使用 nn.ModuleList：

```python
self.hidden_layers = nn.ModuleList([nn.Linear(hidden_size, hidden_size) for _ in range(num_layers)])
```

### 参数更新：
- 由于 nn.ModuleList 会自动注册子模块的参数，因此这些参数可以被优化器访问和更新。例如：

```python
optimizer = torch.optim.SGD(model.parameters(), lr=0.01)
```
