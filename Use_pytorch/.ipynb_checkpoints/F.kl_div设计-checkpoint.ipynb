## PyTorch的kl_div函数设计

```python
import torch
import torch.nn.functional as F

def explain_kl_div():
    """
    解释PyTorch的kl_div为什么这样设计
    """
    # 假设我们有两个分布
    teacher_probs = torch.tensor([0.7, 0.2, 0.1])  # 教师分布 P
    student_log_probs = torch.log(torch.tensor([0.6, 0.3, 0.1]))  # 学生log分布 log(Q)
    
    print("教师分布 P:", teacher_probs)
    print("学生log分布 log(Q):", student_log_probs)
    
    # 手动计算正向KL散度 D_KL(P||Q)= sum(P * log(P/Q)) = sum(P * (logP - logQ))
    manual_kl = (teacher_probs * (torch.log(teacher_probs) - student_log_probs)).sum()
    
    # 使用F.kl_div
    pytorch_kl = F.kl_div(student_log_probs, teacher_probs, reduction='sum')
    
    print(f"\n手动计算KL: {manual_kl.item():.4f}")
    print(f"PyTorch kl_div: {pytorch_kl.item():.4f}")
    
    return manual_kl, pytorch_kl

explain_kl_div()
```
