- nn.Module 的 state_dict() 方法，用于获取模型的状态字典（包含所有可学习参数和缓冲区）

| 参数            | 类型              | 默认值     | 说明                             |
| ------------- | --------------- | ------- | ------------------------------ |
| `destination` | `dict` / `None` | `None`  | 目标字典，用于存储状态。若为 `None`，则创建新字典   |
| `prefix`      | `str`           | `''`    | 键名前缀，用于嵌套模块时区分不同层              |
| `keep_vars`   | `bool`          | `False` | 是否保留 `Variable` 包装（旧版），现为兼容性保留 |


```python
import torch
import torch.nn as nn

class MyModel(nn.Module):
    def __init__(self):
        super().__init__()
        self.fc1 = nn.Linear(10,5)
        self.fc2 = nn.Linear(5,2)
    def forward(self,x):
        return self.fc2(torch.relu(self.fc1(x)))
model = MyModel()
# 基本用法：获取完整状态字典
state = model.state_dict()
print(state.keys)

# 使用参数：指定destination 和prefix
dest = {}
model.state_dict(destination=dest,prefix="model.")
print(dest.keys())

# 常用于保存检查点：
torch.save(model.state_dict(),"model.pth")
```

```python
# 加载
model.load_state_dict(torch.load('model.pth'))
```
