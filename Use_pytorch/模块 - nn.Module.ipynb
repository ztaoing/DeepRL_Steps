nn.Module 是 PyTorch 中的一个核心类，它是所有神经网络模块的基类。

- 如果你想要定义自己的神经网络模型，通常需要继承 nn.Module 并实现其方法。
- 通过继承 nn.Module，你可以方便地构建复杂的神经网络结构，并且能够利用 PyTorch 提供的许多工具和功能来训练和优化你的模型。

作用
nn.Module 提供了以下功能：
- 参数管理：自动管理模型的参数（如权重和偏置），包括参数的初始化、更新和保存。
  - 你可以通过 model.parameters() 来访问模型的所有参数。
- 子模块管理：可以将多个模块组合成一个更大的模块，方便构建复杂的网络结构。
- 前向传播：通过实现 forward 方法定义模型的前向传播逻辑。
- 自动梯度计算：在反向传播时，自动计算梯度并更新参数。

```python
import torch
import torch.nn as nn

# 定义一个简单的线性模型
class MyLinearModel(nn.Module):
    def __init__(self, input_size, output_size):
        # 调用 super().__init__() 来初始化父类。
        super(MyLinearModel, self).__init__()
        self.linear = nn.Linear(input_size, output_size)

    def forward(self, x):
        return self.linear(x)

# 创建模型实例
model = MyLinearModel(10, 5)

# 查看模型的参数
for param in model.parameters():
    print(param)

# 前向传播
inputs = torch.randn(3, 10)  # 假设输入是3个样本，每个样本有10个特征
outputs = model(inputs)
print(outputs)
```

### 默认情况下，PyTorch 会自动初始化模型的参数，但你也可以手动初始化参数。例如：

```python
def init_weights(m):
    if isinstance(m, nn.Linear):
        torch.nn.init.xavier_uniform_(m.weight)
        m.bias.data.fill_(0.01)
model.apply(init_weights)
```
