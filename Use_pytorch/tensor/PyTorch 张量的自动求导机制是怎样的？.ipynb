# PyTorch 张量的自动求导机制是怎样的？

- 这是深度学习框架的“大脑”。简单来说，<font color='red'>PyTorch 会在你进行运算时，动态地记录所有操作，构建一个计算图，然后通过链式法则自动计算梯度。</font>

### 核心组件
- requires_grad：一个开关。如果设置为 True，PyTorch 就会开始追踪该张量的所有历史操作。
- grad_fn：记录了生成该张量的函数（比如加法、乘法）。叶子节点（用户创建的）通常是 None。
- .backward()：反向传播的触发器。它会自动计算所有梯度。
- .grad：存放计算出来的梯度值。

### 工作流程示例

假设我们要计算函数 

$y=x^2$在x=3 处的导数（理论值是 2x=6 ）。

```python
import torch

# 1. 创建张量并开启追踪
x = torch.tensor(3.0,requires_grad=True)

# 2. 进行运行（构建计算图）
y = x ** 2 # y = x^2
# 此时y有一个属性y.grad_fn,指向<PowBackward>
print(y.grad_fn)

# 3. 反向传播（自动求导）
y.backward() # 相当于 对于y求关于x的导数

# 4. 结果
print(x.grad)
```

###### 总结：你只需要负责写前向传播（怎么算结果），调用 .backward() 后，PyTorch 会自动帮你算出反向传播（梯度）。
