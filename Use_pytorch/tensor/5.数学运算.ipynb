数学运算 (Math Operations)
PyTorch 的运算分为元素级、矩阵级和归约操作。
- 元素级运算
基础运算：+, -, *, /, // (整除), % (取模), ** (幂运算)。
- 函数运算：
    - torch.abs(tensor): 绝对值。
    - torch.sqrt(tensor): 平方根。
    - torch.exp(tensor): 指数运算 
    - torch.log(tensor): 对数运算。
    - torch.clamp(tensor, min, max): 将数值限制在指定范围内（常用于截断）。
- 矩阵运算
    - torch.matmul(A, B) 或 A @ B: 矩阵乘法。
    - torch.mm(A, B): 二维矩阵乘法。
    - torch.bmm(A, B): 批量矩阵乘法（Batch Matrix Multiply），用于三维张量。
    - torch.dot(v1, v2): 向量点积。
- 归约操作
    - tensor.sum(dim): 求和。
    - tensor.mean(dim): 求均值。
    - tensor.max(dim): 求最大值（返回值和索引）。
    - tensor.argmax(dim): 求最大值的索引。
    - tensor.min(dim), tensor.argmin(dim): 最小值相关。
    - torch.sort(tensor, dim): 排序。
- 比较操作
    - torch.eq(a, b): 相等比较。
    - torch.gt(a, b): 大于比较。
    - torch.any(tensor): 只要有一个为真则返回真。
    - torch.all(tensor): 全部为真才返回真。

# 🧮 元素级运算 (Element-wise Operations)
元素级运算是指对张量中的每一个元素独立地执行相同的数学操作。
## 基础算术运算
这些运算符和 NumPy 的用法几乎完全一致，非常直观。
- +, -, *, /: 分别对应加、减、乘、除。
- //: 整除，返回商的整数部分。
- %: 取模，返回余数。
- **: 幂运算，例如 x ** 2 计算每个元素的平方。
## 常用函数运算

PyTorch 提供了丰富的数学函数，同样以元素级方式工作。

- torch.abs(tensor): 计算绝对值。
- torch.sqrt(tensor): 计算平方根。
- torch.exp(tensor): 计算自然指数 (e^x)。
- torch.log(tensor): 计算自然对数 (ln x)。
- torch.clamp(tensor, min, max): 将张量中的所有值截断到 [min, max] 范围内，小于 min 的变为 min，大于 max 的变为 max。

```python
import torch
x = torch.tensor([-2.0, -1.0, 0.0, 1.0, 2.0])

print(torch.clamp(x, min=0, max=1)) 
# 输出: tensor([0., 0., 0., 1., 1.])，将所有值限制在 [0, 1] 区间
```

# 矩阵运算 (Matrix Operations)

矩阵运算是线性代数的核心，在神经网络中用于计算层与层之间的变换。

- torch.matmul(A, B) 或 A @ B: <font color='red'>最通用</font>的矩阵乘法。<font color='red'>它支持多种情况，包括向量-矩阵乘法、标准的二维矩阵乘法，以及高维张量的批量矩阵乘法（会自动处理广播）。</font>
- torch.mm(A, B): <font color='red'>严格用于二维张量（矩阵）的乘法</font>。它是 matmul 在二维情况下的一个高效特化版本。
- torch.bmm(A, B): <font color='red'>批量矩阵乘法，专门用于三维张量</font>。它会对 batch 维度中的每一对矩阵分别执行 mm 操作。输入形状通常为 (batch_size, n, m) 和 (batch_size, m, p)。
- torch.dot(v1, v2): 计算两个一维张量（向量）的<font color='red'>点积</font>，即对应元素相乘再求和。

```python
# 二维矩阵乘法
A = torch.randn(3, 4)
B = torch.randn(4, 2)
C = torch.mm(A, B)       # 或者 C = A @ B
print(C.shape)           # torch.Size([3, 2])

# 批量矩阵乘法
A_batch = torch.randn(10, 3, 4) # 10个 3x4 的矩阵
B_batch = torch.randn(10, 4, 2) # 10个 4x2 的矩阵
C_batch = torch.bmm(A_batch, B_batch)
print(C_batch.shape)     # torch.Size([10, 3, 2])
```

# 📉 归约操作 (Reduction Operations)

<font color='red'>归约操作会沿着指定的维度对张量进行汇总，从而降低张量的维度。</font>

- tensor.sum(dim): 沿 dim 维度求和。
- tensor.mean(dim): 沿 dim 维度求平均值。
- tensor.max(dim): 沿 dim 维度求最大值。它返回一个命名元组，包含最大值本身 (values) 和最大值的索引 (indices)。
- tensor.argmax(dim): 直接返回沿 dim 维度最大值的索引。
- tensor.min(dim), tensor.argmin(dim): 与 max 和 argmax 类似，但用于查找最小值。
- torch.sort(tensor, dim): 沿 dim 维度对张量进行排序，同样返回一个包含排序后值 (values) 和对应索引 (indices) 的命名元组。

```python
x = torch.tensor([[1, 4, 2], [3, 0, 5]])

# 沿第1维（列）求最大值
max_vals, max_indices = x.max(dim=1)
print(max_vals)    # tensor([4, 5])
print(max_indices) # tensor([1, 2])
```

## tensor.argmax(dim) 
- 是 PyTorch 中用于<font color='red'>查找最大值位置索引的函数</font>。
- 它的核心逻辑是：<font color='red'>沿着你指定的 dim 维度进行“挤压”或“消除”，并在该维度上寻找最大值的索引。</font>

### 一维张量 (找全局最大)
- 当不指定 dim 或者张量只有一维时，它就是在找整个数组里最大的那个数的下标。

###### 解释：最大值是 50，它在索引 1 的位置。

```python
import torch
x = torch.tensor([10, 50, 30, 20])
# 索引:    0   1   2   3

result = x.argmax() 
# 或者 x.argmax(dim=0)

print(result) # 输出: tensor(1)
```

### 二维张量 (<font color='red'>最常用的场景</font>)
- 这是最容易混淆的地方。请记住口诀：“<font color='red'>指定哪个维度，就消除哪个维度”</font>。

假设我们有一个 2行3列 的张量 x (形状 [2, 3])：

```python
x = torch.tensor([
    [1, 5, 3],  # 第0行
    [6, 2, 4]   # 第1行
])
```

#### 情况 A：dim=0 (跨行比较，消除行)
- 含义：沿着第 0 维（纵向/行）看。也就是列与列比。
- 操作：
    - 第0列：1 和 6 比 -> 6 大 (索引是 1，因为在第1行)
    - 第1列：5 和 2 比 -> 5 大 (索引是 0，因为在第0行)
    - 第2列：3 和 4 比 -> 4 大 (索引是 1，因为在第1行)
- 结果形状：[3] (原来的行数维度被消除了)

```python
print(x.argmax(dim=0)) 
# 输出: tensor([1, 0, 1])
```

#### 情况 B：dim=1 (跨列比较，消除列)
- 含义：沿着第 1 维（横向/列）看。也就是行内自己比（常用于找每张图片的预测类别）。
- 操作：
    - 第0行：[1, 5, 3] -> 5 最大 (索引是 1)
    - 第1行：[6, 2, 4] -> 6 最大 (索引是 0)
- 结果形状：[2] (原来的列数维度被消除了)

```python
print(x.argmax(dim=1)) 
# 输出: tensor([1, 0])
```

### 三维张量 (进阶理解)

假设张量形状是 [2, 3, 4]。

- dim=0：比较两个“大块”数据。结果形状变为 [3, 4]。
逻辑：位置 [0,0,0] 的数 和 位置 [1,0,0] 的数比，谁大取谁的索引（0或1）。
- dim=1：在每一块内部比较“行”。结果形状变为 [2, 4]。
- dim=2：在每一行内部比较“列”（具体数值）。结果形状变为 [2, 3]。

### 实战案例：<font color='red'>计算分类准确率</font>
- 在深度学习分类任务中，模型输出通常是 [Batch_Size, Num_Classes] 的概率分布。<font color='red'>我们需要找出每个样本概率最大的那个类别的索引：</font>

```python
# 假设有 4 个样本，分为 3 类
outputs = torch.tensor([
    [0.1, 0.8, 0.1],   # 样本0：预测第1类概率最大
    [0.3, 0.3, 0.4],   # 样本1：预测第2类概率最大
    [0.9, 0.05, 0.05], # 样本2：预测第0类概率最大
    [0.2, 0.2, 0.6]    # 样本3：预测第2类概率最大
])

# 我们关心每一行（每个样本）的最大值索引
predicted_classes = outputs.argmax(dim=1)

print(predicted_classes) 
# 输出: tensor([1, 2, 0, 2])
```

# ⚖️ 比较操作 (Comparison Operations)

比较操作用于比较张量元素之间的关系，返回一个布尔类型的张量。

- torch.eq(a, b): 逐元素比较是否相等 (a == b)。
- torch.gt(a, b): 逐元素比较是否大于 (a > b)。
- torch.any(tensor): 检查张量中是否至少有一个元素为 True。
- torch.all(tensor): 检查张量中是否所有元素都为 True。

```python
a = torch.tensor([1, 2, 3])
b = torch.tensor([1, 2, 4])

is_equal = torch.eq(a, b)
print(is_equal)      # tensor([ True,  True, False])

print(torch.any(is_equal)) # True，因为有 True 的元素
print(torch.all(is_equal)) # False，因为并非所有元素都为 True
```
