# PyTorch 张量和 NumPy 数组怎么互转？
- PyTorch 的设计初衷就是为了让深度学习像 NumPy 一样易用，因此它们之间的转换非常高效，且<font color='red'>默认情况下共享内存（即修改一方，另一方也会改变）。</font>

## Torch Tensor 转 NumPy Array

```python
import torch
import numpy as np

# 创建一个tensor
t = torch.ones(3)
print(f"tensor:{t}")

# 转换为numpy
n = t.numpy()
print(f"numpy:{n}")

# 验证共享内存
t[0] = 99
print(f"修改tensor后，numpy变为:{n}") # numpy数组也被修改了
```

###### 注意：如果 Tensor 在 GPU 上，必须先用 .cpu() 移回 CPU 才能转换。

## NumPy Array 转 Torch Tensor
- 使用 torch.from_numpy() 方法（推荐）或 torch.tensor()。

```python
# 创建一个numpy数组
n = np.ones(3)

# torch.from_numpy(共享内存)
t1 = torch.from_numpy(n)

# torch.tensor(复制数据，不共享内存)
t2 = torch.tensor(n)

# 修改
n[0]= 55
print(t1) # 变为55（共享）
print(t2) # 仍然是1（不共享）
```
