| 类别 | 常用操作/函数 | 备注 |
| :--- | :--- | :--- |
| 创建 | `torch.tensor`, `zeros`, `ones`, `randn`, `from_numpy` | 支持指定 `dtype` 和 `device` |
| 变形 | `view`, `reshape`, `permute`, `squeeze`, `unsqueeze` | `view` 要求内存连续 |
| 数学 | `+`, `*`, `@`, `sum`, `mean`, `max`, `clamp` | 支持广播机制 |
| 索引 | `[:]`, `[mask]`, `gather`, `index_select` | 类似 NumPy |
| 原地 | `add_`, `mul_`, `zero_`, `copy_` | 节省显存，以 `_` 结尾 |
| 设备 | `.to('cuda')`, `.cpu()` | 训练前必须将数据移至 GPU |
