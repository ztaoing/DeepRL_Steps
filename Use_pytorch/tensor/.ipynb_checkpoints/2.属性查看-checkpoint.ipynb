# 属性查看 (Attributes)

了解张量的“长相”和数据类型。

- tensor.shape 或 tensor.size(): 查看张量的形状（如 [2, 3]）。
- tensor.dtype: 查看数据类型（如 torch.float32, torch.int64）。
- tensor.device: 查看张量所在的设备（cpu 或 cuda）。
- tensor.ndim: 查看维度数量。
- tensor.item(): 仅适用于单元素张量，将其转换为 Python 数字。


| 属性/方法 | 返回值类型 | 含义 | 常用场景 |
| :--- | :--- | :--- | :--- |
| `.shape` | `torch.Size` | 张量的具体形状 | 检查数据维度是否匹配 |
| `.ndim` | `int` | 维度的个数 | 判断是向量还是矩阵 |
| `.dtype` | `torch.dtype` | 数据类型 | 确保计算精度一致 |
| `.device` | `torch.device` | 设备位置 | 确保数据在 GPU 上 |
| `.item()` | `float`/`int` | Python 原生数值 | 打印 Loss、准确率等标量 |

## 形状与维度：shape / size() / ndim

这三个属性帮你搞清楚数据是“几维的”以及“每一维有多大”。

### tensor.shape 与 tensor.size()

<font color='red'>这两个是完全等价的，都返回一个 torch.Size 对象（本质上是一个元组）。</font>

- tensor.shape：这是<font color='red'>属性</font>访问方式，更常用，写法更像 Python 原生风格。
- tensor.size()：这是<font color='red'>方法</font>调用方式，为了兼容旧版本或习惯。

用途：检查矩阵的行数、列数，或者图像的 [Batch, Channel, Height, Width]

```python
import torch

x = torch.randn(2, 3, 4) # 创建一个 3维张量

print(x.shape)      # 输出: torch.Size([2, 3, 4])
print(x.size())     # 输出: torch.Size([2, 3, 4])

# 获取特定维度的大小（例如获取第 1 维的大小，即 3）
print(x.shape[1])   # 输出: 3
```

### tensor.ndim
返回张量的维度数量（即 rank）。
- 标量是 0 维。
- 向量是 1 维。
- 矩阵是 2 维。

```python
print(x.ndim)       # 输出: 3 (因为它有 2, 3, 4 三个维度)
```

## 数据类型与设备：dtype / device

这两个属性决定了张量在计算机底层是如何存储和计算的。

### tensor.dtype

查看张量中元素的数据类型。

- 常见类型：torch.float32 (默认浮点数), torch.int64 (默认整数), torch.float16 (半精度)。
- 重要性：如果数据类型不匹配（例如矩阵乘法时一个是 float32 一个是 float16），程序会报错。

```python
y = torch.tensor([1, 2, 3])
print(y.dtype)      # 输出: torch.int64 (默认整数是 int64)

z = torch.tensor([1.0, 2.0])
print(z.dtype)      # 输出: torch.float32 (默认浮点是 float32)
```

### tensor.device

查看张量存储在 CPU 还是 GPU (CUDA) 上。

- 重要性：PyTorch 默认在 CPU 上创建张量。如果你想用 GPU 加速，必须将张量移动到 GPU。计算时，参与运算的所有张量必须在同一个设备上。

```python
print(y.device)     # 输出: cpu

# 如果移动到 GPU (假设有显卡)
# y_gpu = y.to('cuda')
# print(y_gpu.device) # 输出: cuda:0
```

## 提取数值：item()

这是一个非常实用的“转换”方法。

### tensor.item()
- 作用：从张量中提取唯一的那个数字，并将其转换为 Python 原生的数据类型（如 int 或 float）。
- 限制：只能用于单元素张量（即标量，或者形状为 [1] 的张量）。如果张量里有多个元素，调用它会报错。
- 场景：常用于打印训练时的 Loss 值（例如 print(loss.item())），因为 Loss 通常是一个标量

```python
# 创建一个单元素张量
loss = torch.tensor(3.14159)

# 提取为 Python 浮点数
val = loss.item()
print(val)          # 输出: 3.14159
print(type(val))    # 输出: <class 'float'>

# 错误示范：多元素张量不能用 item()
# x = torch.tensor([1, 2])
# x.item()  <-- 会报错 RuntimeError
```
