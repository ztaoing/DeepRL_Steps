面给出一个“看得见数字”的小例子，一行行打印出 logits.log_softmax(-1) 前后的变化，以及怎样取出指定 token 的 log-prob。

```python
import  torch
logits = torch.tensor([2.5,1.0,0.2,3.0])
print("原始logits:",logits)

# 等价于 F.log_softmax(-1)
log_probs = logits.log_softmax(-1)
print("log-probs:",log_probs)

# 验证：log_probs的和为0，概率和为1
print('log_probs之和：',log_probs.exp().sum())

# 模型在这一步采样处的token_id = 3
token_id = 3
print(f"token={token_id}的对数概率 = ",log_probs[token_id].item())
```
