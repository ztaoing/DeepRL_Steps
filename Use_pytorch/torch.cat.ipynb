```python
import torch

# 创建两个张量
a = torch.tensor([[1, 2], [3, 4]])
b = torch.tensor([[5, 6], [7, 8]])

# 沿着第 0 维拼接（垂直拼接）
result_0 = torch.cat((a, b), dim=0)
print("沿着第 0 维拼接：")
print(result_0)
# 输出：
# tensor([[1, 2],
#         [3, 4],
#         [5, 6],
#         [7, 8]])

# 沿着第 1 维拼接（水平拼接）
result_1 = torch.cat((a, b), dim=1)
print("沿着第 1 维拼接：")
print(result_1)
# 输出：
# tensor([[1, 2, 5, 6],
#         [3, 4, 7, 8]])
```
