### nn.DataParallel 是 PyTorch 中用于在多 GPU 环境下加速模型训练的一个工具。
- 它通过将输入数据分割成多个部分，分别在不同的 GPU 上进行计算，从而实现并行化训练。
- 每个 GPU 上的模型副本会独立地处理自己的数据分片，并在反向传播时将梯度汇总，从而更新模型的参数。

作用
nn.DataParallel 的主要作用是：
- 加速训练：通过在多个 GPU 上并行处理数据，显著提高训练速度。
- 简化代码：它允许你使用单 GPU 的代码结构，而不需要对代码进行大量修改即可实现多 GPU 训练。

```python
import torch
import torch.nn as nn

# 定义一个简单的模型
model = nn.Linear(10, 5)

# 检查是否有可用的 GPU
if torch.cuda.device_count() > 1:
    print("Let's use", torch.cuda.device_count(), "GPUs!")
    # 包装模型以使用 DataParallel
    model = nn.DataParallel(model)

# 将模型移动到 GPU 或 CPU上
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model.to(device)

# 定义损失函数和优化器
criterion = nn.MSELoss()
optimizer = torch.optim.SGD(model.parameters(), lr=0.01)

# 假设有一些输入数据和目标数据
inputs = torch.randn(20, 10).to(device)
targets = torch.randn(20, 5).to(device)

# 前向传播
outputs = model(inputs)
loss = criterion(outputs, targets)

# 反向传播和优化
loss.backward()
optimizer.step()
```

```python
# 保存模型
torch.save(model.state_dict(), 'model.pth')

# 加载模型
model = nn.Linear(10, 5)
model.load_state_dict(torch.load('model.pth'))

# 如果在单 GPU 环境下加载多 GPU 训练的模型
model = nn.Linear(10, 5)

model.load_state_dict(torch.load('model.pth', map_location=torch.device('cpu')))
```
