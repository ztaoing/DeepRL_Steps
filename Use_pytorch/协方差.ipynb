协方差（Covariance） 是一个统计学概念，用于衡量两个随机变量之间的线性关系。具体来说，协方差描述了两个变量如何一起变化。
- 如果两个变量的变化趋势相同（即一个增加时另一个也增加），协方差为正；
- 如果变化趋势相反（即一个增加时另一个减少），协方差为负；
- 如果两个变量之间没有线性关系，协方差接近于零。

```python
import numpy as np

# 示例数据
X = np.array([1, 2, 3, 4, 5])
Y = np.array([2, 3, 4, 5, 6])

# 计算协方差
cov_matrix = np.cov(X, Y)
print("协方差矩阵：")
print(cov_matrix)

# 提取协方差
cov_XY = cov_matrix[0, 1]
print("X 和 Y 的协方差：", cov_XY)
```
