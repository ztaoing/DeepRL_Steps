nn.Sequential 是 PyTorch 中的一个非常方便的容器类，用于按顺序组合多个 nn.Module 对象。
- 它允许你以一种简洁的方式构建神经网络，而无需手动编写 forward 方法。
- nn.Sequential 会自动将输入数据依次传递给每个子模块，并返回最终的输出。

nn.Sequential 的主要作用是：
- 简化代码：通过将多个模块按顺序组合，可以简化代码，减少重复的 forward 方法实现。
- 自动管理模块：nn.Sequential 会自动管理子模块的参数，使得这些参数可以被优化器访问和更新。
- 方便调试：在调试时，可以轻松地插入新的模块或移除现有模块。

```python
import torch
import torch.nn as nn

# 定义一个简单的神经网络
model = nn.Sequential(
    nn.Linear(10, 5),  # 输入层到隐藏层
    nn.ReLU(),         # 激活函数
    nn.Linear(5, 2)    # 隐藏层到输出层
)

# 查看模型的结构
print(model)

# 前向传播
inputs = torch.randn(3, 10)  # 假设有3个样本，每个样本有10个特征
outputs = model(inputs)
print(outputs)
```

### 自动前向传播：
- nn.Sequential 会自动将输入数据依次传递给每个子模块，并返回最终的输出。你不需要手动实现 forward 方法。

### 访问子模块：
你可以通过索引或名称访问 nn.Sequential 中的子模块。例如

```python
print(model[0])  # 访问第一个模块
print(model[1])  # 访问第二个模块
```

### 动态修改：
你可以动态地添加或删除子模块。例如：

```python
model.append(nn.Sigmoid())  # 添加一个新的模块
model.pop()                 # 删除最后一个模块
```

```python
class MyModel(nn.Module):
    def __init__(self):
        super(MyModel, self).__init__()
        self.sequential = nn.Sequential(
            nn.Linear(10, 5),
            nn.ReLU(),
            nn.Linear(5, 2)
        )
        self.additional_layer = nn.Sigmoid()

    def forward(self, x):
        x = self.sequential(x)
        x = self.additional_layer(x)
        return x

# 创建模型实例
model = MyModel()

# 前向传播
inputs = torch.randn(3, 10)  # 假设有3个样本，每个样本有10个特征
outputs = model(inputs)
print(outputs)
```
