## Dataset.from_list 是 🤗 Datasets 库
- （注意不是 torch.utils.data.Dataset）的快捷函数：
- 把一段“普通 Python 列表”直接转成 datasets.Dataset 对象，省去先写 features、再 map、再 cast 的繁琐步骤。
- 写脚本、做 RLHF/指令微调时随手就能把本地 JSON 样例灌进去，不用再手动拼 from_dict + Features

```python
from datasets import Dataset

data = [
    {"prompt": "What is 2+2?", "response": "4"},
    {"prompt": "Capital of France?", "response": "Paris"}
]

ds = Dataset.from_list(data)   # 一步完成
print(ds)
m
```
