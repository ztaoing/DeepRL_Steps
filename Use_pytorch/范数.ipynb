### L1 范数和 L2 范数
L1 范数和 L2 范数是数学中
- 用于衡量向量或矩阵大小的两种常见范数。
- 它们在机器学习、信号处理、图像处理等领域中有着广泛的应用。

```python
from IPython.display import Image
```

```python
Image(filename='./Image/fanshu_l1.png',width=600)
```

应用场景：
- 机器学习中的正则化：L1 正则化（Lasso 回归）可以用于特征选择，通过将一些特征的权重压缩为零，从而实现特征的稀疏表示。
- 信号处理：在信号恢复和压缩感知中，L1 范数可以用于稀疏信号的重建。

```python
Image(filename='./Image/fanshu_l2.png',width=600)
```


应用场景：
- 机器学习中的正则化：L2 正则化（Ridge 回归）可以用于防止模型过拟合，通过限制权重的大小来实现。
- 图像处理：在图像处理中，L2 范数可以用于计算图像的相似度，例如通过计算两个图像的像素差的 L2 范数来衡量它们的相似度。
- 信号处理：在信号处理中，L2 范数可以用于计算信号的能量。

```python
Image(filename='./Image/l1l2.png',width=600)
```

```python

import torch

# 创建一个张量
tensor = torch.tensor([1.0, -2.0, 3.0, -4.0])

# 计算 L1 范数
l1_norm = torch.norm(tensor, p=1)
print("L1 Norm:", l1_norm)  # 输出：10.0
```

```python
import torch

# 创建一个张量
tensor = torch.tensor([1.0, -2.0, 3.0, -4.0])

# 计算 L2 范数
l2_norm = torch.norm(tensor, p=2)
print("L2 Norm:", l2_norm)  # 输出：5.477225575051661
```
