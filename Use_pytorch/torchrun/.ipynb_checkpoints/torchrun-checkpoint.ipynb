# torchrun
torchrun 是 PyTorch 1.9.0+ 引入的分布式训练启动工具，取代了旧的 torch.distributed.launch。它提供了更简洁的命令行界面和更强大的功能，用于启动分布式训练任务。



torchrun 的执行逻辑是一个高度自动化的流程，它将分布式训练中繁琐的初始化和协调工作封装了起来。其核心执行步骤可以概括为以下五个阶段：

## 📝  解析命令行参数
当你执行 torchrun 命令时，它首先会解析你提供的所有参数，例如 --nproc_per_node（每个节点的进程数）、--nnodes（总节点数）、--rdzv_endpoint（集合点地址）等。这些参数是后续所有操作的配置基础。

## ⚙️  设置环境变量
在启动任何进程之前，torchrun 会根据解析的参数，为即将启动的每个进程准备好必需的环境变量。这些变量是分布式进程间通信的“胶水”，主要包括：

    - RANK: 进程的全局唯一编号。
    - LOCAL_RANK: 进程在当前节点内的编号，通常用于绑定对应的 GPU。
    - WORLD_SIZE: 参与训练的总进程数。
    - MASTER_ADDR 和 MASTER_PORT: 主节点的地址和端口，用于初始化通信。
## 🤝  协调进程集合 (Rendezvous)

这是多节点训练中至关重要的一步。torchrun 会使用一个“集合点”（Rendezvous）机制来确保所有节点上的所有进程都准备就绪，并能够相互发现。

    - 所有进程都会尝试连接到由 --rdzv_endpoint 指定的地址。
    - torchrun 会等待，直到所有预期数量的进程都成功连接，形成一个完整的进程组。这个过程保证了分布式训练的同步启动。
## 🚀  启动多进程

一旦所有进程完成集合，torchrun 就会在本地节点上启动指定数量（--nproc_per_node）的 Python 进程。每个进程都会独立运行你提供的训练脚本（例如 train.py）。
## 💻  执行用户脚本

每个被启动的进程会独立执行你的训练脚本。此时，你的脚本可以通过 torch.distributed 包读取之前设置好的环境变量（如 RANK），并调用 dist.init_process_group() 来正式加入分布式训练组，开始进行数据并行或模型并行等训练任务。



| 特性         | 说明                            |
| ---------- | ----------------------------- |
| **弹性训练**   | 支持动态扩缩容节点                     |
| **容错机制**   | 自动重启失败的 worker                |
| **简化配置**   | 无需手动指定 `--nproc_per_node` 等参数 |
| **环境变量传递** | 自动处理 RANK、WORLD\_SIZE 等       |


## 单机多卡训练（最常用）

```python
# 使用 4 张 GPU 训练
torchrun --nproc_per_node=4 train.py

# 完整参数示例
torchrun \
    --nnodes=1 \                    # 节点数
    --nproc_per_node=4 \            # 每个节点的进程数（GPU数）
    --master_addr="localhost" \     # 主节点地址
    --master_port=29500 \           # 通信端口
    train.py \
    --batch_size 32 \
    --epochs 100
```

## 多级多卡训练

```python
# 在 Node 0 (主节点，IP: 192.168.1.10) 上运行：
torchrun \
    --nnodes=2 \
    --nproc_per_node=8 \
    --node_rank=0 \
    --master_addr="192.168.1.10" \
    --master_port=29500 \
    train.py

# 在 Node 1 上运行：
torchrun \
    --nnodes=2 \
    --nproc_per_node=8 \
    --node_rank=1 \
    --master_addr="192.168.1.10" \  # 主节点地址
    --master_port=29500 \
    train.py
```

## 弹性训练

```python
# 允许 1-4 个节点动态加入/离开
torchrun \
    --nnodes=1:4 \          # 最小1个节点，最大4个节点
    --nproc_per_node=8 \
    --max_restarts=3 \      # 最大重启次数
    --rdzv_id=job123 \      # 作业唯一标识
    --rdzv_backend=c10d \   # 协调后端
    --rdzv_endpoint="192.168.1.10:29500" \
    train.py
```

### 关键环境变量

torchrun 会自动设置以下环境变量，你可以在代码中使用：

```python
import os

os.environ["RANK"]          # 全局进程编号 (0 ~ world_size-1)
os.environ["LOCAL_RANK"]    # 本地 GPU 编号 (0 ~ nproc_per_node-1)
os.environ["WORLD_SIZE"]    # 总进程数
os.environ["MASTER_ADDR"]   # 主节点地址
os.environ["MASTER_PORT"]   # 通信端口
```
